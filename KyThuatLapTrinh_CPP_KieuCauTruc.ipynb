# 1. KHÁI NIỆM KIỂU CÓ CẤU TRÚC
---

## 1.1 Vấn Đề Cần Giải Quyết

Giả sử bạn cần lưu thông tin một sinh viên gồm: **họ tên, mã số, điểm toán, điểm lý, điểm hóa**.

**Cách cũ — dùng biến riêng lẻ:**
```cpp
string hoTen1;  int maSo1;  float toan1, ly1, hoa1;
string hoTen2;  int maSo2;  float toan2, ly2, hoa2;
// ... 30 sinh viên = 150 biến! → Không thể quản lý!
```

**Cách cũ — dùng mảng riêng lẻ:**
```cpp
string hoTen[30];  int maSo[30];  float toan[30], ly[30], hoa[30];
// Dữ liệu của 1 người bị xé lẻ ở 5 mảng khác nhau!
// Khi sắp xếp phải đồng bộ 5 mảng cùng lúc → dễ lỗi
```

## 1.2 Giải Pháp: Kiểu Có Cấu Trúc (`struct`)

**`struct`** cho phép **gom nhóm nhiều biến thuộc các kiểu khác nhau** thành một **đơn vị dữ liệu duy nhất** với tên gọi riêng.

```cpp
struct SinhVien {
    string hoTen;     // Trường (field/member)
    int    maSo;
    float  toan, ly, hoa;
};
SinhVien sv[30];   // Mảng 30 sinh viên — gọn gàng, dễ quản lý!
```

## 1.3 Bản Chất Cốt Lõi

**`struct` = Khuôn mẫu (blueprint) để tạo ra kiểu dữ liệu mới do lập trình viên tự định nghĩa.**

**Phép ẩn dụ thực tế:**
```
struct  ←→  Tờ đơn đăng ký (mẫu)
           Có ô "Họ tên", ô "Ngày sinh", ô "Địa chỉ"...

Biến struct  ←→  Tờ đơn đã điền thông tin cụ thể
           Họ tên: Nguyễn Văn A, Ngày sinh: 01/01/2000...
```

## 1.4 So Sánh struct với Kiểu Nguyên Thủy

| | `int`, `float`, `char`... | `struct` |
|---|---|---|
| Do ai định nghĩa | Ngôn ngữ C++ | Lập trình viên tự định nghĩa |
| Lưu được | 1 giá trị đơn | Nhiều giá trị khác kiểu |
| Kích thước | Cố định (4, 8 byte...) | Tổng kích thước các trường |
| Ý nghĩa | Số, ký tự... | Đối tượng thực tế (người, điểm, ngày...) |

## 1.5 Vị Trí của struct trong Lập Trình

```
struct là bước đầu tiên trên con đường OOP:

struct (C)  →  struct có method (C++)  →  class (OOP)
  Dữ liệu      Dữ liệu + Hành vi          Dữ liệu + Hành vi + Đóng gói
```

Trong C++, `struct` và `class` **gần như giống nhau** — chỉ khác ở mặc định truy cập (struct: `public`, class: `private`).

# 2. ĐỊNH NGHĨA VÀ KHAI BÁO KIỂU CẤU TRÚC

## 2.1 Cú Pháp Định Nghĩa

```cpp
struct TênKiểu {

    kiểu_2  trường_2;
    // ...
```

## 2.2 Khai Báo Biến và Khởi Tạo

```cpp
struct DiemSo { int x, y; };

// Cách 1: Khai báo thông thường
DiemSo d1;
d1.x = 3;  d1.y = 4;   // Dùng toán tử . để truy cập trường

// Cách 2: Khởi tạo ngay khi khai báo (aggregate initialization)
DiemSo d2 = {3, 4};    // Theo thứ tự các trường
DiemSo d3 {3, 4};      // C++11 — không cần dấu =

// Cách 3: Khai báo mảng struct
DiemSo mang[5];               // Mảng 5 điểm
DiemSo mang2[] = {{1,2},{3,4},{5,6}};  // Mảng với khởi tạo

// Cách 4: Con trỏ struct
DiemSo* p = &d1;   // p trỏ đến d1
p->x = 10;         // Dùng -> để truy cập qua con trỏ
(*p).y = 20;       // Cách khác — tương đương p->y = 20
```

## 2.3 Toán Tử Truy Cập — `.` và `->` 

```
Biến thường  →  dùng dấu CHẤM  →  sv.hoTen
Con trỏ      →  dùng MŨI TÊN  →  p->hoTen

Quy tắc nhớ: "Dấu chấm cho biến, mũi tên cho con trỏ"
```

## 2.4 struct Lồng Nhau (Nested struct)

```cpp
struct NgaySinh {

struct NguoiDung {

nd.ngaySinh.ngay = 15;    // Truy cập trường lồng nhau
nd.ngaySinh.thang = 8;
nd.ngaySinh.nam = 2000;
```

## 2.5 typedef và using — Đặt Tên Tắt

```cpp
// Cách C cổ điển

// Cách C++ hiện đại

```

## 2.6 Kích Thước struct — Struct Padding

```cpp
struct VD1 { char a; int b; };   // sizeof = 8 (KHÔNG phải 5!)
// Vì CPU đọc dữ liệu theo 'alignment': int cần bắt đầu ở địa chỉ chia hết 4
// → Compiler chèn 3 byte đệm (padding) giữa a và b

// Bộ nhớ thực tế: [a][pad][pad][pad][b b b b]
//                 1   +3       +  4       = 8 bytes
```

⚠️ **Lưu ý:** `sizeof(struct)` không nhất thiết bằng tổng `sizeof` các trường!

````python
%%writefile s2_khai_bao_struct.cpp
#include <iostream>
#include <string>
using namespace std;

// ====== ĐỊNH NGHĨA CÁC STRUCT ======

struct NgaySinh {
    int ngay, thang, nam;
};

struct SinhVien {
    int    maSo;
    string hoTen;
    float  diemTB;
    NgaySinh ns;       // struct lồng nhau
};

// Hàm in ngày sinh
void inNgaySinh(const NgaySinh& ns) {
    cout << ns.ngay << "/" << ns.thang << "/" << ns.nam;
}

// Hàm in thông tin sinh viên
void inSinhVien(const SinhVien& sv) {
    cout << "  [" << sv.maSo << "] "
         << sv.hoTen
         << " | Điểm TB: " << sv.diemTB
         << " | Sinh: ";
    inNgaySinh(sv.ns);
    cout << endl;
}

int main() {
    // ====== CÁCH 1: Khai báo rồi gán từng trường ======
    cout << "=== Cách 1: Gán từng trường ==" << endl;
    SinhVien sv1;
    sv1.maSo  = 1001;
    sv1.hoTen = "Nguyễn Văn An";
    sv1.diemTB = 8.5f;
    sv1.ns.ngay  = 15;
    sv1.ns.thang = 3;
    sv1.ns.nam   = 2003;
    inSinhVien(sv1);

    // ====== CÁCH 2: Khởi tạo ngay khi khai báo ======
    cout << "\n=== Cách 2: Aggregate initialization ==" << endl;
    // {maSo, hoTen, diemTB, {ngay, thang, nam}}
    SinhVien sv2 = {1002, "Trần Thị Bích", 9.2f, {20, 7, 2002}};
    inSinhVien(sv2);

    // ====== CÁCH 3: Mảng struct ======
    cout << "\n=== Cách 3: Mảng struct ==" << endl;
    SinhVien lop[] = {
        {1003, "Lê Văn Cường",  7.8f, {5,  1, 2003}},
        {1004, "Phạm Thị Dung", 8.0f, {12, 9, 2002}},
        {1005, "Hoàng Văn Em",  6.5f, {30, 4, 2003}}
    };
    int n = sizeof(lop) / sizeof(lop[0]);
    for (int i = 0; i < n; i++) inSinhVien(lop[i]);

    // ====== CÁCH 4: Con trỏ struct ======
    cout << "\n=== Cách 4: Con trỏ struct (.  vs ->) ==" << endl;
    SinhVien* p = &sv1;
    cout << "Qua con trỏ  p->hoTen = " << p->hoTen << endl;
    cout << "Qua con trỏ (*p).maSo = " << (*p).maSo << endl;  // Tương đương

    // Sửa qua con trỏ
    p->diemTB = 9.0f;
    cout << "Sau khi sửa qua con trỏ, sv1.diemTB = " << sv1.diemTB << endl;

    // ====== KÍCH THƯỚC struct ======
    cout << "\n=== Kích thước struct ==" << endl;
    struct PaddingVD { char a; int b; double c; };
    cout << "sizeof(char)   = " << sizeof(char)   << endl;
    cout << "sizeof(int)    = " << sizeof(int)    << endl;
    cout << "sizeof(double) = " << sizeof(double) << endl;
    cout << "sizeof(PaddingVD) = " << sizeof(PaddingVD)
         << " (KHÔNG phải " << sizeof(char)+sizeof(int)+sizeof(double)
         << " do padding!)" << endl;

    return 0;
}
````

## 3.1 Truyền struct vào Hàm — 3 Cách

### Cách 1: Truyền theo giá trị (Pass by Value)
```cpp
void inSV(SinhVien sv) { ... }   // sv là bản sao — an toàn, tốn bộ nhớ
```
- ✅ An toàn: hàm không sửa được bản gốc
- ❌ Tốn: Copy toàn bộ struct — chậm nếu struct lớn

### Cách 2: Truyền con trỏ (Pass by Pointer)
```cpp
void capNhat(SinhVien* p) { p->diem = 9.0; }  // Sửa được bản gốc
capNhat(&sv);   // Phải thêm & khi gọi
```
- ✅ Nhanh: chỉ truyền địa chỉ (8 bytes)
- ✅ Sửa được bản gốc
- Cú pháp: dùng `->` bên trong hàm

### Cách 3: Truyền tham chiếu (Pass by Reference) ← Khuyên dùng trong C++
```cpp
void capNhat(SinhVien& sv) { sv.diem = 9.0; }  // Sửa được bản gốc
void docDoc(const SinhVien& sv) { ... }         // Chỉ đọc, không sửa
capNhat(sv);   // Gọi bình thường — không cần &
```
- ✅ Nhanh như con trỏ
- ✅ Cú pháp dùng `.` tự nhiên như biến thường
- `const&` = vừa nhanh vừa an toàn (không sửa được)

## 3.2 Hàm Trả về struct

```cpp
SinhVien taoSinhVien(int maSo, string hoTen, float diem) {
    SinhVien sv;          // Tạo struct cục bộ
    sv.maSo  = maSo;
    sv.hoTen = hoTen;
    sv.diem  = diem;
    return sv;             // Trả về struct — compiler tự tối ưu (RVO)
}

SinhVien sv = taoSinhVien(1001, "Nguyễn A", 8.5);
```

## 3.3 Các Hàm Thao Tác Cơ Bản Với struct

Với mỗi struct, một lập trình viên kinh nghiệm thường xây dựng các hàm:

```
tao()      → Khởi tạo struct với giá trị mặc định hoặc tùy chỉnh
nhap()     → Nhập dữ liệu từ bàn phím
in()       → In thông tin ra màn hình
soSanh()   → So sánh 2 struct (dùng cho sắp xếp)
bang()     → Kiểm tra 2 struct có bằng nhau không
```

## 3.4 Copy struct

```cpp
SinhVien sv1 = {1001, "An", 8.5};
SinhVien sv2 = sv1;    // Copy toàn bộ — C++ tự động sao chép từng trường
sv2.hoTen = "Bích";   // Sửa sv2 không ảnh hưởng sv1
```

⚠️ **Cẩn thận với con trỏ bên trong struct:** Copy nông (shallow copy) — cả hai struct cùng trỏ đến một vùng nhớ!

````python
%%writefile s3_ham_va_struct.cpp
#include <iostream>
#include <string>
using namespace std;

struct SinhVien {
    int    maSo;
    string hoTen;
    float  diemToan;
    float  diemVan;
    float  diemAnh;
};

// ===== HÀM TẠO (Constructor-like) =====
SinhVien taoSV(int maSo, string hoTen, float toan, float van, float anh) {
    SinhVien sv;
    sv.maSo    = maSo;
    sv.hoTen   = hoTen;
    sv.diemToan = toan;
    sv.diemVan  = van;
    sv.diemAnh  = anh;
    return sv;  // Trả về struct
}

// ===== HÀM TÍNH ĐIỂM TRUNG BÌNH =====
float tinhDiemTB(const SinhVien& sv) {   // const& — chỉ đọc, không copy
    return (sv.diemToan + sv.diemVan + sv.diemAnh) / 3.0f;
}

// ===== HÀM XẾP LOẠI =====
string xepLoai(const SinhVien& sv) {
    float dtb = tinhDiemTB(sv);
    if      (dtb >= 9.0f) return "Xuất sắc";
    else if (dtb >= 8.0f) return "Giỏi";
    else if (dtb >= 6.5f) return "Khá";
    else if (dtb >= 5.0f) return "Trung bình";
    else                  return "Yếu";
}

// ===== HÀM IN (tham chiếu const — không sửa, không copy) =====
void inSV(const SinhVien& sv) {
    cout << "  MS: " << sv.maSo
         << " | " << sv.hoTen
         << " | Toán: " << sv.diemToan
         << " | Văn: "  << sv.diemVan
         << " | Anh: "  << sv.diemAnh
         << " | TB: "   << tinhDiemTB(sv)
         << " | "       << xepLoai(sv)
         << endl;
}

// ===== HÀM CẬP NHẬT ĐIỂM (tham chiếu — sửa được bản gốc) =====
void capNhatDiem(SinhVien& sv, float toan, float van, float anh) {
    sv.diemToan = toan;
    sv.diemVan  = van;
    sv.diemAnh  = anh;
}

// ===== HÀM CẬP NHẬT QUA CON TRỎ (so sánh cú pháp) =====
void capNhatQuaConTro(SinhVien* p, float toan) {
    p->diemToan = toan;   // Dùng -> với con trỏ
}

// ===== HÀM TÌM SV ĐIỂM CAO NHẤT (trả về con trỏ) =====
SinhVien* timGioi(SinhVien lop[], int n) {
    SinhVien* gioi = &lop[0];
    for (int i = 1; i < n; i++)
        if (tinhDiemTB(lop[i]) > tinhDiemTB(*gioi))
            gioi = &lop[i];
    return gioi;
}

// ===== HÀM IN CẢ LỚP =====
void inLop(const SinhVien lop[], int n) {
    for (int i = 0; i < n; i++) inSV(lop[i]);
}

int main() {
    // Tạo mảng sinh viên dùng hàm taoSV
    SinhVien lop[] = {
        taoSV(1001, "Nguyễn Văn An",   8.5f, 7.0f, 9.0f),
        taoSV(1002, "Trần Thị Bích",   9.0f, 9.5f, 8.5f),
        taoSV(1003, "Lê Văn Cường",    6.0f, 5.5f, 7.0f),
        taoSV(1004, "Phạm Thị Dung",   4.5f, 5.0f, 4.0f),
        taoSV(1005, "Hoàng Minh Em",   7.5f, 8.0f, 7.0f)
    };
    int n = 5;

    cout << "=== Danh Sách Lớp ==" << endl;
    inLop(lop, n);

    // Cập nhật điểm qua tham chiếu
    cout << "\n=== Cập Nhật Điểm SV 1003 ==" << endl;
    cout << "Trước: "; inSV(lop[2]);
    capNhatDiem(lop[2], 7.5f, 7.0f, 8.0f);  // Sửa qua tham chiếu
    cout << "Sau:   "; inSV(lop[2]);

    // Cập nhật qua con trỏ
    cout << "\n=== Cập Nhật Qua Con Trỏ ==" << endl;
    capNhatQuaConTro(&lop[3], 6.0f);
    inSV(lop[3]);

    // Tìm sinh viên giỏi nhất
    cout << "\n=== Sinh Viên Giỏi Nhất ==" << endl;
    SinhVien* gioi = timGioi(lop, n);
    inSV(*gioi);

    // Copy struct
    cout << "\n=== Minh Họa Copy Struct ==" << endl;
    SinhVien sv_copy = lop[0];       // Copy toàn bộ
    sv_copy.hoTen = "Người Sao Chép"; // Sửa bản copy
    cout << "Gốc:  "; inSV(lop[0]);  // Không thay đổi
    cout << "Copy: "; inSV(sv_copy);

    return 0;
}
````

# 4. KIỂU CẤU TRÚC TỰ TRỎ — DANH SÁCH LIÊN KẾT
---

## 4.1 Khái Niệm Cấu Trúc Tự Trỏ

**Cấu trúc tự trỏ** là `struct` có **một trường là con trỏ trỏ đến chính kiểu struct đó**.

```cpp
struct Node {
    int data;       // Dữ liệu
    Node* next;     // Con trỏ trỏ đến Node khác — tự trỏ!
};
```

Đây là **nền tảng** để xây dựng các cấu trúc dữ liệu động:
- Danh sách liên kết đơn (Singly Linked List)
- Danh sách liên kết đôi (Doubly Linked List)
- Cây nhị phân (Binary Tree)
- Đồ thị (Graph)

## 4.2 Danh Sách Liên Kết Đơn — Bản Chất

```
Mảng thông thường:  Ô nhớ liên tiếp — cứng nhắc
[10]-[20]-[30]-[40]-[50]
 0    1    2    3    4   ← Index cố định

Danh sách liên kết: Ô nhớ RẢI RÁC — linh hoạt
[10|●]──→[20|●]──→[30|●]──→[40|●]──→[50|NULL]
 Node       Node      Node      Node      Node
  ↑
 head (con trỏ đầu danh sách)
```

Mỗi **Node** gồm: **data** (giá trị) + **next** (địa chỉ node tiếp theo)

## 4.3 So Sánh Mảng vs Danh Sách Liên Kết

| Tiêu chí | Mảng | Danh sách liên kết |
|---|---|---|
| Bộ nhớ | Liên tiếp, cố định | Rải rác, động |
| Truy cập phần tử thứ i | O(1) — `a[i]` | O(n) — duyệt từ đầu |
| Chèn ở đầu | O(n) — phải dịch | O(1) — chỉ thay con trỏ |
| Xóa phần tử | O(n) — phải dịch | O(1) — chỉ thay con trỏ |
| Kích thước | Cố định khi khai báo | Thay đổi linh hoạt |
| Cache performance | Tốt | Kém hơn |

## 4.4 Các Thao Tác Cơ Bản

### Tạo node mới:
```cpp
Node* taoNode(int val) {
    Node* p = new Node;   // Cấp phát bộ nhớ
    p->data = val;
    p->next = nullptr;    // Quan trọng: khởi tạo next = NULL
    return p;
}
```

### Chèn vào đầu:
```cpp
void chenDau(Node*& head, int val) {
    Node* moi = taoNode(val);
    moi->next = head;     // Node mới trỏ đến node cũ đầu tiên
    head = moi;           // head bây giờ là node mới
    // Thứ tự 2 dòng trên KHÔNG THỂ đảo!
}
```

### Chèn vào cuối:
```cpp
void chenCuoi(Node*& head, int val) {
    Node* moi = taoNode(val);
    if (!head) { head = moi; return; }  // DS rỗng
    Node* cur = head;
    while (cur->next) cur = cur->next;  // Đi đến cuối
    cur->next = moi;                    // Nối vào cuối
}
```

### Xóa node:
```cpp
// Xóa node đầu
void xoaDau(Node*& head) {
    if (!head) return;      // DS rỗng
    Node* xoa = head;
    head = head->next;      // head tiến lên
    delete xoa;             // Giải phóng bộ nhớ!
}
```

## 4.5 Quy Tắc Vàng Khi Làm Việc Với Danh Sách Liên Kết

```
1. Luôn khởi tạo next = nullptr — tránh dangling pointer
2. Luôn delete node khi không dùng nữa — tránh memory leak  
3. Khi thao tác: lưu con trỏ vào biến temp TRƯỚC khi thay đổi
4. Kiểm tra head == nullptr trước khi duyệt
5. Khi truyền head vào hàm để SỬA: dùng Node*& (tham chiếu con trỏ)
```

````python
%%writefile s4_danh_sach_lien_ket.cpp
#include <iostream>
using namespace std;

// ====== ĐỊNH NGHĨA NODE ======
struct Node {
    int   data;
    Node* next;   // Tự trỏ — trỏ đến Node khác cùng kiểu
};

// ====== HÀM TẠO NODE MỚI ======
Node* taoNode(int val) {
    Node* p = new Node;  // Cấp phát trên heap
    p->data = val;
    p->next = nullptr;   // BẮT BUỘC khởi tạo!
    return p;
}

// ====== IN DANH SÁCH ======
void inDS(Node* head) {
    if (!head) { cout << "[Danh sách rỗng]" << endl; return; }
    Node* cur = head;
    while (cur) {
        cout << cur->data;
        if (cur->next) cout << " → ";
        cur = cur->next;
    }
    cout << " → NULL" << endl;
}

// ====== ĐẾM SỐ NODE ======
int demNode(Node* head) {
    int dem = 0;
    for (Node* cur = head; cur; cur = cur->next) dem++;
    return dem;
}

// ====== CHÈN VÀO ĐẦU ======
// Node*& vì ta cần sửa con trỏ head gốc
void chenDau(Node*& head, int val) {
    Node* moi  = taoNode(val);
    moi->next  = head;   // 1. Node mới trỏ đến node cũ đầu
    head       = moi;    // 2. head = node mới (thứ tự KHÔNG đảo!)
}

// ====== CHÈN VÀO CUỐI ======
void chenCuoi(Node*& head, int val) {
    Node* moi = taoNode(val);
    if (!head) { head = moi; return; }  // DS rỗng
    Node* cur = head;
    while (cur->next) cur = cur->next;  // Đi đến node cuối
    cur->next = moi;                    // Nối node mới vào cuối
}

// ====== CHÈN SAU NODE CÓ GIÁ TRỊ k ======
void chenSau(Node* head, int k, int val) {
    Node* cur = head;
    while (cur && cur->data != k) cur = cur->next;
    if (!cur) { cout << "Không tìm thấy " << k << endl; return; }
    Node* moi  = taoNode(val);
    moi->next  = cur->next;  // 1. Node mới trỏ đến phần sau
    cur->next  = moi;        // 2. Node k trỏ đến node mới
}

// ====== XÓA NODE ĐẦU ======
void xoaDau(Node*& head) {
    if (!head) return;
    Node* xoa = head;        // Lưu lại để delete
    head = head->next;       // head tiến lên
    delete xoa;              // Giải phóng bộ nhớ!
}

// ====== XÓA NODE CÓ GIÁ TRỊ val ======
void xoaGiaTri(Node*& head, int val) {
    if (!head) return;
    // Trường hợp đặc biệt: xóa node đầu
    if (head->data == val) { xoaDau(head); return; }
    // Tìm node đứng TRƯỚC node cần xóa
    Node* prev = head;
    while (prev->next && prev->next->data != val)
        prev = prev->next;
    if (!prev->next) return;  // Không tìm thấy
    Node* xoa  = prev->next;
    prev->next = xoa->next;   // Nối qua node bị xóa
    delete xoa;               // Giải phóng!
}

// ====== TÌM KIẾM ======
Node* timKiem(Node* head, int val) {
    for (Node* cur = head; cur; cur = cur->next)
        if (cur->data == val) return cur;
    return nullptr;
}

// ====== ĐẢO NGƯỢC DANH SÁCH ======
Node* daoNguoc(Node* head) {
    Node* prev = nullptr;
    Node* cur  = head;
    while (cur) {
        Node* next = cur->next;  // Lưu next trước khi cắt
        cur->next  = prev;       // Đảo hướng con trỏ
        prev       = cur;        // Tiến prev
        cur        = next;       // Tiến cur
    }
    return prev;  // prev là head mới
}

// ====== GIẢI PHÓNG TOÀN BỘ ======
void giaiPhong(Node*& head) {
    while (head) xoaDau(head);
}

int main() {
    Node* head = nullptr;  // Danh sách rỗng

    // ====== XÂY DỰNG DANH SÁCH ======
    cout << "=== Xây Dựng Danh Sách ==" << endl;
    chenCuoi(head, 10);
    chenCuoi(head, 20);
    chenCuoi(head, 30);
    chenCuoi(head, 40);
    chenCuoi(head, 50);
    cout << "Sau chen cuoi 10,20,30,40,50: "; inDS(head);

    chenDau(head, 5);
    cout << "Sau chen dau 5:              "; inDS(head);
    cout << "Số node: " << demNode(head) << endl;

    // ====== CHÈN VÀO GIỮA ======
    cout << "\n=== Chèn Vào Giữa ==" << endl;
    chenSau(head, 30, 35);
    cout << "Sau chen 35 sau node[30]: "; inDS(head);

    // ====== XÓA NODE ======
    cout << "\n=== Xóa Node ==" << endl;
    xoaDau(head);
    cout << "Sau xoa dau:        "; inDS(head);
    xoaGiaTri(head, 35);
    cout << "Sau xoa gia tri 35: "; inDS(head);
    xoaGiaTri(head, 50);
    cout << "Sau xoa gia tri 50: "; inDS(head);

    // ====== TÌM KIẾM ======
    cout << "\n=== Tìm Kiếm ==" << endl;
    Node* found = timKiem(head, 30);
    cout << "Tìm 30: " << (found ? "Tìm thấy! data=" + to_string(found->data) : "Không tìm thấy") << endl;
    found = timKiem(head, 99);
    cout << "Tìm 99: " << (found ? "Tìm thấy" : "Không tìm thấy") << endl;

    // ====== ĐẢO NGƯỢC ======
    cout << "\n=== Đảo Ngược Danh Sách ==" << endl;
    cout << "Trước: "; inDS(head);
    head = daoNguoc(head);
    cout << "Sau:   "; inDS(head);

    // ====== GIẢI PHÓNG ======
    giaiPhong(head);
    cout << "\n=== Sau khi giải phóng: ==" << endl;
    inDS(head);

    return 0;
}
````

## 5.7 Sắp Xếp Mảng Con Trỏ — Kỹ Thuật Hiệu Quả

Thay vì hoán đổi struct lớn (tốn thời gian copy), hãy sắp xếp mảng **con trỏ** đến struct:

````python
%%writefile s5_sap_xep_struct.cpp
#include <iostream>
#include <string>
#include <algorithm>
using namespace std;

struct SinhVien {
    int    maSo;
    string hoTen;
    float  diemToan, diemVan, diemAnh;
};

float diemTB(const SinhVien& sv) {
    return (sv.diemToan + sv.diemVan + sv.diemAnh) / 3.0f;
}

void inSV(const SinhVien& sv) {
    cout << "  [" << sv.maSo << "] " << sv.hoTen
         << " | T:" << sv.diemToan
         << " V:" << sv.diemVan
         << " A:" << sv.diemAnh
         << " | TB:" << diemTB(sv)
         << endl;
}

void inLop(SinhVien lop[], int n, const string& tieude) {
    cout << tieude << endl;
    for (int i = 0; i < n; i++) inSV(lop[i]);
}

// ===== HÀM SO SÁNH RỜI RẠC =====
bool theoMaSoTang(const SinhVien& a, const SinhVien& b) {
    return a.maSo < b.maSo;
}
bool theoTenTang(const SinhVien& a, const SinhVien& b) {
    return a.hoTen < b.hoTen;
}
bool theoDiemGiam(const SinhVien& a, const SinhVien& b) {
    return diemTB(a) > diemTB(b);
}

// ===== BUBBLE SORT THỦ CÔNG =====
void bubbleSort(SinhVien lop[], int n) {
    for (int i = 0; i < n-1; i++) {
        bool coHoanDoi = false;
        for (int j = 0; j < n-1-i; j++) {
            if (diemTB(lop[j]) > diemTB(lop[j+1])) {
                swap(lop[j], lop[j+1]);  // Hoán đổi cả struct
                coHoanDoi = true;
            }
        }
        if (!coHoanDoi) break;  // Tối ưu: đã sắp xếp xong sớm
    }
}

int main() {
    SinhVien lop[] = {
        {1003, "Lê Văn Cường",    6.0f, 7.0f, 5.5f},
        {1001, "Nguyễn Văn An",   8.5f, 7.0f, 9.0f},
        {1005, "Hoàng Minh Em",   7.5f, 8.0f, 7.0f},
        {1002, "Trần Thị Bích",   9.0f, 9.5f, 8.5f},
        {1004, "Phạm Thị Dung",   4.5f, 5.0f, 4.0f},
        {1006, "Vũ Thị Fương",    9.0f, 9.0f, 9.0f},  // Cùng điểm với Bích
    };
    int n = 6;

    inLop(lop, n, "=== Danh Sách Ban Đầu ===");

    // ===== SẮP XẾP BẰNG BUBBLE SORT =====
    cout << endl;
    SinhVien lopCopy1[6];
    copy(lop, lop+n, lopCopy1);
    bubbleSort(lopCopy1, n);
    inLop(lopCopy1, n, "=== Bubble Sort — Điểm TB Tăng Dần ===");

    // ===== SẮP XẾP BẰNG std::sort + HÀM SO SÁNH =====
    cout << endl;
    SinhVien lopCopy2[6];
    copy(lop, lop+n, lopCopy2);
    sort(lopCopy2, lopCopy2+n, theoDiemGiam);
    inLop(lopCopy2, n, "=== std::sort — Điểm TB Giảm Dần ===");

    // ===== SẮP XẾP BẰNG LAMBDA =====
    cout << endl;
    SinhVien lopCopy3[6];
    copy(lop, lop+n, lopCopy3);
    sort(lopCopy3, lopCopy3+n, [](const SinhVien& a, const SinhVien& b) {
        return a.hoTen < b.hoTen;
    });
    inLop(lopCopy3, n, "=== Lambda — Theo Tên Tăng Dần ===");

    // ===== SẮP XẾP NHIỀU TIÊU CHÍ =====
    cout << endl;
    SinhVien lopCopy4[6];
    copy(lop, lop+n, lopCopy4);
    sort(lopCopy4, lopCopy4+n, [](const SinhVien& a, const SinhVien& b) {
        float da = diemTB(a), db = diemTB(b);
        if (abs(da - db) > 1e-4f) return da > db;  // Điểm giảm dần
        return a.hoTen < b.hoTen;                   // Bằng điểm → tên tăng dần
    });
    inLop(lopCopy4, n, "=== Nhiều Tiêu Chí: Điểm Giảm, Cùng Điểm → Tên Tăng ===");
    cout << "  (Lưu ý: Bích và Fương cùng TB=9.0, Bích đứng trước vì B < F)" << endl;

    // ===== SẮP XẾP MẢNG CON TRỎ (hiệu quả với struct lớn) =====
    cout << endl;
    cout << "=== Mảng Con Trỏ — Sắp Theo Điểm Toán Giảm ==" << endl;
    SinhVien* ptr[6];
    for (int i = 0; i < n; i++) ptr[i] = &lop[i];
    sort(ptr, ptr+n, [](SinhVien* a, SinhVien* b) {
        return a->diemToan > b->diemToan;  // Chỉ hoán đổi con trỏ!
    });
    for (int i = 0; i < n; i++) inSV(*ptr[i]);
    cout << "  (Mảng lop gốc không thay đổi)" << endl;

    return 0;
}
````

## 6.3 Tìm Kiếm Nhị Phân (Binary Search)

```
Điều kiện TIÊN QUYẾT: Mảng phải được SẮP XẾP theo trường cần tìm!

Ý tưởng: So sánh với phần tử GIỮA:
- Bằng → Tìm thấy!
- Nhỏ hơn → Tìm ở nửa trái
- Lớn hơn → Tìm ở nửa phải

Ví dụ: Tìm maSo=1004 trong {1001, 1002, 1003, 1004, 1005}
  [lo=0, hi=4] mid=2 → a[2]=1003 < 1004 → lo=3
  [lo=3, hi=4] mid=3 → a[3]=1004 == 1004 → Tìm thấy!
```

```cpp
int timNhiPhan(SinhVien lop[], int n, int maSo) {

        else if (lop[mid].maSo < maSo)  lo = mid + 1;
```

````python
%%writefile s6_tim_kiem_struct.cpp
#include <iostream>
#include <string>
#include <algorithm>
#include <vector>
using namespace std;

struct SinhVien {
    int    maSo;
    string hoTen;
    float  diemToan, diemVan, diemAnh;
};

float diemTB(const SinhVien& sv) {
    return (sv.diemToan + sv.diemVan + sv.diemAnh) / 3.0f;
}

string xepLoai(const SinhVien& sv) {
    float d = diemTB(sv);
    if (d >= 9.0f) return "Xuất sắc";
    if (d >= 8.0f) return "Giỏi";
    if (d >= 6.5f) return "Khá";
    if (d >= 5.0f) return "Trung bình";
    return "Yếu";
}

void inSV(const SinhVien& sv) {
    cout << "  [" << sv.maSo << "] " << sv.hoTen
         << " | TB:" << diemTB(sv)
         << " | " << xepLoai(sv) << endl;
}

// ===== TÌM TUYẾN TÍNH THEO MÃ SỐ — trả về index =====
int timTheoMaSo(const SinhVien lop[], int n, int maSo) {
    for (int i = 0; i < n; i++)
        if (lop[i].maSo == maSo) return i;
    return -1;  // Không tìm thấy
}

// ===== TÌM TUYẾN TÍNH THEO TÊN (không phân biệt hoa thường) =====
int timTheoTen(const SinhVien lop[], int n, const string& ten) {
    for (int i = 0; i < n; i++)
        if (lop[i].hoTen == ten) return i;
    return -1;
}

// ===== TÌM NHỊ PHÂN THEO MÃ SỐ (mảng phải sorted theo maSo) =====
int timNhiPhan(const SinhVien lop[], int n, int maSo) {
    int lo = 0, hi = n - 1;
    while (lo <= hi) {
        int mid = lo + (hi - lo) / 2;  // Viết thế này tránh tràn số!
        if      (lop[mid].maSo == maSo) return mid;  // Tìm thấy!
        else if (lop[mid].maSo < maSo)  lo = mid + 1; // Tìm nửa phải
        else                             hi = mid - 1; // Tìm nửa trái
    }
    return -1;
}

// ===== TÌM TẤT CẢ THỎA ĐIỀU KIỆN =====
vector<int> timTatCa(const SinhVien lop[], int n,
                      const string& loai) {
    vector<int> kq;
    for (int i = 0; i < n; i++)
        if (xepLoai(lop[i]) == loai) kq.push_back(i);
    return kq;
}

// ===== TÌM SV CÓ ĐIỂM CAO NHẤT =====
int timCaoNhat(const SinhVien lop[], int n) {
    int best = 0;
    for (int i = 1; i < n; i++)
        if (diemTB(lop[i]) > diemTB(lop[best])) best = i;
    return best;
}

// ===== TÌM SV CÓ ĐIỂM THẤP NHẤT =====
int timThapNhat(const SinhVien lop[], int n) {
    int worst = 0;
    for (int i = 1; i < n; i++)
        if (diemTB(lop[i]) < diemTB(lop[worst])) worst = i;
    return worst;
}

int main() {
    // Mảng đã sắp xếp theo maSo (để dùng binary search)
    SinhVien lop[] = {
        {1001, "Nguyễn Văn An",   8.5f, 7.0f, 9.0f},
        {1002, "Trần Thị Bích",   9.0f, 9.5f, 8.5f},
        {1003, "Lê Văn Cường",    6.0f, 7.0f, 5.5f},
        {1004, "Phạm Thị Dung",   4.5f, 5.0f, 4.0f},
        {1005, "Hoàng Minh Em",   7.5f, 8.0f, 7.0f},
        {1006, "Vũ Thị Fương",    9.0f, 9.0f, 9.0f},
        {1007, "Đặng Văn Giang",  5.5f, 6.0f, 5.0f},
        {1008, "Bùi Thị Hoa",     8.0f, 7.5f, 8.5f},
    };
    int n = 8;

    cout << "=== Danh Sách Lớp ==" << endl;
    for (int i = 0; i < n; i++) inSV(lop[i]);

    // ===== TÌM TUYẾN TÍNH =====
    cout << "\n=== Tìm Tuyến Tính ==" << endl;
    int idx = timTheoMaSo(lop, n, 1005);
    if (idx != -1) {
        cout << "Tìm maSo=1005: "; inSV(lop[idx]);
    }
    idx = timTheoMaSo(lop, n, 9999);
    cout << "Tìm maSo=9999: "
         << (idx != -1 ? "Tìm thấy" : "Không tìm thấy") << endl;

    idx = timTheoTen(lop, n, "Vũ Thị Fương");
    if (idx != -1) { cout << "Tìm tên 'Vũ Thị Fương': "; inSV(lop[idx]); }

    // ===== TÌM NHỊ PHÂN =====
    cout << "\n=== Tìm Nhị Phân (mảng sorted theo maSo) ==" << endl;
    int targets[] = {1001, 1004, 1008, 1099};
    for (int ms : targets) {
        int pos = timNhiPhan(lop, n, ms);
        cout << "Tìm nhị phân maSo=" << ms << ": ";
        if (pos != -1) inSV(lop[pos]);
        else cout << "Không tìm thấy" << endl;
    }

    // ===== TÌM TẤT CẢ THEO LOẠI =====
    cout << "\n=== Tìm Tất Cả Sinh Viên Giỏi ==" << endl;
    auto gioi = timTatCa(lop, n, "Giỏi");
    if (gioi.empty()) cout << "Không có" << endl;
    else for (int i : gioi) inSV(lop[i]);

    cout << "\n=== Tìm Tất Cả Sinh Viên Xuất Sắc ==" << endl;
    auto xs = timTatCa(lop, n, "Xuất sắc");
    if (xs.empty()) cout << "Không có" << endl;
    else for (int i : xs) inSV(lop[i]);

    // ===== TÌM CAO/THẤP NHẤT =====
    cout << "\n=== SV Điểm Cao/Thấp Nhất ==" << endl;
    cout << "Cao nhất: "; inSV(lop[timCaoNhat(lop, n)]);
    cout << "Thấp nhất: "; inSV(lop[timThapNhat(lop, n)]);

    // ===== DÙNG std::find_if =====
    cout << "\n=== Dùng std::find_if (C++ Hiện Đại) ==" << endl;
    // Tìm SV đầu tiên có điểm toán >= 9.0
    auto it = find_if(lop, lop+n, [](const SinhVien& sv) {
        return sv.diemToan >= 9.0f;
    });
    cout << "SV đầu tiên có điểm Toán >= 9: ";
    if (it != lop+n) inSV(*it);
    else cout << "Không có" << endl;

    // Đếm số SV đạt loại Khá
    int soKha = count_if(lop, lop+n, [](const SinhVien& sv) {
        return xepLoai(sv) == "Khá";
    });
    cout << "Số SV xếp loại Khá: " << soKha << endl;

    return 0;
}
````

Bài tổng hợp này kết hợp **tất cả** kiến thức đã học: định nghĩa struct, hàm, sắp xếp, tìm kiếm, và danh sách liên kết để xây dựng một hệ thống quản lý mini.

````python
%%writefile s7_tong_hop.cpp
#include <iostream>
#include <string>
#include <algorithm>
#include <iomanip>  // setw, setprecision
using namespace std;

// ===== CẤU TRÚC DỮ LIỆU =====
struct MonHoc {
    string ten;
    float  diem;
    int    soTinChi;
};

struct SinhVien {
    int     maSo;
    string  hoTen;
    string  lop;
    MonHoc  monHoc[5];   // Mảng struct lồng trong struct
    int     soMon;
};

// ===== HÀM TÍNH ĐIỂM TRUNG BÌNH TÍCH LŨY (GPA) =====
float tinhGPA(const SinhVien& sv) {
    if (sv.soMon == 0) return 0.0f;
    float tongDiemTC = 0, tongTC = 0;
    for (int i = 0; i < sv.soMon; i++) {
        tongDiemTC += sv.monHoc[i].diem * sv.monHoc[i].soTinChi;
        tongTC     += sv.monHoc[i].soTinChi;
    }
    return tongDiemTC / tongTC;
}

string xepLoai(const SinhVien& sv) {
    float g = tinhGPA(sv);
    if (g >= 9.0f) return "Xuất sắc";
    if (g >= 8.0f) return "Giỏi";
    if (g >= 6.5f) return "Khá";
    if (g >= 5.0f) return "Trung bình";
    return "Yếu";
}

// ===== HÀM IN CHI TIẾT =====
void inChiTiet(const SinhVien& sv) {
    cout << "┌── SV: [" << sv.maSo << "] " << sv.hoTen
         << " | Lớp: " << sv.lop << endl;
    for (int i = 0; i < sv.soMon; i++) {
        const MonHoc& m = sv.monHoc[i];
        cout << "│   " << setw(20) << left << m.ten
             << " : " << fixed << setprecision(1) << m.diem
             << " (" << m.soTinChi << " TC)" << endl;
    }
    cout << "└── GPA: " << fixed << setprecision(2) << tinhGPA(sv)
         << " | " << xepLoai(sv) << endl;
}

void inTomTat(const SinhVien& sv) {
    cout << "  [" << setw(4) << sv.maSo << "] "
         << setw(20) << left << sv.hoTen
         << " | GPA: " << fixed << setprecision(2) << tinhGPA(sv)
         << " | " << xepLoai(sv) << endl;
}

// ===== HÀM THỐNG KÊ =====
void thongKe(const SinhVien lop[], int n) {
    int dem[5] = {0};  // XS, G, K, TB, Y
    string loai[] = {"Xuất sắc","Giỏi","Khá","Trung bình","Yếu"};
    float tong = 0;
    for (int i = 0; i < n; i++) {
        tong += tinhGPA(lop[i]);
        string xl = xepLoai(lop[i]);
        for (int j = 0; j < 5; j++)
            if (xl == loai[j]) { dem[j]++; break; }
    }
    cout << "  GPA trung bình lớp: " << fixed << setprecision(2) << tong/n << endl;
    for (int j = 0; j < 5; j++)
        if (dem[j] > 0)
            cout << "  " << setw(12) << left << loai[j]
                 << ": " << dem[j] << " SV"
                 << " (" << fixed << setprecision(1)
                 << 100.0f*dem[j]/n << "%)" << endl;
}

int main() {
    SinhVien lop[] = {
        {1001, "Nguyễn Văn An",   "CS01",
            {{"Toán",8.5f,4},{"Lý",7.0f,3},{"Hóa",9.0f,3},{"Anh",8.0f,3},{"Lập trình",9.5f,4}}, 5},
        {1002, "Trần Thị Bích",   "CS01",
            {{"Toán",9.5f,4},{"Lý",9.0f,3},{"Hóa",8.5f,3},{"Anh",9.0f,3},{"Lập trình",9.0f,4}}, 5},
        {1003, "Lê Văn Cường",    "CS02",
            {{"Toán",6.0f,4},{"Lý",7.0f,3},{"Hóa",5.5f,3},{"Anh",6.5f,3},{"Lập trình",7.0f,4}}, 5},
        {1004, "Phạm Thị Dung",   "CS02",
            {{"Toán",4.5f,4},{"Lý",5.0f,3},{"Hóa",4.0f,3},{"Anh",5.5f,3},{"Lập trình",4.0f,4}}, 5},
        {1005, "Hoàng Minh Em",   "CS01",
            {{"Toán",7.5f,4},{"Lý",8.0f,3},{"Hóa",7.0f,3},{"Anh",7.5f,3},{"Lập trình",8.5f,4}}, 5},
    };
    int n = 5;

    // ===== IN CHI TIẾT =====
    cout << "╔══════════════════════════════════════╗" << endl;
    cout << "║     HỆ THỐNG QUẢN LÝ SINH VIÊN       ║" << endl;
    cout << "╚══════════════════════════════════════╝" << endl;
    for (int i = 0; i < n; i++) { inChiTiet(lop[i]); cout << endl; }

    // ===== SẮP XẾP THEO GPA GIẢM DẦN =====
    cout << "=== Bảng Xếp Hạng (GPA Giảm Dần) ==" << endl;
    SinhVien lopSorted[5];
    copy(lop, lop+n, lopSorted);
    sort(lopSorted, lopSorted+n, [](const SinhVien& a, const SinhVien& b) {
        return tinhGPA(a) > tinhGPA(b);
    });
    for (int i = 0; i < n; i++) {
        cout << "  #" << (i+1) << " ";
        inTomTat(lopSorted[i]);
    }

    // ===== TÌM KIẾM =====
    cout << "\n=== Tìm Kiếm ==" << endl;
    // Tìm theo mã số
    int target = 1003;
    auto it = find_if(lop, lop+n, [&](const SinhVien& sv) {
        return sv.maSo == target;
    });
    cout << "Tìm maSo=" << target << ": ";
    if (it != lop+n) inTomTat(*it);
    else cout << "Không tìm thấy" << endl;

    // Tìm SV xuất sắc
    cout << "\nSV có GPA >= 9.0:" << endl;
    for (int i = 0; i < n; i++)
        if (tinhGPA(lop[i]) >= 9.0f) inTomTat(lop[i]);

    // ===== THỐNG KÊ =====
    cout << "\n=== Thống Kê Lớp ==" << endl;
    thongKe(lop, n);

    return 0;
}
````

### 🗂️ SO SÁNH CÁC PHƯƠNG THỨC TRUYỀN THAM SỐ STRUCT

| Cách | Cú pháp | Sửa gốc? | Tốc độ | Nên dùng khi |
|---|---|---|---|---|
| By value | `f(T x)` | ❌ Không | Chậm (copy) | Struct nhỏ, cần bản sao |
| By pointer | `f(T* p)` | ✅ Có | Nhanh | Tương thích C, dynamic allocation |
| By reference | `f(T& x)` | ✅ Có | Nhanh | **Cần sửa gốc** |
| By const ref | `f(const T& x)` | ❌ Không | Nhanh | **Chỉ đọc** ← Khuyên dùng |

### 🗂️ MẢNG VS DANH SÁCH LIÊN KẾT

| | Mảng struct | Danh sách liên kết |
|---|---|---|
| Truy cập `a[i]` | O(1) | O(n) |
| Chèn đầu | O(n) | O(1) |
| Chèn cuối | O(1)* | O(n) |
| Xóa giữa | O(n) | O(n) tìm + O(1) xóa |
| Kích thước | Cố định | Linh hoạt |
| Binary search | ✅ Được | ❌ Không |
| Cache | Tốt | Kém hơn |
