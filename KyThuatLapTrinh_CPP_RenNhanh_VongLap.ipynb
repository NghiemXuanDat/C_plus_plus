### 🔑 Bản chất cốt lõi — Rẽ nhánh là gì?

Bình thường, chương trình chạy **từ trên xuống dưới** theo thứ tự. **Cấu trúc rẽ nhánh** cho phép chương trình **"chọn con đường"** dựa trên điều kiện — giống như ngã tư đường:

```
   [Điều kiện]
      /    \
   Đúng    Sai
    ↓        ↓
 [Làm A]  [Làm B]
      \    /
    [Tiếp tục]
```

Trong C++, có **3 công cụ** rẽ nhánh:
1. `if - else` — Lựa chọn 2 hướng
2. `if - else if - else` — Lựa chọn nhiều hướng liên tiếp
3. `switch - case` — Lựa chọn theo giá trị cụ thể

## 1.1 Cấu Trúc `if - else`

### 🔑 Bản chất

`if-else` là cấu trúc **"nếu... thì... ngược lại..."** — đưa ra quyết định nhị phân (chỉ 2 lựa chọn: Đúng hoặc Sai).

### 📌 Cú pháp đầy đủ

```cpp
if (điều_kiện) {
    // Thực thi khi điều kiện ĐÚNG (true / != 0)
} else {
    // Thực thi khi điều kiện SAI (false / == 0)
}
```

### 📌 Các biến thể

```cpp
// Biến thể 1: Chỉ có if, không có else
if (x > 0) {
    cout << "Dương";
}
// Nếu x <= 0: chương trình bỏ qua khối if và đi tiếp

// Biến thể 2: Không có ngoặc nhọn (chỉ 1 lệnh)
if (x > 0) cout << "Dương";   // Chỉ áp dụng cho 1 lệnh duy nhất
else       cout << "Không dương";

// Biến thể 3: Toán tử 3 ngôi (ternary) — rút gọn if-else
int max = (a > b) ? a : b;   // Đọc: "nếu a>b thì lấy a, ngược lại lấy b"
```

### 🧠 Điều kiện trong C++ — Hiểu đúng

Trong C++, **mọi giá trị khác 0 đều là `true`**, chỉ `0` là `false`:
```cpp
if (1)   // true — luôn vào
if (0)   // false — không bao giờ vào
if (-5)  // true — khác 0
if (x)   // true nếu x != 0
```

### ⚠️ Lỗi kinh điển: `=` nhầm với `==`

```cpp
// SAI — gán x = 5 (luôn true, không bao giờ sai)
if (x = 5) { ... }   // BUG nguy hiểm! Compiler không báo lỗi!

// ĐÚNG — so sánh x có bằng 5 không
if (x == 5) { ... }

// TRICK phòng tránh: Đặt hằng số bên trái (Yoda condition)
if (5 == x) { ... }  // Nếu vô tình viết = thì compiler báo lỗi ngay
```

### 🔥 Skill: if lồng nhau (nested if)
```cpp
// If lồng nhau — dùng khi cần kiểm tra điều kiện sau khi đã qua điều kiện trước
if (tuoi >= 18) {
    if (coTheCC) {
        cout << "Được lái xe";
    } else {
        cout << "Đủ tuổi nhưng không có thẻ";
    }
} else {
    cout << "Chưa đủ tuổi";
}
```

### ⚠️ Lưu ý: Dangling else
```cpp
// Else luôn khớp với if GẦN NHẤT phía trên
if (a > 0)
    if (b > 0)
        cout << "Cả hai dương";
    else                         // Else này thuộc if(b>0), KHÔNG phải if(a>0)!
        cout << "a dương, b không dương";
// → Luôn dùng ngoặc nhọn {} để tránh nhầm lẫn!
```

````python
%%writefile bai1_1_if_else.cpp
#include <iostream>
using namespace std;

int main() {
    // ====== VÍ DỤ 1: if-else cơ bản ======
    cout << "=== Ví dụ 1: Kiểm tra số chẵn/lẻ ==" << endl;
    int n = 17;
    if (n % 2 == 0) {
        cout << n << " là số chẵn" << endl;
    } else {
        cout << n << " là số lẻ" << endl;
    }

    // ====== VÍ DỤ 2: Toán tử 3 ngôi (ternary) ======
    cout << "\n=== Ví dụ 2: Toán tử 3 ngôi ==" << endl;
    int a = 15, b = 8;
    int maxVal = (a > b) ? a : b;  // Đọc: a>b ? "lấy a" : "lấy b"
    int minVal = (a < b) ? a : b;
    cout << "Max(" << a << ", " << b << ") = " << maxVal << endl;
    cout << "Min(" << a << ", " << b << ") = " << minVal << endl;

    // Ternary trong cout — gọn hơn
    cout << n << " là số " << (n % 2 == 0 ? "chẵn" : "lẻ") << endl;

    // ====== VÍ DỤ 3: if lồng nhau ======
    cout << "\n=== Ví dụ 3: if lồng nhau - Xét tam giác ==" << endl;
    // Kiểm tra 3 số có tạo thành tam giác không
    // và tam giác đó loại gì
    int x = 5, y = 5, z = 5;
    if (x + y > z && x + z > y && y + z > x) {  // Điều kiện tam giác
        cout << "(" << x << "," << y << "," << z << ") tạo thành tam giác" << endl;
        if (x == y && y == z) {
            cout << "→ Tam giác đều" << endl;
        } else if (x == y || y == z || x == z) {
            cout << "→ Tam giác cân" << endl;
        } else {
            cout << "→ Tam giác thường" << endl;
        }
    } else {
        cout << "Không tạo thành tam giác" << endl;
    }

    // ====== VÍ DỤ 4: Chỉ có if, không có else ======
    cout << "\n=== Ví dụ 4: if không có else ==" << endl;
    int diem = 85;
    cout << "Điểm: " << diem;
    if (diem >= 90) cout << " → Được học bổng!";  // Chỉ in nếu >= 90
    cout << endl;  // Luôn xuống dòng dù có hay không có học bổng

    // ====== VÍ DỤ 5: Minh họa lỗi dangling else ======
    cout << "\n=== Ví dụ 5: Dangling else ==" << endl;
    int p = 5, q = -3;

    // CÁCH SAI — dễ nhầm (không dùng ngoặc)
    // if (p > 0)
    //     if (q > 0) cout << "Cả hai dương";
    //     else cout << "..." // Của if(q>0), không phải if(p>0)!

    // CÁCH ĐÚNG — luôn dùng ngoặc nhọn
    if (p > 0) {
        if (q > 0) {
            cout << "Cả hai dương" << endl;
        } else {
            cout << "p dương (p=" << p << "), q không dương (q=" << q << ")" << endl;
        }
    } else {
        cout << "p không dương" << endl;
    }

    return 0;
}
````

## 1.2 Cấu Trúc `if - else if - else`

### 🔑 Bản chất

Khi có **nhiều hơn 2 trường hợp**, ta cần `else if` để kiểm tra **nhiều điều kiện theo chuỗi**. Chương trình kiểm tra từ trên xuống, gặp điều kiện đúng đầu tiên thì thực thi và **bỏ qua tất cả phần còn lại**.

```
Kiểm tra ĐK1
    ↓ Sai
Kiểm tra ĐK2
    ↓ Sai
Kiểm tra ĐK3
    ↓ Sai
Thực thi else (mặc định)
```

### 📌 Cú pháp

```cpp
if (điều_kiện_1) {
    // Thực thi khi ĐK1 đúng
} else if (điều_kiện_2) {
    // Thực thi khi ĐK1 sai VÀ ĐK2 đúng
} else if (điều_kiện_3) {
    // Thực thi khi ĐK1, ĐK2 sai VÀ ĐK3 đúng
} else {
    // Thực thi khi TẤT CẢ điều kiện trên đều sai
}
```

### 🔥 Skill quan trọng: Thứ tự điều kiện ảnh hưởng đến kết quả!

```cpp
// ❌ SAI — điều kiện rộng trước, hẹp sau: không bao giờ vào case điểm A!
if (diem >= 50) cout << "Đạt";         // Diem 95 vào đây rồi, kết thúc
else if (diem >= 90) cout << "Xuất sắc"; // Không bao giờ tới đây!

// ✅ ĐÚNG — điều kiện hẹp (cụ thể) trước, rộng sau
if (diem >= 90) cout << "Xuất sắc";    // Kiểm tra hẹp nhất trước
else if (diem >= 70) cout << "Khá";
else if (diem >= 50) cout << "Đạt";
else cout << "Chưa đạt";
```

### 🔥 Skill: Khi nào dùng `else if` vs `if` riêng lẻ?

```cpp
// Dùng else if: các trường hợp LOẠI TRỪ nhau — chỉ 1 trường hợp đúng
// (không thể vừa < 0 vừa > 0)
if (x < 0)      cout << "Âm";
else if (x > 0) cout << "Dương";
else            cout << "Bằng 0";

// Dùng if riêng lẻ: các trường hợp CÓ THỂ cùng xảy ra
// (có thể vừa là số lẻ vừa > 10)
if (x % 2 != 0) cout << "Số lẻ ";
if (x > 10)     cout << "Lớn hơn 10 ";
if (x < 0)      cout << "Số âm ";
```

````python
%%writefile bai1_2_else_if.cpp
#include <iostream>
using namespace std;

// Hàm xếp loại học lực
string xepLoai(int diem) {
    // Điều kiện hẹp (cụ thể nhất) trước!
    if (diem >= 90)      return "Xuất sắc (A+)";
    else if (diem >= 80) return "Giỏi (A)";
    else if (diem >= 70) return "Khá (B)";
    else if (diem >= 60) return "Trung bình khá (C)";
    else if (diem >= 50) return "Trung bình (D)";
    else                 return "Yếu (F)";
}

// Hàm nhận dạng ký tự
string nhanDangKyTu(char c) {
    if (c >= 'A' && c <= 'Z') return "Chữ hoa";
    else if (c >= 'a' && c <= 'z') return "Chữ thường";
    else if (c >= '0' && c <= '9') return "Chữ số";
    else if (c == ' ')  return "Dấu cách";
    else return "Ký tự đặc biệt";
}

// Hàm mô tả số (nhiều điều kiện KHÔNG loại trừ nhau → dùng if riêng lẻ)
void moTaSo(int n) {
    cout << n << " là: ";
    if (n > 0)      cout << "[dương] ";
    if (n < 0)      cout << "[âm] ";
    if (n == 0)     cout << "[bằng 0] ";
    if (n % 2 == 0) cout << "[chẵn] ";
    else            cout << "[lẻ] ";
    if (n > 100)    cout << "[lớn hơn 100] ";
    cout << endl;
}

int main() {
    // ====== Xếp loại học lực ======
    cout << "=== Xếp Loại Học Lực ==" << endl;
    int diemTest[] = {95, 83, 72, 65, 53, 40, 100, 0};
    for (int d : diemTest) {
        cout << "Điểm " << d << " → " << xepLoai(d) << endl;
    }

    // ====== Nhận dạng ký tự ======
    cout << "\n=== Nhận Dạng Ký Tự ==" << endl;
    char kyTuTest[] = {'A', 'z', '5', ' ', '!', '@'};
    for (char c : kyTuTest) {
        cout << "'" << c << "' → " << nhanDangKyTu(c) << endl;
    }

    // ====== if riêng lẻ vs else if ======
    cout << "\n=== Mô Tả Số (if riêng lẻ) ==" << endl;
    moTaSo(7);
    moTaSo(-4);
    moTaSo(0);
    moTaSo(150);

    // ====== Minh họa thứ tự điều kiện quan trọng ======
    cout << "\n=== Thứ Tự Điều Kiện Quan Trọng ==" << endl;
    int diem = 95;

    cout << "Điểm " << diem << ":" << endl;

    // Thứ tự SAI — điều kiện rộng trước
    if (diem >= 50)       cout << "  [THỨ TỰ SAI]  → Đạt (không bao giờ in Xuất sắc!)" << endl;
    else if (diem >= 90)  cout << "  → Xuất sắc (không bao giờ tới đây)" << endl;

    // Thứ tự ĐÚNG — điều kiện hẹp trước
    if (diem >= 90)       cout << "  [THỨ TỰ ĐÚNG] → Xuất sắc" << endl;
    else if (diem >= 50)  cout << "  → Đạt" << endl;

    return 0;
}
````

## 1.3 Cấu Trúc `switch - case`

### 🔑 Bản chất

`switch-case` là cấu trúc rẽ nhánh **chuyên dụng** khi cần so sánh **một biến với nhiều giá trị cụ thể** — như một **bảng tra cứu** thay vì dãy `if-else if` dài dòng.

**Phép ẩn dụ:** `switch` giống như **bảng số phòng trong khách sạn** — bạn đến số phòng nào thì vào phòng đó, không cần kiểm tra từng phòng một.

### 📌 Cú pháp đầy đủ

```cpp
switch (biểu_thức) {       // biểu thức phải cho ra int, char, enum
    case giá_trị_1:
        // Thực thi khi biểu_thức == giá_trị_1
        break;             // THOÁT khỏi switch — quan trọng!

    case giá_trị_2:
        // Thực thi khi biểu_thức == giá_trị_2
        break;

    default:               // Tùy chọn — như else, khi không khớp case nào
        // Thực thi khi không khớp bất kỳ case nào
        break;
}
```

### ⚠️ ĐIỀU QUAN TRỌNG NHẤT: Fall-through

Nếu **quên `break`**, chương trình sẽ **"rơi xuống" (fall-through)** và thực thi tiếp `case` bên dưới — đây là nguồn gốc BUG phổ biến nhất với `switch`!

```cpp
int n = 2;
switch (n) {
    case 1: cout << "Một ";   // Không có break!
    case 2: cout << "Hai ";   // Không có break!
    case 3: cout << "Ba ";    // Không có break!
    case 4: cout << "Bốn ";   break;  // Có break
}
// Output: "Hai Ba Bốn" — rơi thẳng từ case 2 xuống 3, 4!
```

### 🔥 Skill: Dùng fall-through có chủ đích — Nhóm nhiều case

```cpp
// Xử lý nhiều case giống nhau
switch (ngay) {
    case 1:   // Thứ 2
    case 2:   // Thứ 3
    case 3:   // Thứ 4
    case 4:   // Thứ 5
    case 5:   // Thứ 6 — TẤT CẢ rơi vào đây
        cout << "Ngày trong tuần"; break;
    case 6:   // Thứ 7
    case 0:   // Chủ nhật
        cout << "Cuối tuần"; break;
}
```

### 📌 So sánh `switch` vs `if-else if`

| Tiêu chí | `switch-case` | `if-else if` |
|---|---|---|
| Loại điều kiện | Chỉ so sánh **bằng** với giá trị nguyên/char | Mọi loại điều kiện |
| Kiểu dữ liệu | `int`, `char`, `enum` (KHÔNG phải `float`, `string`) | Mọi kiểu |
| Tốc độ | Nhanh hơn với nhiều case (jump table) | Tuần tự |
| Đọc hiểu | Rõ ràng khi nhiều case | Linh hoạt hơn |
| Fall-through | Có (cần cẩn thận) | Không |

### ⚠️ Giới hạn quan trọng của switch
- Không dùng được với `float`, `double`, `string`
- Giá trị case phải là **hằng số** (không phải biến)
- Không dùng được biểu thức phạm vi (không viết `case x > 5:`)

````python
%%writefile bai1_3_switch_case.cpp
#include <iostream>
using namespace std;

int main() {
    // ====== VÍ DỤ 1: Switch cơ bản — menu ngày trong tuần ======
    cout << "=== Ví dụ 1: Ngày Trong Tuần ==" << endl;
    int ngay = 3;  // 1=Thứ 2, 2=Thứ 3, ..., 7=CN
    cout << "Ngày " << ngay << " là: ";
    switch (ngay) {
        case 1: cout << "Thứ Hai";   break;
        case 2: cout << "Thứ Ba";    break;
        case 3: cout << "Thứ Tư";    break;
        case 4: cout << "Thứ Năm";   break;
        case 5: cout << "Thứ Sáu";   break;
        case 6: cout << "Thứ Bảy";   break;
        case 7: cout << "Chủ Nhật";  break;
        default: cout << "Không hợp lệ"; break;
    }
    cout << endl;

    // ====== VÍ DỤ 2: Fall-through CÓ CHỦ ĐÍCH — nhóm case ======
    cout << "\n=== Ví dụ 2: Fall-through Có Chủ Đích ==" << endl;
    for (int d = 1; d <= 7; d++) {
        cout << "Ngày " << d << ": ";
        switch (d) {
            case 1: case 2: case 3: case 4: case 5:
                cout << "📅 Ngày làm việc" << endl; break;
            case 6: case 7:
                cout << "🎉 Cuối tuần!"    << endl; break;
        }
    }

    // ====== VÍ DỤ 3: Switch với char — máy tính đơn giản ======
    cout << "\n=== Ví dụ 3: Máy Tính Đơn Giản ==" << endl;
    double a = 10.5, b = 3.0;
    char phepTinhArr[] = {'+', '-', '*', '/'};

    for (char phepTinh : phepTinhArr) {
        cout << a << " " << phepTinh << " " << b << " = ";
        switch (phepTinh) {
            case '+':
                cout << a + b; break;
            case '-':
                cout << a - b; break;
            case '*':
                cout << a * b; break;
            case '/':
                if (b != 0) cout << a / b;
                else        cout << "Lỗi: chia cho 0!";
                break;
            default:
                cout << "Phép tính không hợp lệ";
        }
        cout << endl;
    }

    // ====== VÍ DỤ 4: Minh họa BUG fall-through khi quên break ======
    cout << "\n=== Ví dụ 4: BUG Fall-through (quên break) ==" << endl;
    int x = 2;
    cout << "x = " << x << ", switch không có break:" << endl;
    cout << "Output: ";
    switch (x) {
        case 1: cout << "[case1] ";  // Không có break
        case 2: cout << "[case2] ";  // Không có break — x=2 vào đây
        case 3: cout << "[case3] ";  // Tiếp tục rơi xuống!
        case 4: cout << "[case4] ";  break;  // Dừng ở đây
        case 5: cout << "[case5] ";  break;
    }
    cout << "← Thấy bug chưa? Vào case2 rồi rơi xuống case3, case4!" << endl;

    return 0;
}
````

## 📌 Tóm Tắt Phần 1: Khi Nào Dùng Gì?

| Tình huống | Dùng |
|---|---|
| 2 trường hợp đơn giản | `if - else` hoặc toán tử `? :` |
| Nhiều trường hợp với khoảng giá trị (>=, <=, >, <) | `if - else if - else` |
| Nhiều trường hợp với giá trị CỤ THỂ (int, char) | `switch - case` |
| Nhiều điều kiện CÓ THỂ cùng đúng | Nhiều `if` riêng lẻ |

```
⚠️ 3 LỖI HAY GẶP NHẤT:
  1. Viết = thay vì == trong điều kiện
  2. Không có break trong switch → fall-through
  3. Thứ tự điều kiện sai trong else if (rộng trước, hẹp sau)
```

# 🔁 PHẦN 2: CẤU TRÚC VÒNG LẶP (LOOPS)
---

### 🔑 Bản chất cốt lõi — Vòng lặp là gì?

**Vòng lặp** cho phép thực thi một khối lệnh **nhiều lần** mà không phải viết lại code. Không có vòng lặp, muốn in 1000 số ta phải viết 1000 dòng `cout`.

**Phép ẩn dụ:** Vòng lặp giống như **máy photocopy** — thay vì tự tay chép 100 tờ, máy tự làm 100 lần cho bạn.

**Mọi vòng lặp đều có 4 thành phần:**
```
1. KHỞI TẠO   → Thiết lập trạng thái ban đầu (chạy 1 lần)
2. ĐIỀU KIỆN  → Kiểm tra trước/sau mỗi vòng (true = tiếp tục, false = dừng)
3. THÂN LẶP   → Code thực thi mỗi vòng
4. CẬP NHẬT   → Thay đổi biến để tiến đến điều kiện dừng
```

> ⚠️ **Nếu điều kiện không bao giờ sai → Vòng lặp vô tận → Chương trình treo!**

Trong C++, có 3 loại vòng lặp:
- `for` — biết trước số lần lặp
- `while` — kiểm tra điều kiện TRƯỚC
- `do-while` — kiểm tra điều kiện SAU (luôn chạy ít nhất 1 lần)

## 2.1 Vòng Lặp `for`

### 🔑 Bản chất

`for` là vòng lặp **"biết trước số lần"** — cả 3 thành phần (khởi tạo, điều kiện, cập nhật) được gom vào 1 dòng, giúp code gọn và rõ ràng.

### 📌 Cú pháp & Giải phẫu

```cpp
//      ┌ Khởi tạo  ┬ Điều kiện  ┬ Cập nhật ┐
for (int i = 0;    i < n;        i++       ) {
    // Thân lặp — thực thi khi điều kiện ĐÚNG
}

// Luồng thực thi:
// 1. Khởi tạo: int i = 0    (chỉ chạy 1 lần)
// 2. Kiểm tra: i < n?       (nếu sai → thoát)
// 3. Thực thi thân lặp
// 4. Cập nhật: i++
// 5. Quay lại bước 2
```

### 📌 Các biến thể quan trọng

```cpp
// Đếm tiến (tăng dần)
for (int i = 0; i < n; i++)      // i = 0, 1, 2, ..., n-1
for (int i = 1; i <= n; i++)     // i = 1, 2, 3, ..., n

// Đếm lùi (giảm dần)
for (int i = n-1; i >= 0; i--)  // i = n-1, n-2, ..., 0
for (int i = n; i >= 1; i--)    // i = n, n-1, ..., 1

// Bước nhảy tùy ý
for (int i = 0; i < n; i += 2)  // i = 0, 2, 4, 6, ... (chỉ số chẵn)
for (int i = 1; i < n; i += 2)  // i = 1, 3, 5, 7, ... (chỉ số lẻ)

// Range-based for (C++11) — duyệt mảng/container gọn nhất
int arr[] = {1, 2, 3, 4, 5};
for (int x : arr) cout << x;   // Đọc từng phần tử
for (int& x : arr) x *= 2;     // Sửa từng phần tử (cần &)

// For vô tận — kết hợp với break
for (;;) {    // Không có khởi tạo, điều kiện, cập nhật
    if (điều_kiện_dừng) break;
}
```

### 🔥 Skill: For lồng nhau (nested for)
```cpp
// In bảng cửu chương — 2 vòng lặp lồng nhau
for (int i = 1; i <= 9; i++) {        // Vòng ngoài: dòng
    for (int j = 1; j <= 9; j++) {    // Vòng trong: cột
        cout << i * j << "\t";
    }
    cout << endl;
}
// Mỗi lần vòng ngoài chạy 1 bước, vòng trong chạy HẾT 9 bước
// Tổng số lần thực thi thân trong = 9 × 9 = 81 lần
```

### ⚠️ Lưu ý quan trọng
- Biến `i` khai báo trong `for` chỉ tồn tại **trong phạm vi vòng for** (C++11 trở đi)
- Tránh sửa đổi biến lặp trong thân lặp — gây bug khó tìm
- `i++` và `++i` đều được trong for, nhưng `++i` nhanh hơn tí (với object phức tạp)

````python
%%writefile bai2_1_for.cpp
#include <iostream>
#include <vector>
using namespace std;

int main() {
    // ====== VÍ DỤ 1: For cơ bản ======
    cout << "=== Ví dụ 1: Tính Tổng 1..10 ==" << endl;
    int tong = 0;
    for (int i = 1; i <= 10; i++) {
        tong += i;
        cout << "i=" << i << ", tong=" << tong << endl;
    }
    cout << "Kết quả: " << tong << endl;

    // ====== VÍ DỤ 2: Đếm lùi ======
    cout << "\n=== Ví dụ 2: Đếm Ngược ==" << endl;
    for (int i = 5; i >= 1; i--) {
        cout << i << "... ";
    }
    cout << "Phóng!" << endl;

    // ====== VÍ DỤ 3: Bước nhảy tùy ý ======
    cout << "\n=== Ví dụ 3: Số Chẵn và Lẻ ==" << endl;
    cout << "Số chẵn từ 2 đến 20: ";
    for (int i = 2; i <= 20; i += 2) cout << i << " ";
    cout << endl;

    cout << "Số lẻ từ 1 đến 19:  ";
    for (int i = 1; i <= 19; i += 2) cout << i << " ";
    cout << endl;

    // ====== VÍ DỤ 4: Range-based for (C++11) ======
    cout << "\n=== Ví dụ 4: Range-based For ==" << endl;
    int arr[] = {5, 3, 8, 1, 9, 2, 7};

    // Đọc phần tử
    cout << "Mảng: ";
    for (int x : arr) cout << x << " ";
    cout << endl;

    // Sửa phần tử (cần &)
    for (int& x : arr) x *= 2;  // Nhân đôi mỗi phần tử
    cout << "Nhân đôi: ";
    for (int x : arr) cout << x << " ";
    cout << endl;

    // Với vector
    vector<string> fruits = {"táo", "cam", "xoài", "chuối"};
    cout << "Hoa quả: ";
    for (const string& f : fruits) cout << f << " ";
    cout << endl;

    // ====== VÍ DỤ 5: For lồng nhau — Hình tam giác ======
    cout << "\n=== Ví dụ 5: Vẽ Hình Tam Giác ==" << endl;
    int n = 5;
    for (int i = 1; i <= n; i++) {       // Vòng ngoài: số hàng
        for (int j = 1; j <= i; j++) {   // Vòng trong: số sao mỗi hàng
            cout << "* ";
        }
        cout << endl;
    }

    // ====== VÍ DỤ 6: For lồng nhau — Bảng nhân ======
    cout << "\n=== Ví dụ 6: Bảng Nhân 3x3 ==" << endl;
    for (int i = 1; i <= 3; i++) {
        for (int j = 1; j <= 3; j++) {
            cout << i << "x" << j << "=" << i*j;
            cout << (j < 3 ? "  " : "\n");  // Tab hoặc xuống dòng
        }
    }

    return 0;
}
````

## 2.2 Vòng Lặp `while`

### 🔑 Bản chất

`while` là vòng lặp **"lặp khi điều kiện còn đúng"** — kiểm tra điều kiện **TRƯỚC** khi thực thi thân lặp. Nếu điều kiện sai ngay từ đầu → thân lặp **không chạy lần nào**.

```
     ┌──────────────────┐
     ↓                  │
[Kiểm tra điều kiện]    │
     │ Đúng             │
     ↓                  │
 [Thân lặp] ────────────┘
     │ Sai
     ↓
[Ra khỏi vòng lặp]
```

### 📌 Cú pháp

```cpp
// Khởi tạo trước vòng lặp
int i = 0;
while (i < n) {         // Kiểm tra điều kiện
    // Thân lặp
    i++;                // Cập nhật — ĐỪNG QUÊN không sẽ lặp vô tận!
}
```

### 📌 Khi nào dùng `while` thay vì `for`?

```cpp
// Dùng while khi: Không biết trước số lần lặp

// Ví dụ 1: Xử lý từng chữ số của một số
int n = 12345;
while (n > 0) {
    cout << n % 10;  // Lấy chữ số cuối
    n /= 10;         // Bỏ chữ số cuối
}  // Số lần lặp = số chữ số của n — không biết trước

// Ví dụ 2: Đọc dữ liệu cho đến khi gặp giá trị dừng
int x;
while (cin >> x && x != -1) {
    // Xử lý x
}  // Dừng khi nhập -1 hoặc hết input

// Ví dụ 3: Tìm kiếm — không biết trước tìm ở đâu
int i = 0;
while (i < n && arr[i] != target) i++;  // Tìm đến khi gặp
```

### 🔥 Skill: `while(true)` với `break` — Vòng lặp "game"
```cpp
// Vòng lặp chính của game hoặc menu
while (true) {
    hienThiMenu();
    int luaChon;
    cin >> luaChon;
    if (luaChon == 0) break;  // Thoát khi chọn 0
    xuLyLuaChon(luaChon);
}
```

### ⚠️ Lưu ý cẩn thận với while
- **Luôn đảm bảo điều kiện sẽ sai** tại một thời điểm nào đó
- Cập nhật biến lặp **bên trong** thân lặp
- `while (cin >> x)` — cách đọc đến hết file rất phổ biến trong thi

````python
%%writefile bai2_2_while.cpp
#include <iostream>
using namespace std;

// Hàm đảo ngược số nguyên
long long daoNguocSo(long long n) {
    long long dao = 0;
    bool amTinh = (n < 0);
    if (amTinh) n = -n;
    while (n > 0) {
        dao = dao * 10 + n % 10;  // Thêm chữ số cuối vào bên phải
        n /= 10;                   // Bỏ chữ số cuối
    }
    return amTinh ? -dao : dao;
}

// Hàm đếm số chữ số
int demChuSo(long long n) {
    if (n == 0) return 1;
    if (n < 0) n = -n;
    int dem = 0;
    while (n > 0) {
        dem++;
        n /= 10;
    }
    return dem;
}

// Thuật toán GCD — cực đẹp với while
int gcd(int a, int b) {
    while (b != 0) {
        int temp = b;
        b = a % b;
        a = temp;
    }
    return a;
}

int main() {
    // ====== VÍ DỤ 1: While cơ bản ======
    cout << "=== Ví dụ 1: Đếm Ngược ==" << endl;
    int dem = 5;
    while (dem > 0) {
        cout << dem << " ";
        dem--;  // Đừng quên cập nhật!
    }
    cout << "Hết!" << endl;

    // ====== VÍ DỤ 2: Xử lý từng chữ số ======
    cout << "\n=== Ví dụ 2: Xử Lý Từng Chữ Số ==" << endl;
    long long n = 123456789;
    cout << "Số: " << n << endl;
    cout << "Chữ số (từ phải): ";
    long long temp = n;
    while (temp > 0) {
        cout << temp % 10 << " ";
        temp /= 10;
    }
    cout << endl;
    cout << "Tổng chữ số: ";
    int tongCS = 0;
    temp = n;
    while (temp > 0) { tongCS += temp % 10; temp /= 10; }
    cout << tongCS << endl;

    // ====== VÍ DỤ 3: Đảo ngược số ======
    cout << "\n=== Ví dụ 3: Đảo Ngược Số ==" << endl;
    long long testNums[] = {12345, 100, 9, -678};
    for (long long x : testNums)
        cout << x << " → " << daoNguocSo(x) << endl;

    // ====== VÍ DỤ 4: Đếm chữ số ======
    cout << "\n=== Ví dụ 4: Đếm Chữ Số ==" << endl;
    long long testNums2[] = {0, 7, 42, 1000, 999999};
    for (long long x : testNums2)
        cout << x << " có " << demChuSo(x) << " chữ số" << endl;

    // ====== VÍ DỤ 5: GCD bằng while ======
    cout << "\n=== Ví dụ 5: GCD Thuật Toán Euclid ==" << endl;
    cout << "gcd(48, 18) = " << gcd(48, 18) << endl;   // 6
    cout << "gcd(100, 75) = " << gcd(100, 75) << endl; // 25

    // ====== VÍ DỤ 6: Tìm kiếm trong mảng ======
    cout << "\n=== Ví dụ 6: Tìm Kiếm Tuyến Tính ==" << endl;
    int arr[] = {4, 7, 2, 9, 1, 5, 8};
    int arrSize = 7, target = 9;
    int idx = 0;
    while (idx < arrSize && arr[idx] != target) idx++;

    if (idx < arrSize)
        cout << "Tìm thấy " << target << " tại vị trí " << idx << endl;
    else
        cout << "Không tìm thấy " << target << endl;

    return 0;
}
````

## 2.3 Vòng Lặp `do-while`

### 🔑 Bản chất — Điểm khác biệt cốt lõi

`do-while` **thực thi thân lặp TRƯỚC, kiểm tra điều kiện SAU** — đảm bảo thân lặp **chạy ít nhất 1 lần** dù điều kiện có sai từ đầu.

```
     ┌──────────────────┐
     ↓                  │
 [Thân lặp] ← Chạy trước│
     ↓                  │
[Kiểm tra điều kiện]    │
     │ Đúng ────────────┘
     │ Sai
     ↓
[Ra khỏi vòng lặp]
```

### 📌 Cú pháp

```cpp
do {
    // Thân lặp — luôn chạy ít nhất 1 lần!
    // Cập nhật biến
} while (điều_kiện);   // Chú ý: có dấu ; sau while!
```

### 📌 So sánh 3 loại vòng lặp

```cpp
int i = 10;  // Điều kiện sai ngay từ đầu!

for (int i = 10; i < 5; i++)
    cout << i;    // Không in gì — kiểm tra trước

while (i < 5)
    cout << i;    // Không in gì — kiểm tra trước

do {
    cout << i;    // In ra 10 — chạy trước rồi mới kiểm tra
} while (i < 5);
```

### 🔥 Khi nào dùng `do-while`?

**Trường hợp điển hình nhất:** Nhập liệu có kiểm tra hợp lệ — luôn cần nhập ít nhất 1 lần!

```cpp
int tuoi;
do {
    cout << "Nhập tuổi (1-120): ";
    cin >> tuoi;
    if (tuoi < 1 || tuoi > 120)
        cout << "Tuổi không hợp lệ, nhập lại!" << endl;
} while (tuoi < 1 || tuoi > 120);  // Lặp đến khi hợp lệ
```

### ⚠️ Lưu ý
- Đừng quên dấu `;` sau `while(điều_kiện)` — lỗi cú pháp phổ biến!
- `do-while` ít dùng hơn `for` và `while`, nhưng trong bài thi nhập liệu nó cực kỳ tiện
- Mọi `do-while` đều có thể viết lại bằng `while` (nhưng dài hơn)

````python
%%writefile bai2_3_do_while.cpp
#include <iostream>
using namespace std;

int main() {
    // ====== VÍ DỤ 1: So sánh while vs do-while ======
    cout << "=== Ví dụ 1: While vs Do-While khi ĐK sai từ đầu ==" << endl;

    int x = 10;

    cout << "while (" << x << " < 5): ";
    int w = x;
    while (w < 5) {
        cout << w << " ";
        w++;
    }
    cout << "(không in gì cả)" << endl;

    cout << "do-while (" << x << " < 5): ";
    int d = x;
    do {
        cout << d << " ";  // In ra 10 dù 10 < 5 là sai!
        d++;
    } while (d < 5);
    cout << "(in ra 10 một lần rồi dừng)" << endl;

    // ====== VÍ DỤ 2: Số Collatz (bài toán 3n+1) ======
    // Từ bất kỳ số nào, luôn về 1
    // Quy tắc: nếu chẵn → n/2, nếu lẻ → 3n+1
    cout << "\n=== Ví dụ 2: Dãy Collatz ==" << endl;
    int n = 27;
    cout << "Dãy Collatz bắt đầu từ " << n << ":" << endl;
    int buoc = 0;
    do {
        cout << n;
        if (n % 2 == 0) n = n / 2;
        else            n = 3 * n + 1;
        cout << " → ";
        buoc++;
        if (buoc % 10 == 0) cout << endl;  // Xuống dòng mỗi 10 số
    } while (n != 1);
    cout << "1 (kết thúc sau " << buoc << " bước)" << endl;

    // ====== VÍ DỤ 3: Giải phương trình lặp (phương pháp số) ======
    // Tính căn bậc 2 bằng phương pháp Newton-Raphson
    // Tìm x sao cho x^2 = S: x_next = (x + S/x) / 2
    cout << "\n=== Ví dụ 3: Tính Căn Bậc Hai (Newton-Raphson) ==" << endl;
    double S = 2.0;     // Tính căn 2
    double xi = S / 2;  // Giá trị ban đầu
    double eps = 1e-10; // Độ chính xác
    int iter = 0;
    do {
        double xi_next = (xi + S / xi) / 2.0;
        iter++;
        // Dừng khi thay đổi < eps
        if (iter <= 5)  // Chỉ in 5 bước đầu
            cout << "  Bước " << iter << ": x = " << xi_next << endl;
        if (xi_next - xi < eps && xi_next - xi > -eps) {
            xi = xi_next;
            break;
        }
        xi = xi_next;
    } while (true);
    cout << "Căn bậc 2 của " << S << " ≈ " << xi
         << " (" << iter << " bước)" << endl;

    // ====== VÍ DỤ 4: Mô phỏng nhập liệu có validate ======
    // (Dùng mảng test thay vì cin để chạy không cần nhập)
    cout << "\n=== Ví dụ 4: Mô Phỏng Nhập Liệu Hợp Lệ ==" << endl;
    // Giả lập người dùng nhập: -5, 0, 150, 25 (chỉ 25 là hợp lệ 1-100)
    int inputs[] = {-5, 0, 150, 25};
    int inputIdx = 0;
    int diem;
    do {
        diem = inputs[inputIdx++];
        cout << "Nhập điểm: " << diem;
        if (diem < 0 || diem > 100)
            cout << " ← Không hợp lệ! Nhập lại." << endl;
        else
            cout << " ← Hợp lệ." << endl;
    } while (diem < 0 || diem > 100);
    cout << "Điểm hợp lệ cuối cùng: " << diem << endl;

    return 0;
}
````

## 2.4 Câu Lệnh `break` và `continue`

### 🔑 Bản chất — 2 công cụ điều khiển vòng lặp

Đây là 2 lệnh dùng **bên trong** vòng lặp (và `break` dùng trong `switch`) để **thay đổi luồng thực thi** theo ý muốn.

---

### 📌 `break` — Thoát hoàn toàn

```
Tác dụng: Thoát NGAY LẬP TỨC khỏi vòng lặp (hoặc switch)
          Nhảy đến dòng lệnh ngay SAU vòng lặp
```

```cpp
for (int i = 0; i < 10; i++) {
    if (i == 5) break;    // Thoát ngay khi i = 5
    cout << i;            // Chỉ in 0, 1, 2, 3, 4
}
// Nhảy đến đây
```

**Ứng dụng thực tế:**
- Tìm kiếm: Tìm thấy rồi → dừng ngay, không tìm tiếp
- Điều kiện lỗi: Gặp dữ liệu xấu → dừng xử lý
- Vòng lặp vô tận `while(true)` cần `break` để thoát

---

### 📌 `continue` — Bỏ qua lần này, lặp tiếp

```
Tác dụng: Bỏ qua phần còn lại của thân lặp HIỆN TẠI
          Nhảy đến bước CẬP NHẬT (for) hoặc KIỂM TRA ĐIỀU KIỆN (while)
```

```cpp
for (int i = 0; i < 10; i++) {
    if (i % 2 == 0) continue;  // Số chẵn → bỏ qua
    cout << i;                  // Chỉ in số lẻ: 1, 3, 5, 7, 9
}
```

**Ứng dụng thực tế:**
- Bỏ qua phần tử không thỏa mãn điều kiện khi xử lý mảng
- Lọc dữ liệu: bỏ qua dữ liệu không hợp lệ

---

### ⚠️ Break trong vòng lặp lồng nhau — QUAN TRỌNG!

// break và continue chỉ ảnh hưởng đến vòng lặp GẦN NHẤT chứa nó!

````python
%%writefile bai2_4_break_continue.cpp
#include <iostream>
using namespace std;

int main() {
    // ====== VÍ DỤ 1: break cơ bản ======
    cout << "=== Ví dụ 1: break ==" << endl;
    cout << "In từ 0, dừng khi gặp số chia hết cho 7:" << endl;
    for (int i = 1; i <= 100; i++) {
        if (i % 7 == 0) {
            cout << "[DỪNG tại i=" << i << "]" << endl;
            break;
        }
        cout << i << " ";
    }

    // ====== VÍ DỤ 2: continue cơ bản ======
    cout << "\n=== Ví dụ 2: continue ==" << endl;
    cout << "In số từ 1-20, bỏ qua bội của 3:" << endl;
    for (int i = 1; i <= 20; i++) {
        if (i % 3 == 0) continue;  // Bỏ qua, không in
        cout << i << " ";
    }
    cout << endl;

    // ====== VÍ DỤ 3: Ứng dụng break — Tìm kiếm ======
    cout << "\n=== Ví dụ 3: Tìm Số Nguyên Tố Đầu Tiên > 50 ==" << endl;
    auto laNguyenTo = [](int n) -> bool {
        if (n < 2) return false;
        for (int i = 2; i * i <= n; i++)
            if (n % i == 0) return false;
        return true;
    };
    for (int i = 51; ; i++) {  // Vòng lặp không giới hạn
        if (laNguyenTo(i)) {
            cout << "Số nguyên tố đầu tiên > 50: " << i << endl;
            break;  // Tìm thấy, dừng ngay
        }
    }

    // ====== VÍ DỤ 4: Ứng dụng continue — Lọc dữ liệu ======
    cout << "\n=== Ví dụ 4: Tính Trung Bình Bỏ Qua Số Âm ==" << endl;
    int data[] = {5, -3, 8, -1, 12, 0, 7, -2, 4, 9};
    int dem = 0;
    double tongDuong = 0;
    for (int x : data) {
        if (x <= 0) {
            cout << x << "(bỏ qua) ";
            continue;  // Bỏ qua số âm và 0
        }
        cout << x << "(+) ";
        tongDuong += x;
        dem++;
    }
    cout << "\nTổng số dương: " << tongDuong
         << ", Đếm: " << dem
         << ", TB: " << (dem > 0 ? tongDuong / dem : 0) << endl;

    // ====== VÍ DỤ 5: break trong vòng lặp lồng nhau ======
    cout << "\n=== Ví dụ 5: break Chỉ Thoát Vòng Trong ==" << endl;
    for (int i = 0; i < 3; i++) {
        cout << "i=" << i << ": ";
        for (int j = 0; j < 5; j++) {
            if (j == 3) {
                cout << "[break j=3] ";
                break;  // Chỉ thoát vòng j!
            }
            cout << "j=" << j << " ";
        }
        cout << endl;  // Vòng i vẫn chạy!
    }

    // ====== VÍ DỤ 6: Thoát vòng lồng nhau bằng flag ======
    cout << "\n=== Ví dụ 6: Thoát Cả 2 Vòng Lồng ==" << endl;
    int matrix[3][3] = {
        {1, 2, 3},
        {4, 99, 6},  // 99 ở vị trí (1,1)
        {7, 8, 9}
    };
    int targetVal = 99;
    int foundI = -1, foundJ = -1;
    bool timThay = false;

    for (int i = 0; i < 3 && !timThay; i++) {
        for (int j = 0; j < 3 && !timThay; j++) {
            if (matrix[i][j] == targetVal) {
                foundI = i; foundJ = j;
                timThay = true;  // Kích hoạt dừng cả 2 vòng
            }
        }
    }

    if (timThay)
        cout << "Tìm thấy " << targetVal
             << " tại [" << foundI << "][" << foundJ << "]" << endl;
    else
        cout << "Không tìm thấy " << targetVal << endl;

    // ====== VÍ DỤ 7: continue trong while ======
    cout << "\n=== Ví dụ 7: continue trong While ==" << endl;
    cout << "Các bội của 3 trong [1,20] không phải bội 9: ";
    int k = 0;
    while (k < 20) {
        k++;
        if (k % 3 != 0) continue;  // Không phải bội 3 → bỏ qua
        if (k % 9 == 0) continue;  // Là bội 9 → cũng bỏ qua
        cout << k << " ";          // Chỉ in bội 3 nhưng không phải bội 9
    }
    cout << endl;

    return 0;
}
````

## 2.5 Bài Tập Tổng Hợp — Kết Hợp Tất Cả

### Bài toán: In bảng số nguyên tố và phân tích

Bài tập này kết hợp:
- `for` để duyệt phạm vi
- `if-else if` để phân loại
- `switch` cho menu
- `while` cho xử lý số
- `break` và `continue` để kiểm soát luồng

````python
%%writefile bai2_5_tong_hop.cpp
#include <iostream>
#include <cmath>
using namespace std;

bool laNguyenTo(int n) {
    if (n < 2) return false;
    if (n == 2) return true;
    if (n % 2 == 0) return false;
    for (int i = 3; i * i <= n; i += 2)
        if (n % i == 0) return false;
    return true;
}

// Phân loại số
void phanLoaiSo(int n) {
    cout << "=== Phân Loại Số " << n << " ===" << endl;

    // if-else if cho nhiều điều kiện
    if (n < 0)      cout << "  Loại: Số âm" << endl;
    else if (n == 0) cout << "  Loại: Bằng 0" << endl;
    else if (n == 1) cout << "  Loại: Số 1 (đặc biệt)" << endl;
    else if (laNguyenTo(n)) cout << "  Loại: Số nguyên tố" << endl;
    else            cout << "  Loại: Hợp số" << endl;

    // Chẵn lẻ
    if (n != 0)
        cout << "  Tính chất: " << (n % 2 == 0 ? "Chẵn" : "Lẻ") << endl;

    // Phân tích thừa số
    if (n > 1 && !laNguyenTo(n)) {
        cout << "  Phân tích: " << n << " = ";
        int temp = n;
        bool first = true;
        for (int p = 2; p * p <= temp; p++) {
            int cnt = 0;
            while (temp % p == 0) { cnt++; temp /= p; }
            if (cnt > 0) {
                if (!first) cout << " x ";
                cout << p;
                if (cnt > 1) cout << "^" << cnt;
                first = false;
            }
        }
        if (temp > 1) { if (!first) cout << " x "; cout << temp; }
        cout << endl;
    }
}

// Tìm nguyên tố trong đoạn bằng cách kết hợp for + if + continue
void timNguyenToTrongDoan(int L, int R) {
    cout << "=== Nguyên Tố Trong [" << L << ", " << R << "] ===" << endl;
    int dem = 0;
    for (int i = L; i <= R; i++) {
        if (!laNguyenTo(i)) continue;  // Không phải NT → bỏ qua
        cout << i << " ";
        dem++;
        if (dem % 10 == 0) cout << endl;  // Xuống dòng mỗi 10 số
    }
    cout << "\nTổng: " << dem << " số nguyên tố" << endl;
}

// Switch cho menu báo cáo
void baoCoSo(int n, int loai) {
    switch (loai) {
        case 1:
            // Tổng chữ số
            { int s = 0, t = (n < 0 ? -n : n);
              while (t > 0) { s += t % 10; t /= 10; }
              cout << "Tổng chữ số của " << n << " = " << s; }
            break;
        case 2:
            // Đảo ngược
            { long long d = 0, t = (n < 0 ? -n : n);
              while (t > 0) { d = d*10 + t%10; t /= 10; }
              cout << "Đảo ngược " << n << " = " << (n<0?-d:d); }
            break;
        case 3:
            // Số chữ số
            { int cnt = 0, t = (n < 0 ? -n : n);
              if (t == 0) cnt = 1;
              while (t > 0) { cnt++; t /= 10; }
              cout << n << " có " << cnt << " chữ số"; }
            break;
        default:
            cout << "Loại báo cáo không hợp lệ";
    }
    cout << endl;
}

int main() {
    // Phân loại các số
    int testNums[] = {-5, 0, 1, 2, 17, 18, 36, 97};
    for (int n : testNums) {
        phanLoaiSo(n);
        cout << endl;
    }

    // Tìm nguyên tố
    timNguyenToTrongDoan(1, 100);
    cout << endl;

    // Báo cáo số
    cout << "=== Báo Cáo (Switch) ===" << endl;
    for (int loai = 1; loai <= 3; loai++)
        baoCoSo(123456, loai);

    return 0;
}
````

### 🗂️ PHẦN 1: CẤU TRÚC RẼ NHÁNH

| Cấu trúc | Cú pháp | Dùng khi |
|---|---|---|
| `if` đơn | `if (đk) { ... }` | Làm gì đó nếu điều kiện đúng |
| `if-else` | `if (đk) {...} else {...}` | 2 nhánh loại trừ nhau |
| `if-else if-else` | Chuỗi else if | Nhiều nhánh với điều kiện phức tạp |
| Ternary `?:` | `đk ? a : b` | Gán biến theo điều kiện |
| `switch-case` | `switch(x) { case v: ... break; }` | So sánh bằng với int/char |

### 🗂️ PHẦN 2: VÒNG LẶP

| Vòng lặp | Cú pháp | Dùng khi | Kiểm tra điều kiện |
|---|---|---|---|
| `for` | `for(init; đk; update)` | Biết số lần lặp | Trước |
| `range-for` | `for(int x : arr)` | Duyệt container | - |
| `while` | `while(đk) {...}` | Không biết số lần | Trước |
| `do-while` | `do {...} while(đk);` | Chạy ít nhất 1 lần | Sau |

### 🗂️ ĐIỀU KHIỂN VÒNG LẶP

| Lệnh | Tác dụng | Ảnh hưởng |
|---|---|---|
| `break` | Thoát hoàn toàn | Chỉ vòng lặp gần nhất |
| `continue` | Bỏ qua lần lặp hiện tại | Chỉ vòng lặp gần nhất |
