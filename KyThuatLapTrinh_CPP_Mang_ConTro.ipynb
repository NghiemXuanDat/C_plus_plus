## 1.1 Mảng Là Gì?

### 🔑 Bản chất cốt lõi

**Mảng** là một **dãy các ô nhớ liên tiếp nhau trên RAM**, tất cả cùng kiểu dữ liệu, được truy cập qua **một tên biến duy nhất + chỉ số (index)**.

Hãy hình dung mảng như một **dãy ngăn tủ đánh số từ 0**:
```
Tủ:   [  5  ] [  3  ] [  8  ] [  1  ] [  9  ]
Index:   0       1       2       3       4
```

### 🧠 Tại sao cần mảng?
- Không có mảng: cần `int a, b, c, d, e` — 5 biến riêng lẻ, không thể dùng vòng lặp xử lý.
- Có mảng: `int a[5]` — 1 biến, dùng vòng lặp `for` xử lý 5 phần tử dễ dàng.
- **Khi cần lưu 1000 số:** Không mảng = không thể. Có mảng = `int a[1000]`.

### ⚠️ Lưu ý quan trọng nhất
| Điều cần nhớ | Chi tiết |
|---|---|
| **Index bắt đầu từ 0** | Mảng 5 phần tử: index từ 0 đến 4 |
| **Không kiểm tra biên** | C++ KHÔNG báo lỗi khi truy cập ngoài mảng — gây bug nguy hiểm |
| **Kích thước cố định** | Khai báo `int a[5]` thì mãi là 5 (trừ khi dùng cấp phát động) |
| **Bộ nhớ liên tiếp** | Các phần tử nằm kề nhau trong RAM — đây là nền tảng của con trỏ |

## 1.2 Biến Mảng — Khai Báo & Khởi Tạo

### 📌 Cú pháp khai báo
```cpp
kieuDuLieu tenMang[kichThuoc];
```

### 📌 Các cách khởi tạo
```cpp
int a[5];                     // Khai báo không khởi tạo — giá trị rác!
int b[5] = {1, 2, 3, 4, 5};  // Khởi tạo đầy đủ
int c[5] = {1, 2};           // Phần còn lại tự động = 0
int d[] = {10, 20, 30};      // Compiler tự tính kích thước = 3
int e[5] = {0};              // Tất cả = 0 (cách hay dùng để reset)
```

### 🔥 Skill của lập trình viên kinh nghiệm
- Luôn khởi tạo mảng khi khai báo, hoặc dùng `memset(a, 0, sizeof(a))` để tránh giá trị rác.
- Dùng `sizeof(a)/sizeof(a[0])` để tính số phần tử mảng — không cần nhớ kích thước cứng.
- Khai báo kích thước mảng bằng hằng số `const int N = 100` thay vì số cứng — dễ sửa.

````python
%%writefile bai1_2_bien_mang.cpp
#include <iostream>
using namespace std;

int main() {
    // --- Cách 1: Khai báo và khởi tạo thủ công ---
    int a[5] = {10, 20, 30, 40, 50};

    cout << "=== Mảng a ==" << endl;
    // Truy cập từng phần tử bằng index
    cout << "a[0] = " << a[0] << endl;  // 10
    cout << "a[4] = " << a[4] << endl;  // 50

    // --- Cách 2: In toàn bộ mảng bằng vòng lặp ---
    cout << "\nToàn bộ mảng a: ";
    int n = sizeof(a) / sizeof(a[0]);  // Tự tính số phần tử = 5
    for (int i = 0; i < n; i++) {
        cout << a[i] << " ";
    }
    cout << endl;

    // --- Cách 3: Khởi tạo một phần ---
    int b[5] = {1, 2};  // b = {1, 2, 0, 0, 0}
    cout << "\nMảng b (khởi tạo một phần): ";
    for (int i = 0; i < 5; i++) cout << b[i] << " ";
    cout << endl;

    // --- Lưu ý nguy hiểm: truy cập ngoài biên ---
    // cout << a[10]; // NGUY HIỂM! Không báo lỗi nhưng đọc vùng nhớ lung tung

    return 0;
}
````

## 1.3 Mảng 1 Chiều — Nhập / Xuất / Xử Lý

### 🔑 Bản chất
Mảng 1 chiều = **danh sách tuyến tính**. Hầu hết bài toán thực tế đều bắt đầu từ đây.

### 📌 Các thao tác cơ bản
| Thao tác | Cách làm |
|---|---|
| Nhập từ bàn phím | `for` + `cin >> a[i]` |
| Xuất ra màn hình | `for` + `cout << a[i]` |
| Tìm max/min | Duyệt toàn bộ, so sánh từng phần tử |
| Tính tổng/trung bình | Cộng dồn vào biến `sum` |
| Tìm kiếm | Duyệt tuyến tính hoặc tìm kiếm nhị phân |
| Sắp xếp | Bubble sort, selection sort, hoặc `sort()` từ STL |

### 🔥 Skill quan trọng: Vừa nhập vừa xử lý
Nhiều bài thi yêu cầu tính toán ngay khi nhập — không cần lưu toàn bộ mảng:
```cpp
int x, sum = 0;
for (int i = 0; i < n; i++) {
    cin >> x;
    sum += x;  // Xử lý ngay, không cần lưu vào mảng
}
```

````python
%%writefile bai1_3_mang_1chieu.cpp
#include <iostream>
#include <algorithm>  // Dùng sort()
using namespace std;

int main() {
    // --- Dùng mảng cố định để demo (không cần nhập từ bàn phím) ---
    int a[] = {5, 3, 8, 1, 9, 2, 7, 4, 6};
    int n = sizeof(a) / sizeof(a[0]);  // n = 9

    // 1. In mảng ban đầu
    cout << "Mảng ban đầu: ";
    for (int i = 0; i < n; i++) cout << a[i] << " ";
    cout << endl;

    // 2. Tìm max và min
    int maxVal = a[0], minVal = a[0];
    int maxIdx = 0, minIdx = 0;
    for (int i = 1; i < n; i++) {
        if (a[i] > maxVal) { maxVal = a[i]; maxIdx = i; }
        if (a[i] < minVal) { minVal = a[i]; minIdx = i; }
    }
    cout << "Max = " << maxVal << " tại index " << maxIdx << endl;
    cout << "Min = " << minVal << " tại index " << minIdx << endl;

    // 3. Tính tổng và trung bình
    int sum = 0;
    for (int i = 0; i < n; i++) sum += a[i];
    cout << "Tổng = " << sum << ", Trung bình = " << (double)sum / n << endl;

    // 4. Tìm kiếm tuyến tính (linear search)
    int target = 7;
    bool found = false;
    for (int i = 0; i < n; i++) {
        if (a[i] == target) {
            cout << "Tìm thấy " << target << " tại index " << i << endl;
            found = true;
            break;  // Dừng ngay khi tìm thấy
        }
    }
    if (!found) cout << "Không tìm thấy " << target << endl;

    // 5. Sắp xếp tăng dần bằng sort() (nhanh nhất)
    sort(a, a + n);  // sort(đầu_mảng, sau_phần_tử_cuối)
    cout << "Mảng sau khi sắp xếp: ";
    for (int i = 0; i < n; i++) cout << a[i] << " ";
    cout << endl;

    return 0;
}
````

## 1.4 Xâu Ký Tự (String / C-String)

### 🔑 Bản chất — 2 loại string trong C++

**Loại 1: C-String** (mảng ký tự kiểu C — cổ điển)
```cpp
char s[100] = "Hello";
```
Đây thực chất là mảng `char`: `['H','e','l','l','o','\0']`  
Ký tự `\0` (null terminator) **bắt buộc phải có** ở cuối — đánh dấu kết thúc chuỗi.

**Loại 2: std::string** (kiểu C++ hiện đại — dùng trong thi/dự án)
```cpp
#include <string>
string s = "Hello";
```
Tiện lợi hơn, tự quản lý bộ nhớ, có nhiều hàm hỗ trợ.

### 📌 So sánh 2 loại
| Tiêu chí | C-String (`char[]`) | `std::string` |
|---|---|---|
| Nối chuỗi | `strcat(a, b)` | `a + b` hoặc `a += b` |
| So sánh | `strcmp(a, b)` | `a == b` |
| Độ dài | `strlen(a)` | `a.length()` hoặc `a.size()` |
| An toàn | Dễ lỗi tràn bộ nhớ | An toàn hơn |
| Hiệu năng | Nhanh hơn | Đủ nhanh cho hầu hết bài |

### ⚠️ Lưu ý quan trọng với C-String
- `char s[10] = "Hello"` — cần khai báo đủ chỗ (ít nhất độ dài + 1 cho `\0`)
- Đọc cả dòng có khoảng trắng: dùng `cin.getline(s, 100)` thay vì `cin >> s`
- Với `std::string`: dùng `getline(cin, s)`

### 🔥 Skill: Đọc string có khoảng trắng
```cpp
string s;
getline(cin, s);   // Đọc cả dòng kể cả dấu cách
```

````python
%%writefile bai1_4_xau_kytu.cpp
#include <iostream>
#include <cstring>   // strlen, strcmp, strcat
#include <string>    // std::string
using namespace std;

int main() {
    // ====== PHẦN 1: C-String (char array) ======
    cout << "=== C-String ==" << endl;

    char s1[20] = "Hello";
    char s2[20] = "World";

    cout << "s1 = " << s1 << endl;
    cout << "Độ dài s1 = " << strlen(s1) << endl;  // 5 (không tính \0)

    // Nối chuỗi
    strcat(s1, " ");   // s1 = "Hello "
    strcat(s1, s2);    // s1 = "Hello World"
    cout << "Sau khi nối: " << s1 << endl;

    // So sánh chuỗi (strcmp trả về 0 nếu bằng nhau)
    char a[10] = "abc", b[10] = "abc";
    if (strcmp(a, b) == 0) cout << "a và b bằng nhau" << endl;

    // Duyệt từng ký tự
    cout << "Ký tự trong s2: ";
    for (int i = 0; s2[i] != '\0'; i++) {  // Dừng khi gặp ký tự null
        cout << s2[i] << " ";
    }
    cout << endl;

    // ====== PHẦN 2: std::string (C++ hiện đại) ======
    cout << "\n=== std::string ==" << endl;

    string str1 = "Xin chào";
    string str2 = " Việt Nam";

    cout << "str1 = " << str1 << endl;
    cout << "Độ dài = " << str1.length() << endl;

    // Nối chuỗi — đơn giản hơn nhiều!
    string str3 = str1 + str2;
    cout << "Sau nối: " << str3 << endl;

    // So sánh — tự nhiên như so sánh số
    if (str1 == "Xin chào") cout << "str1 khớp!" << endl;

    // Truy cập ký tự theo index
    cout << "Ký tự đầu: " << str1[0] << endl;  // 'X'

    // Cắt chuỗi con: substr(vị_trí_bắt_đầu, độ_dài)
    cout << "3 ký tự đầu: " << str1.substr(0, 3) << endl;  // "Xin"

    // Tìm vị trí chuỗi con
    size_t pos = str3.find("Việt");
    if (pos != string::npos)  // npos = không tìm thấy
        cout << "\"Việt\" tìm thấy tại index " << pos << endl;

    return 0;
}
````

## 1.5 Mảng Đánh Dấu (Marking Array / Frequency Array)

### 🔑 Bản chất — Kỹ thuật đỉnh cao

**Mảng đánh dấu** là kỹ thuật dùng **chỉ số của mảng như một cái "khóa"** để lưu thông tin về giá trị đó.

**Ý tưởng cốt lõi:**
```
Thay vì hỏi: "Số 5 có trong mảng không?"
→ Tạo mảng bool mark[]: mark[5] = true nghĩa là số 5 ĐÃ XUẤT HIỆN

Thay vì đếm bằng vòng lặp lồng nhau O(n²)
→ Dùng mảng đếm freq[]: freq[5] = số lần số 5 xuất hiện
```

### 📌 Ứng dụng thực tế
| Bài toán | Không dùng kỹ thuật | Dùng mảng đánh dấu |
|---|---|---|
| Kiểm tra số xuất hiện | Vòng lặp tuyến tính O(n) | Kiểm tra ngay O(1) |
| Đếm tần suất | Vòng lặp lồng O(n²) | Một lần duyệt O(n) |
| Tìm số bị thiếu | Phức tạp | Dễ dàng |
| Loại bỏ trùng lặp | Sắp xếp + so sánh | Đơn giản hơn |

### 🔥 Skill của dân thi lập trình
Kỹ thuật này là **nền tảng của thuật toán Counting Sort** — sắp xếp O(n) thay vì O(n log n).

**Điều kiện áp dụng:** Giá trị phần tử nằm trong khoảng đã biết (ví dụ: 0..1000).

### ⚠️ Lưu ý
- Phải biết giới hạn trên của giá trị để khai báo đủ kích thước mảng đánh dấu.
- Nhớ khởi tạo mảng về 0 trước khi dùng.

````python
%%writefile bai1_5_mang_danhday.cpp
#include <iostream>
#include <cstring>  // memset
using namespace std;

int main() {
    int a[] = {3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5};
    int n = sizeof(a) / sizeof(a[0]);  // n = 11

    cout << "Mảng gốc: ";
    for (int i = 0; i < n; i++) cout << a[i] << " ";
    cout << endl;

    // ====== KỸ THUẬT 1: Mảng đánh dấu xuất hiện ======
    // mark[x] = true nếu x xuất hiện trong mảng
    bool mark[10] = {false};  // Giả sử giá trị trong khoảng 0..9

    for (int i = 0; i < n; i++) {
        mark[a[i]] = true;  // Đánh dấu a[i] đã xuất hiện
    }

    cout << "\nCác số xuất hiện trong mảng: ";
    for (int x = 0; x <= 9; x++) {
        if (mark[x]) cout << x << " ";
    }
    cout << endl;

    // Kiểm tra số 7 có xuất hiện không — O(1)!
    cout << "Số 7 có trong mảng? " << (mark[7] ? "Có" : "Không") << endl;
    cout << "Số 5 có trong mảng? " << (mark[5] ? "Có" : "Không") << endl;

    // ====== KỸ THUẬT 2: Mảng đếm tần suất ======
    // freq[x] = số lần x xuất hiện trong mảng
    int freq[10] = {0};  // Khởi tạo tất cả = 0

    for (int i = 0; i < n; i++) {
        freq[a[i]]++;  // Tăng đếm cho giá trị a[i]
    }

    cout << "\nTần suất xuất hiện:" << endl;
    for (int x = 0; x <= 9; x++) {
        if (freq[x] > 0)
            cout << "  Số " << x << " xuất hiện " << freq[x] << " lần" << endl;
    }

    // ====== ỨNG DỤNG: Tìm số thiếu trong dãy 1..n ======
    cout << "\n--- Tìm số bị thiếu ---" << endl;
    int b[] = {1, 2, 4, 5, 6};  // Thiếu số 3
    int m = 6;  // Dãy phải có 1..6
    bool hasMark[7] = {false};

    for (int i = 0; i < 5; i++) hasMark[b[i]] = true;

    cout << "Các số bị thiếu trong 1..6: ";
    for (int x = 1; x <= m; x++) {
        if (!hasMark[x]) cout << x << " ";
    }
    cout << endl;

    return 0;
}
````

## 1.6 Chèn Và Xóa Phần Tử Mảng

### 🔑 Bản chất — Tại sao mảng khó chèn/xóa?

Vì bộ nhớ mảng **liên tiếp và cố định**, khi chèn/xóa phải **dịch chuyển các phần tử** — không thể chỉ "nhét vào" hay "kéo ra" như danh sách liên kết.

### 📌 Chèn phần tử tại vị trí k
```
Trước: [1, 2, 3, 4, 5], chèn 99 vào vị trí 2
Bước 1: Dịch phải từ cuối về vị trí k
         [1, 2, 3, 3, 4, 5]  (copy a[3]->a[4], a[2]->a[3])
Bước 2: Gán a[k] = giá_trị_mới
         [1, 2, 99, 3, 4, 5]
Bước 3: Tăng n lên 1
```

### 📌 Xóa phần tử tại vị trí k
```
Trước: [1, 2, 99, 3, 4, 5], xóa vị trí 2
Bước 1: Dịch trái từ vị trí k+1 về cuối
         [1, 2, 3, 4, 5, 5]  (copy a[3]->a[2], a[4]->a[3]...)
Bước 2: Giảm n đi 1
         [1, 2, 3, 4, 5] (ô cuối vẫn còn giá trị cũ nhưng n đã giảm nên bỏ qua)
```

### ⚠️ Lưu ý quan trọng
- **Chèn:** Dịch từ CUỐI về vị trí k (tránh ghi đè). Nếu dịch từ đầu sẽ bị mất dữ liệu.
- **Xóa:** Dịch từ vị trí k+1 về CUỐI.
- **Phải đảm bảo mảng có đủ chỗ** trước khi chèn (n < kích_thước_khai_báo).

### 🔥 Skill: Xóa nhanh (không giữ thứ tự)
Nếu không cần giữ thứ tự: hoán đổi phần tử cần xóa với phần tử cuối rồi giảm n — O(1) thay vì O(n)!

````python
%%writefile bai1_6_chen_xoa.cpp
#include <iostream>
using namespace std;

// Hàm in mảng
void inMang(int a[], int n) {
    cout << "[";
    for (int i = 0; i < n; i++) {
        cout << a[i];
        if (i < n-1) cout << ", ";
    }
    cout << "]" << endl;
}

// Chèn giá trị val vào vị trí k
void chenVaoViTri(int a[], int &n, int k, int val) {
    // Dịch phải từ cuối về vị trí k — QUAN TRỌNG: đi từ cuối về!
    for (int i = n; i > k; i--) {
        a[i] = a[i-1];
    }
    a[k] = val;  // Gán giá trị mới
    n++;         // Tăng kích thước
}

// Xóa phần tử tại vị trí k
void xoaTaiViTri(int a[], int &n, int k) {
    // Dịch trái từ vị trí k+1 về cuối
    for (int i = k; i < n-1; i++) {
        a[i] = a[i+1];
    }
    n--;  // Giảm kích thước
}

// Xóa nhanh (không giữ thứ tự) — O(1)
void xoaNhanh(int a[], int &n, int k) {
    a[k] = a[n-1];  // Ghi đè bằng phần tử cuối
    n--;            // Giảm n
}

int main() {
    int a[20] = {10, 20, 30, 40, 50};
    int n = 5;

    cout << "Mảng ban đầu:       "; inMang(a, n);

    // --- Chèn 99 vào vị trí 2 ---
    chenVaoViTri(a, n, 2, 99);
    cout << "Sau chèn 99 tại [2]: "; inMang(a, n);
    // Kết quả: [10, 20, 99, 30, 40, 50]

    // --- Chèn 0 vào đầu (vị trí 0) ---
    chenVaoViTri(a, n, 0, 0);
    cout << "Sau chèn 0 tại [0]:  "; inMang(a, n);
    // Kết quả: [0, 10, 20, 99, 30, 40, 50]

    // --- Xóa phần tử tại vị trí 3 (là 99) ---
    xoaTaiViTri(a, n, 3);
    cout << "Sau xóa tại [3]:     "; inMang(a, n);
    // Kết quả: [0, 10, 20, 30, 40, 50]

    // --- Xóa nhanh tại vị trí 1 (không giữ thứ tự) ---
    cout << "\n--- Xóa nhanh (không giữ thứ tự) ---" << endl;
    int b[] = {1, 2, 3, 4, 5};
    int m = 5;
    cout << "Trước: "; inMang(b, m);
    xoaNhanh(b, m, 1);  // Xóa vị trí 1 (giá trị 2)
    cout << "Sau xóa nhanh vị trí 1: "; inMang(b, m);
    // Kết quả: [1, 5, 3, 4] — thứ tự thay đổi nhưng nhanh hơn!

    return 0;
}
````

# 🎯 PHẦN 2: CON TRỎ (POINTER)
---

> Con trỏ là **chủ đề khó nhất và quan trọng nhất** trong C/C++. Hiểu con trỏ = hiểu bộ nhớ = viết code tốt hơn hẳn.

## 2.1 Lý Thuyết Con Trỏ

### 🔑 Bản chất cốt lõi — Con trỏ là gì?

**Con trỏ là một biến đặc biệt** — nó không lưu dữ liệu thông thường (số, ký tự...) mà lưu **địa chỉ ô nhớ** của một biến khác.

### 🧠 Phép ẩn dụ thực tế
```
Hãy nghĩ RAM như một tòa nhà chung cư:
- Mỗi căn hộ = một ô nhớ, có số phòng (địa chỉ) riêng
- Biến thường = nội dung bên trong phòng (số 42, ký tự 'A'...)
- Con trỏ = tờ giấy GHI SỐ PHÒNG của phòng khác

int x = 42;    // Phòng 1000: chứa giá trị 42
int* p = &x;   // p = tờ giấy ghi "1000" (địa chỉ của x)
*p = 99;       // Đi đến phòng 1000 và đổi nội dung thành 99
```

### 📌 3 khái niệm nền tảng

**1. Toán tử `&` (address-of)** — Lấy địa chỉ của biến
```cpp
int x = 42;
cout << &x;  // In ra địa chỉ ô nhớ của x, ví dụ: 0x7ffd1234
```

**2. Khai báo con trỏ** — Dùng dấu `*` sau kiểu dữ liệu
```cpp
int* p;      // p là con trỏ đến int
double* q;   // q là con trỏ đến double
char* r;     // r là con trỏ đến char
```

**3. Toán tử `*` (dereference / giải tham chiếu)** — Đến địa chỉ mà con trỏ trỏ đến và đọc/ghi giá trị
```cpp
int x = 42;
int* p = &x;
cout << *p;  // Đọc giá trị tại địa chỉ p trỏ đến → in ra 42
*p = 99;     // Ghi giá trị 99 vào ô nhớ mà p trỏ đến → x = 99
```

### ⚠️ Lưu ý cực kỳ quan trọng
| Lỗi nguy hiểm | Giải thích |
|---|---|
| **Con trỏ NULL** | `int* p = nullptr; *p = 5;` → Crash ngay lập tức |
| **Con trỏ dangling** | Trỏ vào biến đã bị xóa/hủy → Hành vi không xác định |
| **Con trỏ chưa khởi tạo** | `int* p; *p = 5;` → Ghi vào vùng nhớ ngẫu nhiên → rất nguy hiểm |
| **Quên giải phóng bộ nhớ** | Dùng `new` mà không `delete` → Memory leak |

### 🔥 Quy tắc vàng: Luôn khởi tạo con trỏ
```cpp
int* p = nullptr;  // Nếu chưa biết trỏ đến đâu, dùng nullptr
if (p != nullptr) *p = 5;  // Luôn kiểm tra trước khi dùng
```

````python
%%writefile bai2_1_con_tro.cpp
#include <iostream>
using namespace std;

int main() {
    // ====== PHẦN 1: Khái niệm cơ bản ======
    int x = 42;

    cout << "x = " << x << endl;           // Giá trị của x
    cout << "&x = " << &x << endl;         // Địa chỉ ô nhớ của x

    int* p = &x;  // p lưu địa chỉ của x

    cout << "p = " << p << endl;           // Giá trị của p = địa chỉ của x
    cout << "*p = " << *p << endl;         // Giải tham chiếu: đọc giá trị tại địa chỉ p

    // p và &x phải bằng nhau
    cout << "p == &x? " << (p == &x ? "Đúng" : "Sai") << endl;

    // ====== PHẦN 2: Thay đổi giá trị qua con trỏ ======
    cout << "\n--- Thay đổi x qua con trỏ ---" << endl;
    cout << "x trước khi thay đổi: " << x << endl;

    *p = 99;  // Ghi 99 vào ô nhớ mà p trỏ đến (tức là ghi vào x)

    cout << "Sau khi *p = 99:" << endl;
    cout << "  x = " << x << endl;   // x cũng thay đổi = 99!
    cout << "  *p = " << *p << endl; // = 99

    // ====== PHẦN 3: Con trỏ và phép toán ======
    cout << "\n--- Số học con trỏ ---" << endl;
    int arr[] = {10, 20, 30, 40, 50};
    int* q = arr;  // q trỏ đến đầu mảng

    cout << "*q = " << *q << endl;      // 10 (phần tử đầu)
    cout << "*(q+1) = " << *(q+1) << endl;  // 20 (cộng 1 = dịch sang phần tử tiếp theo)
    cout << "*(q+2) = " << *(q+2) << endl;  // 30

    // Tiến con trỏ
    q++;  // q tiến 1 phần tử (tiến sizeof(int) bytes)
    cout << "Sau q++, *q = " << *q << endl;  // 20

    // ====== PHẦN 4: Kích thước con trỏ ======
    cout << "\n--- Kích thước ---" << endl;
    cout << "sizeof(int) = " << sizeof(int) << " bytes" << endl;
    cout << "sizeof(int*) = " << sizeof(int*) << " bytes" << endl;   // 8 bytes (64-bit)
    cout << "sizeof(char*) = " << sizeof(char*) << " bytes" << endl; // Cũng 8 bytes!
    // → Mọi con trỏ đều có kích thước bằng nhau (8 bytes trên 64-bit)
    // vì chúng đều lưu địa chỉ, địa chỉ có kích thước cố định

    return 0;
}
````

### 📌 Con Trỏ Trong Hàm — Truyền Tham Chiếu

Đây là ứng dụng **quan trọng nhất** của con trỏ trong thực tế.

**Vấn đề:** C++ mặc định truyền tham số **theo giá trị (by value)** — hàm nhận bản sao, không thể sửa biến gốc.

**Giải pháp:** Truyền con trỏ (hoặc tham chiếu `&`) để hàm sửa được biến gốc.

```cpp
// Truyền by value — KHÔNG thay đổi được a, b gốc
void swap_wrong(int a, int b) { int t = a; a = b; b = t; }  // Vô nghĩa!

// Truyền con trỏ — Thay đổi được!
void swap_ptr(int* a, int* b) { int t = *a; *a = *b; *b = t; }

// Truyền tham chiếu (C++ style) — Gọn hơn
void swap_ref(int& a, int& b) { int t = a; a = b; b = t; }
```

````python
%%writefile bai2_1b_controtronghàm.cpp
#include <iostream>
using namespace std;

// Hàm swap bằng con trỏ
void swapPointer(int* a, int* b) {
    int temp = *a;
    *a = *b;
    *b = temp;
}

// Hàm swap bằng tham chiếu (C++ thuần)
void swapRef(int& a, int& b) {
    int temp = a;
    a = b;
    b = temp;
}

// Hàm tính min và max cùng lúc — trả 2 giá trị qua con trỏ
void minMax(int arr[], int n, int* mn, int* mx) {
    *mn = *mx = arr[0];
    for (int i = 1; i < n; i++) {
        if (arr[i] < *mn) *mn = arr[i];
        if (arr[i] > *mx) *mx = arr[i];
    }
}

int main() {
    int x = 5, y = 10;

    cout << "Trước swap: x=" << x << ", y=" << y << endl;

    swapPointer(&x, &y);  // Truyền địa chỉ
    cout << "Sau swapPointer: x=" << x << ", y=" << y << endl;

    swapRef(x, y);  // Truyền tham chiếu — đơn giản hơn
    cout << "Sau swapRef: x=" << x << ", y=" << y << endl;

    // Hàm trả về nhiều giá trị qua con trỏ
    int arr[] = {3, 1, 7, 2, 9, 4};
    int mn, mx;
    minMax(arr, 6, &mn, &mx);
    cout << "\nMin = " << mn << ", Max = " << mx << endl;

    return 0;
}
````

## 2.2 Con Trỏ Cấp 2 (Pointer to Pointer)

### 🔑 Bản chất

**Con trỏ cấp 2** là con trỏ trỏ đến... **một con trỏ khác**.

```
int x = 42;
int*  p  = &x;   // p  lưu địa chỉ của x
int** pp = &p;   // pp lưu địa chỉ của p

Chuỗi: pp → p → x
         **pp = 42 (đi 2 bước mới đến x)
```

### 📌 Sơ đồ bộ nhớ
```
Địa chỉ:   0x1000      0x2000      0x3000
Biến:      [ x=42 ]    [ p=0x1000 ] [ pp=0x2000 ]
            ↑           ↑            ↑
           x           p trỏ đến x  pp trỏ đến p
```

### 📌 Truy cập qua con trỏ cấp 2
```cpp
pp   = địa chỉ của p
*pp  = giá trị tại pp = p = địa chỉ của x
**pp = giá trị tại *pp = giá trị tại địa chỉ x = 42
```

### 🔥 Ứng dụng thực tế quan trọng
1. **Thay đổi con trỏ bên trong hàm** — khi cần hàm cấp phát bộ nhớ và trả về qua tham số.
2. **Mảng 2 chiều động** — mảng của mảng.
3. **Xử lý `argv` trong `main(int argc, char** argv)`** — danh sách chuỗi.

### ⚠️ Lưu ý
- Mỗi lần thêm `*` trong khai báo = thêm một cấp độ gián tiếp.
- Khi giải tham chiếu: `*` tương ứng với số cấp. Con trỏ cấp 2 cần `**` để đến dữ liệu gốc.

````python
%%writefile bai2_2_con_tro_cap2.cpp
#include <iostream>
using namespace std;

// Hàm nhận con trỏ cấp 2 để thay đổi con trỏ gốc
void capPhat(int** pp) {
    *pp = new int(100);  // Cấp phát bộ nhớ và gán cho *pp (tức là con trỏ gốc)
}

int main() {
    // ====== PHẦN 1: Con trỏ cấp 2 cơ bản ======
    int x = 42;
    int* p = &x;    // Con trỏ cấp 1: trỏ đến x
    int** pp = &p;  // Con trỏ cấp 2: trỏ đến p

    cout << "=== Con trỏ cấp 2 ==" << endl;
    cout << "x    = " << x    << "  (giá trị)" << endl;
    cout << "&x   = " << &x   << "  (địa chỉ của x)" << endl;
    cout << "p    = " << p    << "  (p = địa chỉ của x)" << endl;
    cout << "*p   = " << *p   << "  (giá trị tại địa chỉ p = x)" << endl;
    cout << "&p   = " << &p   << "  (địa chỉ của p)" << endl;
    cout << "pp   = " << pp   << "  (pp = địa chỉ của p)" << endl;
    cout << "*pp  = " << *pp  << "  (*pp = p = địa chỉ của x)" << endl;
    cout << "**pp = " << **pp << "  (**pp = x = 42)" << endl;

    // ====== PHẦN 2: Thay đổi x qua pp ======
    **pp = 999;  // Đi 2 cấp để đến x, gán 999
    cout << "\nSau **pp = 999:" << endl;
    cout << "x = " << x << "  (x đã thay đổi)" << endl;

    // ====== PHẦN 3: Ứng dụng — cấp phát qua hàm ======
    cout << "\n=== Ứng dụng thực tế ==" << endl;
    int* myPtr = nullptr;
    cout << "myPtr trước hàm: " << myPtr << endl;  // nullptr = 0

    capPhat(&myPtr);  // Truyền địa chỉ của con trỏ
    cout << "myPtr sau hàm: " << myPtr << endl;  // Đã có địa chỉ hợp lệ
    cout << "*myPtr = " << *myPtr << endl;       // 100

    delete myPtr;  // Giải phóng bộ nhớ

    // ====== PHẦN 4: Con trỏ cấp 2 với mảng chuỗi ======
    // Mô phỏng argv (danh sách chuỗi)
    cout << "\n=== Mảng chuỗi (char**) ==" << endl;
    const char* names[] = {"Alice", "Bob", "Charlie"};
    const char** ptr = names;  // Con trỏ cấp 2 trỏ đến mảng chuỗi

    for (int i = 0; i < 3; i++) {
        cout << "names[" << i << "] = " << ptr[i] << endl;
    }

    return 0;
}
````

## 2.3 Con Trỏ Và Mảng 1 Chiều

### 🔑 Bản chất — Mối quan hệ cốt lõi

> **"Tên mảng là con trỏ hằng trỏ đến phần tử đầu tiên của mảng"**

Đây là điều **quan trọng nhất** cần hiểu:
```cpp
int a[5] = {10, 20, 30, 40, 50};
//  a == &a[0]  → Tên mảng = địa chỉ phần tử đầu tiên
```

### 📌 Hai cách truy cập tương đương 100%
```cpp
int a[5] = {10, 20, 30, 40, 50};
int* p = a;

// Dùng subscript notation (cách thông thường)
a[0]  a[1]  a[2]  a[3]  a[4]

// Dùng pointer arithmetic (cách con trỏ)
*(p+0) *(p+1) *(p+2) *(p+3) *(p+4)

// THẬM CHÍ có thể viết lộn xộn:
p[2] == *(p+2) == a[2] == *(a+2)  // Tất cả đều là 30!
```

### 📌 Số học con trỏ (Pointer Arithmetic)
```
p + 1  không phải cộng thêm 1 byte
       mà cộng thêm sizeof(kiểu_dữ_liệu) bytes

int* p:    p + 1 = p + 4 bytes  (sizeof(int) = 4)
double* q: q + 1 = q + 8 bytes  (sizeof(double) = 8)
char* r:   r + 1 = r + 1 byte   (sizeof(char) = 1)
```

### 📌 Điểm khác biệt quan trọng giữa mảng và con trỏ
| | Tên mảng `a` | Con trỏ `p` |
|---|---|---|
| Có thể thay đổi địa chỉ | ❌ Không (`a` là hằng) | ✅ Có (`p = &x`) |
| `sizeof` | Kích thước toàn bộ mảng | Kích thước con trỏ (8 bytes) |
| Có thể `p++` | ❌ `a++` là lỗi | ✅ Được |

### 🔥 Skill: Duyệt mảng bằng con trỏ (style lập trình hệ thống)
```cpp
for (int* p = a; p < a + n; p++) {
    cout << *p << " ";
}
```

````python
%%writefile bai2_3_contro_mang.cpp
#include <iostream>
using namespace std;

// Hàm nhận mảng qua con trỏ
void inMang(int* arr, int n) {
    cout << "[";
    for (int i = 0; i < n; i++) {
        cout << arr[i];  // arr[i] == *(arr+i)
        if (i < n-1) cout << ", ";
    }
    cout << "]";
}

// Hàm tổng dùng pointer arithmetic
int tongMang(int* arr, int n) {
    int sum = 0;
    int* end = arr + n;   // Con trỏ đến sau phần tử cuối
    for (int* p = arr; p < end; p++) {
        sum += *p;  // Đọc giá trị tại vị trí con trỏ
    }
    return sum;
}

int main() {
    int a[5] = {10, 20, 30, 40, 50};
    int n = 5;

    // ====== Mảng và con trỏ là như nhau ======
    cout << "=== Mảng vs Con Trỏ ==" << endl;
    cout << "a    = " << a    << "  (địa chỉ đầu mảng)" << endl;
    cout << "&a[0]= " << &a[0] << "  (địa chỉ a[0])" << endl;
    cout << "Bằng nhau? " << (a == &a[0] ? "Đúng" : "Sai") << endl;

    int* p = a;  // p trỏ vào đầu mảng

    cout << "\n=== 4 cách truy cập phần tử thứ 2 (index 2) ==" << endl;
    cout << "a[2]   = " << a[2]    << endl;  // Cách thông thường
    cout << "*(a+2) = " << *(a+2)  << endl;  // Pointer arithmetic trên tên mảng
    cout << "p[2]   = " << p[2]    << endl;  // Con trỏ dùng subscript
    cout << "*(p+2) = " << *(p+2)  << endl;  // Pointer arithmetic trên con trỏ
    // Tất cả đều in ra 30

    // ====== Duyệt mảng bằng con trỏ ======
    cout << "\n=== Duyệt bằng con trỏ ==" << endl;
    cout << "Cách 1 (index): ";
    for (int i = 0; i < n; i++) cout << a[i] << " ";
    cout << endl;

    cout << "Cách 2 (pointer): ";
    for (int* q = a; q < a + n; q++) {  // Điều kiện: q chưa vượt qua cuối
        cout << *q << " ";
    }
    cout << endl;

    // ====== Truyền mảng vào hàm ======
    cout << "\n=== Truyền mảng vào hàm ==" << endl;
    cout << "Mảng: ";
    inMang(a, n);  // a tự động chuyển thành con trỏ
    cout << endl;
    cout << "Tổng = " << tongMang(a, n) << endl;

    // ====== sizeof khác nhau! ======
    cout << "\n=== sizeof ==" << endl;
    cout << "sizeof(a) = " << sizeof(a) << " bytes  (toàn bộ mảng = 5*4)" << endl;
    cout << "sizeof(p) = " << sizeof(p) << " bytes  (chỉ là địa chỉ = 8 bytes)" << endl;

    return 0;
}
````

## 2.4 Cấp Phát Mảng Động (Dynamic Memory Allocation)

### 🔑 Bản chất — Tại sao cần cấp phát động?

**Bài toán:** Kích thước mảng chỉ biết lúc chạy chương trình, không biết lúc viết code.

```cpp
// Vấn đề: Khai báo tĩnh cần biết n trước
int n;
cin >> n;
int a[n];  // ⚠️ VLA — không chuẩn C++, tránh dùng!

// Giải pháp: Cấp phát động
int* a = new int[n];  // ✅ Chuẩn C++, an toàn
```

### 📌 Cú pháp cấp phát và giải phóng

**Cấp phát 1 biến:**
```cpp
int* p = new int;       // Cấp phát 1 int (không khởi tạo)
int* p = new int(42);   // Cấp phát và khởi tạo = 42
delete p;               // Giải phóng 1 biến
p = nullptr;            // Quan trọng: reset về nullptr sau delete
```

**Cấp phát mảng:**
```cpp
int* a = new int[n];    // Cấp phát mảng n phần tử
int* a = new int[n]();  // Cấp phát + khởi tạo tất cả = 0
delete[] a;             // Giải phóng mảng (phải dùng delete[])
a = nullptr;
```

### 📌 Stack vs Heap — Sự khác biệt
| | Stack | Heap |
|---|---|---|
| Khai báo tĩnh | `int a[100]` | `int* a = new int[100]` |
| Quản lý | Tự động (hàm trả về = giải phóng) | Thủ công (`delete[]`) |
| Kích thước | Nhỏ (~1-8 MB) | Lớn (vài GB) |
| Tốc độ | Nhanh hơn | Chậm hơn một chút |
| Lifetime | Trong phạm vi khai báo | Tồn tại đến khi `delete` |

### ⚠️ Các lỗi nguy hiểm nhất
```cpp
// 1. Quên delete → Memory leak (rò rỉ bộ nhớ)
int* p = new int[1000];
// ... quên delete[] p → mỗi lần gọi = 4KB bị mất!

// 2. Delete 2 lần → Double free → crash
delete[] p;
delete[] p;  // NGUY HIỂM!

// 3. Nhầm delete với delete[]
int* arr = new int[10];
delete arr;    // SAI! Phải dùng delete[]
delete[] arr;  // ĐÚNG

// 4. Dùng sau khi delete → Dangling pointer
delete[] p;
cout << p[0];  // NGUY HIỂM! p là dangling pointer
```

### 🔥 Quy tắc vàng: RAII Pattern
Mọi `new` đều phải có `delete` tương ứng. Kiểm tra trước khi đóng hàm.

````python
%%writefile bai2_4_cap_phat_dong.cpp
#include <iostream>
#include <algorithm>  // sort
using namespace std;

int main() {
    // ====== PHẦN 1: Cấp phát 1 biến ======
    cout << "=== Cấp phát 1 biến ==" << endl;

    int* p = new int(42);  // Cấp phát 1 int, khởi tạo = 42
    cout << "*p = " << *p << endl;

    *p = 99;  // Sửa giá trị
    cout << "Sau khi sửa: *p = " << *p << endl;

    delete p;     // Giải phóng bộ nhớ
    p = nullptr;  // Reset về nullptr (quan trọng!)
    cout << "Đã giải phóng, p = " << p << endl;

    // ====== PHẦN 2: Cấp phát mảng động ======
    cout << "\n=== Cấp phát mảng động ==" << endl;

    int n = 5;  // Trong thực tế: cin >> n;
    int* arr = new int[n]();  // () → khởi tạo tất cả = 0

    // Gán giá trị
    for (int i = 0; i < n; i++) {
        arr[i] = (i + 1) * 10;  // 10, 20, 30, 40, 50
    }

    cout << "Mảng: ";
    for (int i = 0; i < n; i++) cout << arr[i] << " ";
    cout << endl;

    // Sắp xếp mảng động — dùng sort bình thường
    sort(arr, arr + n, greater<int>());  // Sắp xếp giảm dần
    cout << "Sau sắp xếp giảm: ";
    for (int i = 0; i < n; i++) cout << arr[i] << " ";
    cout << endl;

    delete[] arr;   // Giải phóng mảng — PHẢI dùng delete[]
    arr = nullptr;

    // ====== PHẦN 3: Thay đổi kích thước mảng (realloc kiểu C++) ======
    cout << "\n=== Mở rộng mảng (resize) ==" << endl;

    int oldN = 3;
    int* a = new int[oldN]{1, 2, 3};

    cout << "Mảng ban đầu (n=3): ";
    for (int i = 0; i < oldN; i++) cout << a[i] << " ";
    cout << endl;

    // Cần mở rộng lên 5 phần tử
    int newN = 5;
    int* b = new int[newN]();  // Cấp phát mảng mới lớn hơn

    for (int i = 0; i < oldN; i++) b[i] = a[i];  // Sao chép dữ liệu cũ
    b[3] = 4;  // Thêm dữ liệu mới
    b[4] = 5;

    delete[] a;  // Giải phóng mảng cũ
    a = b;       // a giờ trỏ đến mảng mới
    b = nullptr; // Chỉ có 1 con trỏ quản lý

    cout << "Mảng sau khi mở rộng (n=5): ";
    for (int i = 0; i < newN; i++) cout << a[i] << " ";
    cout << endl;

    delete[] a;  // Giải phóng cuối cùng
    a = nullptr;

    // ====== PHẦN 4: Kiểm tra cấp phát thất bại ======
    cout << "\n=== An toàn khi cấp phát ==" << endl;
    try {
        int* safe = new int[1000];  // Cấp phát nhỏ, luôn thành công
        safe[0] = 42;
        cout << "Cấp phát thành công, safe[0] = " << safe[0] << endl;
        delete[] safe;
        safe = nullptr;
    } catch (bad_alloc& e) {
        cout << "Cấp phát thất bại: " << e.what() << endl;
    }

    cout << "\nTất cả bộ nhớ đã được giải phóng sạch!" << endl;
    return 0;
}
````

### 🗂️ PHẦN 1: MẢNG

| Chủ đề | Cú pháp / Công thức | Lưu ý |
|---|---|---|
| Khai báo | `int a[N];` | Index: 0 đến N-1 |
| Khởi tạo | `int a[5] = {1,2,3,4,5};` | Phần dư tự động = 0 |
| Số phần tử | `sizeof(a)/sizeof(a[0])` | Chỉ dùng được trong scope khai báo |
| Sort tăng | `sort(a, a+n)` | Cần `#include <algorithm>` |
| Sort giảm | `sort(a, a+n, greater<int>())` | - |
| Mảng đánh dấu | `bool mark[MAX] = {false}; mark[x] = true;` | Kiểm tra O(1) |
| Mảng đếm | `int freq[MAX] = {0}; freq[x]++;` | Đếm tần suất O(n) |
| Chèn tại k | Dịch phải từ cuối về k, rồi `a[k]=val; n++` | Đi từ CUỐI về |
| Xóa tại k | Dịch trái từ k+1, rồi `n--` | - |
| Xóa nhanh | `a[k] = a[n-1]; n--` | Mất thứ tự, O(1) |

### 🗂️ PHẦN 2: CON TRỎ

| Chủ đề | Cú pháp / Công thức | Lưu ý |
|---|---|---|
| Khai báo | `int* p;` | Phải khởi tạo trước khi dùng |
| Lấy địa chỉ | `p = &x;` | - |
| Giải tham chiếu | `*p` | Đọc/ghi giá trị tại địa chỉ |
| nullptr | `int* p = nullptr;` | Luôn kiểm tra trước khi dùng |
| Con trỏ cấp 2 | `int** pp = &p;` | `**pp` để đến giá trị gốc |
| Mảng = con trỏ | `a == &a[0]` | `a[i] == *(a+i)` |
| Pointer arithmetic | `p+1` = tiến `sizeof(type)` bytes | - |
| Duyệt bằng con trỏ | `for(int* p=a; p<a+n; p++)` | Style C thuần |
| Cấp phát | `int* p = new int[n]();` | `()` để khởi tạo = 0 |
| Giải phóng | `delete[] p; p = nullptr;` | PHẢI có sau mỗi new[] |
| Truyền hàm | `void f(int* p)` hoặc `void f(int& x)` | Sửa được biến gốc |

### ⚡ CHECKLIST TRƯỚC KHI NỘP BÀI
- [ ] Index mảng không vượt quá `n-1`
- [ ] Mọi con trỏ đều được khởi tạo (không để hoang)
- [ ] Mọi `new` đều có `delete` (hoặc `new[]` có `delete[]`) tương ứng
- [ ] Sau `delete`: reset con trỏ về `nullptr`
- [ ] Khi chèn mảng: dịch từ CUỐI về trước, không dịch từ đầu
- [ ] Mảng đánh dấu: phải khởi tạo về 0/false trước
- [ ] String: dùng `getline` khi input có khoảng trắng

## 🎁 BONUS: Bài Tập Tổng Hợp — Áp Dụng Tất Cả

Bài tập này kết hợp **mảng động + con trỏ + mảng đánh dấu** trong một bài hoàn chỉnh.

````python
%%writefile bonus_tong_hop.cpp
/*
 * BÀI TẬP TỔNG HỢP:
 * Nhập n số nguyên, thực hiện:
 * 1. In mảng gốc
 * 2. Tìm và in các số xuất hiện nhiều hơn 1 lần
 * 3. Xóa các phần tử trùng lặp (giữ lần xuất hiện đầu tiên)
 * 4. In mảng sau khi xóa
 */
#include <iostream>
using namespace std;

void inMang(int* a, int n) {
    for (int i = 0; i < n; i++) cout << a[i] << (i < n-1 ? " " : "\n");
}

int main() {
    // Dữ liệu cố định để demo
    int data[] = {3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5};
    int n = sizeof(data) / sizeof(data[0]);

    // Cấp phát mảng động sao chép dữ liệu
    int* a = new int[n];
    for (int i = 0; i < n; i++) a[i] = data[i];

    cout << "Mảng gốc: ";
    inMang(a, n);

    // Giả sử giá trị trong [0, 10]
    const int MAX_VAL = 11;
    int* freq = new int[MAX_VAL]();  // Mảng đếm tần suất, khởi tạo = 0

    // Đếm tần suất
    for (int i = 0; i < n; i++) freq[a[i]]++;

    // Tìm các số xuất hiện > 1 lần
    cout << "Số xuất hiện nhiều hơn 1 lần: ";
    for (int x = 0; x < MAX_VAL; x++) {
        if (freq[x] > 1) cout << x << "(" << freq[x] << " lần) ";
    }
    cout << endl;

    // Xóa trùng lặp — dùng mảng đánh dấu
    bool* seen = new bool[MAX_VAL]();  // seen[x] = đã thấy x chưa
    int newN = 0;

    for (int i = 0; i < n; i++) {
        if (!seen[a[i]]) {       // Nếu chưa thấy giá trị này
            a[newN++] = a[i];    // Giữ lại
            seen[a[i]] = true;   // Đánh dấu đã thấy
        }
        // Nếu đã thấy rồi: bỏ qua (không chép vào)
    }

    cout << "Mảng sau khi xóa trùng: ";
    inMang(a, newN);

    // Giải phóng tất cả
    delete[] a;    a = nullptr;
    delete[] freq; freq = nullptr;
    delete[] seen; seen = nullptr;

    cout << "Hoàn thành! Bộ nhớ đã giải phóng." << endl;
    return 0;
}
````

## 💡 LỜI KHUYÊN CUỐI — TỪ NGƯỜI LẬP TRÌNH KINH NGHIỆM

### 🧠 Về tư duy
1. **Hình dung RAM như mảng ô nhớ có địa chỉ** — mọi thứ (biến, mảng, con trỏ) đều là ô nhớ.
2. **Con trỏ chỉ là địa chỉ** — đừng sợ, hãy vẽ sơ đồ ra giấy khi rối.
3. **Mảng và con trỏ gần như tương đương** — biết một là biết cả hai.

### 🔥 Các "skill" đỉnh nhất cần nhớ
1. `sizeof(a)/sizeof(a[0])` — đếm phần tử mảng không cần nhớ kích thước.
2. Mảng đánh dấu/đếm tần suất — chuyển bài toán O(n²) thành O(n).
3. Truyền con trỏ vào hàm — thay đổi biến gốc từ hàm con.
4. Xóa nhanh O(1) — hoán đổi với cuối khi không cần giữ thứ tự.
5. `new int[n]()` (có dấu `()`) — khởi tạo mảng về 0 ngay khi cấp phát.

### ⚠️ Lỗi hay gặp nhất trong kỳ thi
1. Quên `delete[]` — gây memory leak (không fail test nhưng trừ điểm).
2. Dùng `a[n]` thay vì `a[n-1]` — out of bounds.
3. Chèn mảng dịch từ đầu thay vì từ cuối — ghi đè mất dữ liệu.
4. Nhầm `delete` và `delete[]`.
5. Không reset con trỏ về `nullptr` sau khi delete.
