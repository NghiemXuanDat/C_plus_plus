# PHẦN I: NỀN TẢNG OOP

## I.1. Khái niệm Lập trình Hướng đối tượng (OOP)

### 🎯 Định nghĩa chính xác

**OOP (Object-Oriented Programming)** là một **mô hình lập trình** (programming paradigm) tổ chức code xung quanh **đối tượng** (objects) — những thực thể kết hợp **dữ liệu + hành vi** — thay vì tổ chức xung quanh **hàm và logic tuần tự**.

### 🧩 Hai thành phần cốt lõi của một Object

| Thành phần | Thuật ngữ kỹ thuật | Ý nghĩa |
|-----------|-------------------|---------|
| **Thuộc tính** (Attributes/Properties) | **Data Members** | Dữ liệu mô tả trạng thái của đối tượng |
| **Hành vi** (Behaviors/Actions) | **Member Functions / Methods** | Những thao tác đối tượng có thể thực hiện |

### 🌍 Tư duy "nhìn thế giới theo đối tượng"

```
Thế giới thực          →      OOP
─────────────────────────────────────────────
Con người              →  class Person
  - tên, tuổi             - name, age
  - đi, nói, ăn            - walk(), talk(), eat()

Tài khoản ngân hàng   →  class BankAccount
  - số dư, chủ TK          - balance, ownerName
  - nạp, rút, kiểm tra     - deposit(), withdraw(), check()

Xe hơi                →  class Car
  - màu, tốc độ, nhiên liệu - color, speed, fuel
  - chạy, phanh, còi        - drive(), brake(), horn()
```

### 🏗️ Tại sao OOP ra đời?

Trước OOP, lập trình **procedural** (thủ tục) gặp vấn đề nghiêm trọng với phần mềm lớn:

| Vấn đề | Mô tả |
|--------|-------|
| **Khó bảo trì** | Dữ liệu toàn cục — ai cũng sửa được, khó theo dõi |
| **Khó tái sử dụng** | Code viết cho project A khó dùng lại cho project B |
| **Khó mở rộng** | Thêm tính năng mới → phải sửa rải rác ở nhiều hàm |
| **Coupling cao** | Hàm và dữ liệu phụ thuộc lẫn nhau chằng chịt |

**OOP giải quyết bằng cách**: Đóng gói dữ liệu + hành vi vào một đơn vị độc lập (class), kiểm soát quyền truy cập, và cho phép tái sử dụng qua kế thừa.

## I.2. So sánh OOP và Lập trình Thủ tục (Procedural Programming)

### 📊 Bảng so sánh tổng quan

| Tiêu chí | Procedural | OOP |
|----------|-----------|-----|
| **Trọng tâm** | Hàm (functions) | Đối tượng (objects) |
| **Đơn vị cơ bản** | Function | Class / Object |
| **Dữ liệu** | Toàn cục hoặc truyền qua tham số | Đóng gói bên trong object |
| **Luồng xử lý** | Top-down, tuần tự | Tương tác giữa các objects |
| **Tái sử dụng** | Copy code hoặc include | Kế thừa (Inheritance) |
| **Kiểm soát truy cập** | Không có | Có (private/public/protected) |
| **Ngôn ngữ tiêu biểu** | C, Pascal, BASIC | C++, Java, Python, C# |

### 🔑 Điểm khác biệt cốt lõi nhất

> **Procedural**: Hàm và dữ liệu **tách rời** nhau → Hàm "đứng ngoài" xử lý dữ liệu.
>
> **OOP**: Hàm và dữ liệu **sống chung** trong class → Object **tự biết** cách xử lý dữ liệu của mình.

````python
%%writefile compare_paradigms.cpp
#include <iostream>
#include <string>
using namespace std;

// ========================================================
// CACH 1: PROCEDURAL - Du lieu va ham tach roi nhau
// ========================================================
// Van de: Bat ky ham nao cung co the doc/sua truc tiep du lieu
struct StudentData {
    string name;
    float mathScore;
    float literatureScore;
};

// Ham nam o ngoai - không co su bao ve
float averageProc(StudentData s) {
    return (s.mathScore + s.literatureScore) / 2.0f;
}

void printStudentProc(StudentData s) {
    cout << s.name << " | TB: " << averageProc(s) << endl;
}

bool passedProc(StudentData s) {
    return averageProc(s) >= 5.0f;
}

// ========================================================
// CACH 2: OOP - Du lieu + hanh vi goi gon trong class
// ========================================================
class Student {
private:
    // Du lieu duoc bao ve, khong ai sua truc tiep duoc
    string name;
    float mathScore;
    float literatureScore;

public:
    // Constructor: khoi tao du lieu
    Student(string n, float m, float l) : name(n), mathScore(m), literatureScore(l) {}

    // Hanh vi gan lien voi object - object tu xu ly du lieu cua chinh no
    float average() const {
        return (mathScore + literatureScore) / 2.0f;
    }

    void printInfo() const {
        cout << name << " | TB: " << average() << endl;
    }

    bool passed() const {
        return average() >= 5.0f;
    }

    string getGrade() const {
        float avg = average();
        if (avg >= 9.0f)  return "Xuat sac";
        if (avg >= 8.0f)  return "Gioi";
        if (avg >= 7.0f)  return "Kha";
        if (avg >= 5.0f)  return "Trung binh";
        return "Yeu";
    }
};

int main() {
    cout << "==========================================" << endl;
    cout << "   SO SANH PROCEDURAL vs OOP   " << endl;
    cout << "==========================================" << endl;

    cout << "
[PROCEDURAL] - Ham va du lieu tach roi:" << endl;
    StudentData s1 = {"Nguyen Van A", 7.5f, 6.5f};
    printStudentProc(s1);
    cout << "Ket qua: " << (passedProc(s1) ? "DAU" : "TRUOT") << endl;
    // Van de: Co the sua truc tiep - khong kiem soat duoc!
    s1.mathScore = -999.0f;  // Gia tri phi ly - khong ai ngan duoc!
    cout << "Sau khi sua truc tiep: " << averageProc(s1) << " (sai!)" << endl;

    cout << "
[OOP] - Du lieu duoc bao ve, object tu xu ly:" << endl;
    Student st1("Tran Thi B", 8.0f, 9.0f);
    st1.printInfo();
    cout << "Xep loai: " << st1.getGrade() << endl;
    cout << "Ket qua: " << (st1.passed() ? "DAU" : "TRUOT") << endl;
    // st1.mathScore = -999;  // LOI BIEN DICH - khong the sua truc tiep!
    cout << "Du lieu an toan - khong the sua truc tiep tu ngoai" << endl;

    return 0;
}
````

## I.3. Các Nguyên lý Cơ bản của OOP – Bốn Trụ cột (EIPA)

> 🏛️ **Viết tắt dễ nhớ**: **E**ncapsulation – **I**nheritance – **P**olymorphism – **A**bstraction

---

### 🔒 1. Encapsulation – Đóng gói

**Định nghĩa**: Gói **dữ liệu + hành vi** vào cùng một đơn vị (class) và **kiểm soát quyền truy cập** từ bên ngoài.

**Cốt lõi**: Ẩn chi tiết nội bộ, chỉ lộ ra "giao diện" cần thiết.

**Lợi ích thực tế**:
- Bảo vệ dữ liệu khỏi bị thay đổi sai cách
- Cho phép thay đổi cách lưu trữ nội bộ mà **không ảnh hưởng** code bên ngoài
- Dễ debug: biết chắc dữ liệu chỉ thay đổi qua getter/setter

**Ví dụ thực tế**: Remote TV — bạn chỉ biết nút bấm (interface), không cần biết mạch điện bên trong (implementation).

---

### 🧬 2. Inheritance – Kế thừa

**Định nghĩa**: Lớp con (**derived class**) thừa hưởng thuộc tính + hành vi của lớp cha (**base class**), có thể thêm mới hoặc ghi đè.

**Cốt lõi**: Mô hình quan hệ "**IS-A**" — Dog IS-A Animal, Circle IS-A Shape.

**Lợi ích**: Tái sử dụng code, tránh trùng lặp, tạo hệ thống phân cấp tự nhiên.

**Ví dụ**: `class Dog : public Animal` — Dog kế thừa tất cả từ Animal, thêm các đặc tính riêng của chó.

---

### 🎭 3. Polymorphism – Đa hình

**Định nghĩa**: Cùng một **giao diện (tên hàm)** nhưng cho **nhiều hành vi khác nhau** tùy theo loại đối tượng.

**Cốt lõi**: *"Một tên, nhiều hành vi"*

**Hai loại**:
- **Compile-time** (Static): Function overloading, Operator overloading
- **Runtime** (Dynamic): Virtual functions — đây là đa hình "thực sự" của OOP

**Ví dụ**: Gọi `sound()` → Dog kêu "Gâu!", Cat kêu "Meo!", Bird kêu "Ríu rít!" — cùng lệnh gọi, kết quả khác nhau.

---

### 🎨 4. Abstraction – Trừu tượng hóa

**Định nghĩa**: **Ẩn đi sự phức tạp**, chỉ lộ ra những gì người dùng cần biết và tương tác.

**Cốt lõi**: Tập trung vào **"WHAT"** (làm gì), không quan tâm **"HOW"** (làm thế nào nội bộ).

**Ví dụ**: Gọi `sort()` — bạn chỉ cần biết nó sắp xếp, không cần biết thuật toán Quick Sort hay Merge Sort bên trong.

---

> 📌 **Phân biệt Abstraction vs Encapsulation**:
> - **Abstraction** = ẩn **sự phức tạp** ở cấp độ **thiết kế** (design)
> - **Encapsulation** = ẩn **dữ liệu nội bộ** ở cấp độ **triển khai** (implementation)
> - Abstraction là **ý tưởng**, Encapsulation là **cơ chế thực hiện** ý tưởng đó.

````python
%%writefile four_pillars.cpp
#include <iostream>
#include <string>
#include <vector>
using namespace std;

// ============================================================
// MINH HOA 4 NGUYEN LY OOP
// Su dung he thong quan ly Dong vat trong soo thu
// ============================================================

// ======= ABSTRACTION: Lop truu tuong =======
// Dinh nghia "giao dien" ma moi dong vat PHAI co
// Khach hang chi can biet: dong vat co the speak() va describe()
// Khong can biet CHI TIET ben trong
class Animal {
protected:
    string name;
    int age;

public:
    Animal(string n, int a) : name(n), age(a) {}

    // Pure virtual = bat buoc lop con phai override
    // Day la ABSTRACTION: dat ra giao uoc "moi animal phai biet noi"
    virtual string speak() const = 0;
    virtual string getType() const = 0;

    // Giao dien chung: ai cung goi duoc, khong biet ben trong lam gi
    void introduce() const {
        cout << "  [" << getType() << "] " << name
             << " (" << age << " tuoi): " << speak() << endl;
    }

    string getName() const { return name; }
    int getAge() const { return age; }

    virtual ~Animal() {}
};

// ======= INHERITANCE: Ke thua =======
// Dog IS-A Animal (quan he "la mot loai")
class Dog : public Animal {
private:
    string breed;  // Thuoc tinh rieng cua Dog - khong co trong Animal

public:
    // ENCAPSULATION: Du lieu (breed, name, age) bi an
    // Chi tiep can qua constructor va getter
    Dog(string n, int a, string b) : Animal(n, a), breed(b) {}

    // POLYMORPHISM: Ghi de hanh vi speak() theo cach cua Dog
    string speak() const override { return "Gau! Gau! Gau!"; }
    string getType() const override { return "Cho"; }
    string getBreed() const { return breed; }
};

// Cat IS-A Animal
class Cat : public Animal {
private:
    bool isIndoor;

public:
    Cat(string n, int a, bool indoor) : Animal(n, a), isIndoor(indoor) {}

    string speak() const override { return "Meo~ Meoooo~"; }
    string getType() const override { return "Meo"; }
    string getHabitat() const { return isIndoor ? "Trong nha" : "Ngoai troi"; }
};

// Bird IS-A Animal
class Bird : public Animal {
private:
    double wingspan;

public:
    Bird(string n, int a, double ws) : Animal(n, a), wingspan(ws) {}

    string speak() const override { return "Riu rit! Tip tip!"; }
    string getType() const override { return "Chim"; }
};

// ============================================================
// Ham nhan POINTER TO BASE CLASS - loi the cua Polymorphism
// Khong can biet la Dog, Cat hay Bird - cu goi introduce() la xong
void makeAllSpeak(const vector<Animal*>& zoo) {
    cout << "  --- Tat ca dong vat trong soo thu len tieng ---" << endl;
    for (const Animal* a : zoo) {
        a->introduce();  // POLYMORPHISM: moi animal noi theo cach rieng
    }
}

int main() {
    cout << "=== MINH HOA 4 NGUYEN LY OOP ===" << endl;

    // Tao cac doi tuong
    Dog dog1("Rex", 3, "German Shepherd");
    Cat cat1("Mimi", 2, true);
    Bird bird1("Tweety", 1, 0.3);
    Dog dog2("Lucky", 5, "Golden Retriever");

    // ===== 1. ENCAPSULATION =====
    cout << "
[1] ENCAPSULATION - Du lieu duoc bao ve:" << endl;
    cout << "  Ten: " << dog1.getName() << endl;   // Dung getter
    cout << "  Giong: " << dog1.getBreed() << endl; // Dung getter
    // dog1.name = "Hack"; // LOI BIEN DICH - name la protected!
    cout << "  (Khong the sua truc tiep dog1.name tu ben ngoai)" << endl;

    // ===== 2. INHERITANCE =====
    cout << "
[2] INHERITANCE - Ke thua Animal:" << endl;
    cout << "  Dog co introduce() tu Animal: ";
    dog1.introduce();  // Method nay duoc ke thua tu Animal
    cout << "  Cat co introduce() tu Animal: ";
    cat1.introduce();
    cout << "  Dog them getBreed() rieng: " << dog1.getBreed() << endl;

    // ===== 3. POLYMORPHISM =====
    cout << "
[3] POLYMORPHISM - Cung goi introduce(), hanh vi khac nhau:" << endl;
    vector<Animal*> zoo = {&dog1, &cat1, &bird1, &dog2};
    makeAllSpeak(zoo);

    // ===== 4. ABSTRACTION =====
    cout << "
[4] ABSTRACTION - An di chi tiet, chi lo giao dien:" << endl;
    cout << "  Ham makeAllSpeak() khong biet loai cu the," << endl;
    cout << "  chi biet chung la Animal va co the introduce()" << endl;
    cout << "  -> Day la suc manh cua ABSTRACTION!" << endl;

    return 0;
}
````

# PHẦN II: LỚP VÀ ĐỐI TƯỢNG (CLASS & OBJECT)

## II.1 & II.2. Khái niệm Class và Object

### 📐 Class (Lớp) — "Bản thiết kế"

**Class** là một **khuôn mẫu** (blueprint/template) định nghĩa:
- **Kiểu dữ liệu** mà object sẽ lưu trữ (data members)
- **Hành vi** mà object có thể thực hiện (member functions)

> 💡 Class chỉ là **định nghĩa** — không tạo ra đối tượng thực, không chiếm bộ nhớ cho dữ liệu.

### 🚗 Object (Đối tượng) — "Thực thể cụ thể"

**Object** là một **thể hiện** (instance) cụ thể được tạo ra từ class:
- Mỗi object có **bộ nhớ riêng** cho dữ liệu của nó
- Nhiều objects có thể được tạo từ cùng một class
- Mỗi object có **trạng thái riêng** (data khác nhau)

```
     CLASS Car                    OBJECTS
  ┌─────────────┐         ┌─────────────────────────────┐
  │ Khuôn mẫu  │ ──────→  │ myCar: color="đỏ", speed=0 │
  │  color      │         ├─────────────────────────────┤
  │  speed      │ ──────→  │ yourCar: color="xanh",speed=0│
  │  drive()    │         └─────────────────────────────┘
  └─────────────┘
        ↑
   1 bản thiết kế → nhiều ngôi nhà thực tế
```

## II.7. Từ khóa `this`

**`this`** là **con trỏ ẩn** (implicit pointer) trong mỗi non-static member function, luôn trỏ đến **object đang gọi hàm** đó.

**Ba trường hợp dùng `this`:**

```cpp
class Point {
    int x, y;
public:
    // 1. Phân biệt tham số trùng tên với data member
    void setXY(int x, int y) {
        this->x = x;  // this->x = data member; x bên phải = tham số
        this->y = y;
    }

    // 2. Method chaining — trả về *this để gọi tiếp
    Point& moveX(int dx) { x += dx; return *this; }
    Point& moveY(int dy) { y += dy; return *this; }

    // 3. Truyền object hiện tại vào hàm khác
    void registerAt(SomeManager& m) { m.add(this); }
};

// Method chaining nhờ return *this
point.moveX(5).moveY(3).moveX(2);  // Gọi liên tiếp!
```

## II.8. Phạm vi Thành viên Lớp (Class Scope)

- Bên **trong** class: Các thành viên có thể dùng nhau tự do, bất kể thứ tự khai báo
- Bên **ngoài** class: Phải dùng tên lớp (hoặc object) để truy cập

```cpp
class MyClass {
    void funcA() { funcB(); }  // OK! funcB có thể khai báo sau funcA
    void funcB() { /* ... */ }
};

// Bên ngoài:
MyClass obj;
obj.publicMethod();        // Qua object
MyClass::staticMethod();   // Qua tên lớp (cho static)
```

````python
%%writefile class_object_full.cpp
#include <iostream>
#include <string>
using namespace std;

// ============================================================
// DEMO TOAN DIEN: CLASS, OBJECT, THIS, SCOPE ::
// Vi du: Quan ly Tai khoan Ngan hang
// ============================================================

class BankAccount {
// ==== PRIVATE: Du lieu duoc bao ve ====
private:
    string ownerName;
    double balance;
    int accountNumber;
    static int totalAccounts;   // Static: dem tong so TK
    static int nextAccountNum;  // Static: so TK tiep theo

    // Private helper - logic noi bo, khong lo ra ngoai
    bool isValidAmount(double amount) const {
        return amount > 0;
    }

// ==== PUBLIC: Giao dien cong khai ====
public:
    // --- Constructor (se hoc sau, dung nhu mac nhien) ---
    BankAccount(string name, double initialBalance) {
        // Dung this-> vi tham so trung ten voi member
        this->ownerName = name;
        this->balance = (initialBalance >= 0) ? initialBalance : 0;
        this->accountNumber = ++nextAccountNum;
        totalAccounts++;
        cout << "  [+] Tao TK #" << accountNumber
             << " | Chu: " << ownerName
             << " | So du ban dau: " << balance << endl;
    }

    // --- Ham dinh nghia TRONG lop (auto inline) ---
    string getOwner() const { return ownerName; }
    double getBalance() const { return balance; }
    int getAccountNumber() const { return accountNumber; }

    // --- Nap tien ---
    void deposit(double amount) {
        if (isValidAmount(amount)) {  // Goi private helper
            balance += amount;
            cout << "  Nap " << amount
                 << " | So du moi: " << balance << endl;
        } else {
            cout << "  [!] So tien nap khong hop le!" << endl;
        }
    }

    // --- Rut tien - khai bao trong, dinh nghia ngoai ---
    bool withdraw(double amount);

    // --- METHOD CHAINING voi this ---
    // Moi ham tra ve *this de goi tiep
    BankAccount& setOwner(string name) {
        this->ownerName = name;
        return *this;   // Tra ve chinh object nay
    }

    BankAccount& addBonus(double bonus) {
        if (bonus > 0) balance += bonus;
        return *this;
    }

    // --- Ham hien thi ---
    void printInfo() const {
        cout << "  +------------------------------------------+" << endl;
        cout << "  | TK #" << accountNumber
             << " | Chu: " << ownerName
             << " | So du: " << balance << endl;
        cout << "  +------------------------------------------+" << endl;
    }

    // --- Static methods ---
    static int getTotalAccounts() { return totalAccounts; }

    // --- Destructor ---
    ~BankAccount() {
        totalAccounts--;
        cout << "  [-] Dong TK #" << accountNumber
             << " cua " << ownerName << endl;
    }
};

// ==== Dinh nghia bien STATIC ngoai class (BAT BUOC!) ====
int BankAccount::totalAccounts = 0;
int BankAccount::nextAccountNum = 1000;

// ==== Dinh nghia ham withdraw NGOAI class (scope ::) ====
bool BankAccount::withdraw(double amount) {
    if (!isValidAmount(amount)) {
        cout << "  [!] So tien rut khong hop le!" << endl;
        return false;
    }
    if (amount > balance) {
        cout << "  [!] Khong du so du! Can " << amount
             << " nhung chi co " << balance << endl;
        return false;
    }
    balance -= amount;
    cout << "  Rut " << amount << " | So du moi: " << balance << endl;
    return true;
}

int main() {
    cout << "======================================" << endl;
    cout << "  DEMO CLASS & OBJECT - NGAN HANG  " << endl;
    cout << "======================================" << endl;

    // [1] Tao doi tuong tren STACK
    cout << "
[1] Tao doi tuong tren Stack:" << endl;
    BankAccount acc1("Nguyen Van A", 5000000.0);
    BankAccount acc2("Tran Thi B", 3000000.0);
    cout << "  Tong so TK hien tai: " << BankAccount::getTotalAccounts() << endl;

    // [2] Truy cap qua toan tu .
    cout << "
[2] Truy cap bang toan tu '.':" << endl;
    cout << "  Chu: " << acc1.getOwner() << endl;
    cout << "  So du: " << acc1.getBalance() << endl;
    acc1.deposit(1000000);
    acc1.withdraw(500000);
    acc1.withdraw(9000000);  // Thu rut qua so du

    // [3] Tao doi tuong tren HEAP (con tro)
    cout << "
[3] Doi tuong tren Heap (con tro + ->):" << endl;
    BankAccount* pAcc = new BankAccount("Le Van C", 2000000.0);
    pAcc->deposit(500000);    // Truy cap qua -> (khac voi . o tren)
    pAcc->printInfo();
    delete pAcc;               // BAT BUOC giai phong bo nho!
    pAcc = nullptr;            // Good practice

    // [4] Method Chaining voi this
    cout << "
[4] Method Chaining (return *this):" << endl;
    acc2.setOwner("Tran Thi B (VIP)")  // setOwner tra ve *this
        .addBonus(500000)               // addBonus tra ve *this
        .addBonus(200000);              // Goi tiep luon!
    acc2.printInfo();

    cout << "
--- Ket thuc main --- (acc1, acc2 tu dong bi huy)" << endl;
    return 0;
}
````

## III.1 & III.2. Data Members và Member Functions

### 📦 Data Members (Biến thành viên)

Data members là **biến khai báo bên trong class**, lưu trạng thái của object.

```cpp
class Car {
    string brand;      // Non-static: MỖI object có bản copy riêng
    double speed;
    bool isRunning;
    static int count;  // Static: TẤT CẢ objects dùng CHUNG một bản
};
```

**Phân loại:**
- **Non-static**: Mỗi object có một bản riêng (thay đổi ở object A không ảnh hưởng object B)
- **Static**: Tất cả objects chia sẻ, chỉ có một bản duy nhất trong bộ nhớ

### ⚙️ Member Functions (Hàm thành viên)

Hàm thành viên định nghĩa **hành vi** của object. Đặc điểm quan trọng: có thể truy cập trực tiếp **tất cả** data members (kể cả private) của chính class đó.

---

## III.3 & III.4. Định nghĩa hàm: Trong lớp vs Ngoài lớp

| | Trong lớp (inline) | Ngoài lớp (với `::`) |
|-|-------------------|---------------------|
| **Vị trí** | Trong dấu `{}` của class | Bên ngoài class, dùng `TenLop::` |
| **Inline** | Compiler tự động cân nhắc inline | Không inline (trừ khi ghi `inline`) |
| **Phù hợp với** | Hàm ngắn (getter/setter, 1-3 dòng) | Hàm dài/phức tạp, tách ra file `.cpp` |
| **Dự án thực tế** | Viết trong file `.h` | Viết trong file `.cpp` |

---

## III.5. Hàm hằng – `const` Member Function

```cpp
class Rectangle {
    int w, h;
public:
    int area() const;   // const đặt SAU dấu )
};
```

**`const` sau `)` nghĩa là gì chính xác?**
- Bên trong hàm, `this` có kiểu `const Rectangle*` — **pointer to const**
- Hàm **cam kết không thay đổi** bất kỳ data member nào
- **Object hằng** (`const Rectangle r`) chỉ được gọi `const` member functions

**Quy tắc thực hành (bắt buộc với dev có kinh nghiệm):**
> Bất kỳ member function nào không cần sửa dữ liệu → **luôn** đánh dấu `const`.

**Mutable — ngoại lệ của const:**
```cpp
class Logger {
    mutable int callCount = 0;  // mutable: cho phép sửa ngay cả trong const function
public:
    void log(string msg) const {
        callCount++;  // OK vì callCount là mutable
    }
};
```

## III.6. Hàm nội tuyến – `inline` Function

**Cách hoạt động:**
- Hàm thường: CPU nhảy đến địa chỉ hàm → thực thi → nhảy về (overhead)
- `inline`: Compiler **chèn code** vào ngay vị trí gọi → không có overhead nhảy

```cpp
inline int max(int a, int b) { return a > b ? a : b; }

// Khi gọi: int m = max(3, 5);
// Compiler biến thành: int m = (3 > 5) ? 3 : 5;  // Không có function call!
```

**Khi nào nên dùng inline?**
- Hàm **rất ngắn** (≤ 3 dòng)
- Hàm được gọi **rất nhiều lần** trong vòng lặp
- Getter/Setter là ứng dụng cổ điển nhất

> ⚠️ `inline` chỉ là **gợi ý** cho compiler — compiler quyết định có inline hay không. Hàm quá phức tạp sẽ bị bỏ qua.
>
> 💡 Hàm định nghĩa **bên trong class** tự động được compiler coi là inline.

````python
%%writefile methods_demo.cpp
#include <iostream>
#include <string>
#include <stdexcept>
using namespace std;

// ============================================================
// DEMO: DATA MEMBERS, MEMBER FUNCTIONS, CONST, GETTER/SETTER
// Vi du: Quan ly san pham trong kho hang
// ============================================================

class Product {
private:
    // Data members
    string name;
    double price;
    int quantity;
    string category;
    mutable int viewCount;  // mutable: co the thay doi ngay ca trong ham const

    // Private helper - kiem tra gia hop le
    bool isValidPrice(double p) const { return p >= 0; }
    bool isValidQuantity(int q) const { return q >= 0; }

public:
    // === CONSTRUCTOR ===
    Product(string n, double p, int q, string cat)
        : name(n), price(p), quantity(q), category(cat), viewCount(0) {}

    // === GETTER (const - khong thay doi du lieu) ===

    // Ham dinh nghia TRONG lop (auto inline)
    string getName() const { return name; }
    string getCategory() const { return category; }

    // Ham 'const' nhung van tang viewCount nho 'mutable'
    double getPrice() const {
        viewCount++;  // OK vi viewCount la mutable!
        return price;
    }

    int getQuantity() const { return quantity; }

    // Getter tinh toan
    double getTotalValue() const {
        return price * quantity;  // Tinh gia tri kho hang
    }

    bool isAvailable() const { return quantity > 0; }
    bool isExpensive() const { return price > 1000000; }

    int getViewCount() const { return viewCount; }

    // === SETTER (co validation) ===

    bool setPrice(double p) {
        if (!isValidPrice(p)) {
            cout << "  [!] Gia khong hop le: " << p << endl;
            return false;
        }
        price = p;
        cout << "  Cap nhat gia: " << price << endl;
        return true;
    }

    bool setQuantity(int q) {
        if (!isValidQuantity(q)) {
            cout << "  [!] So luong khong hop le: " << q << endl;
            return false;
        }
        quantity = q;
        return true;
    }

    // === MEMBER FUNCTIONS dinh nghia NGOAI lop ===
    // (Khai bao o day, dinh nghia sau class)
    bool restock(int amount);
    bool sell(int amount);
    void printInfo() const;
    void applyDiscount(double percent);
};

// === Dinh nghia NGOAI lop bang scope resolution :: ===

bool Product::restock(int amount) {
    if (amount <= 0) {
        cout << "  [!] So luong nhap kho phai > 0" << endl;
        return false;
    }
    quantity += amount;
    cout << "  Nhap kho: +" << amount << " | Ton kho moi: " << quantity << endl;
    return true;
}

bool Product::sell(int amount) {
    if (amount <= 0) {
        cout << "  [!] So luong ban phai > 0" << endl;
        return false;
    }
    if (amount > quantity) {
        cout << "  [!] Khong du hang! Can " << amount
             << " nhung chi con " << quantity << endl;
        return false;
    }
    quantity -= amount;
    cout << "  Ban ra: " << amount << " | Con lai: " << quantity << endl;
    return true;
}

void Product::printInfo() const {
    // Const function: cam ket khong thay doi bat ky thu gi (ngoai mutable)
    cout << "  +----------------------------------+" << endl;
    cout << "  | San pham: " << name << endl;
    cout << "  | Gia: " << price << " VND" << endl;
    cout << "  | Ton kho: " << quantity << " cai" << endl;
    cout << "  | Danh muc: " << category << endl;
    cout << "  | Gia tri kho: " << getTotalValue() << " VND" << endl;
    cout << "  | Trang thai: " << (isAvailable() ? "Con hang" : "Het hang") << endl;
    cout << "  +----------------------------------+" << endl;
}

void Product::applyDiscount(double percent) {
    if (percent <= 0 || percent >= 100) {
        cout << "  [!] % giam gia khong hop le" << endl;
        return;
    }
    double newPrice = price * (1.0 - percent / 100.0);
    cout << "  Giam " << percent << "%: " << price << " -> " << newPrice << endl;
    price = newPrice;
}

// ============================================================
// Dem so lan truy cap -> ung dung cua mutable
// ============================================================
void checkPrices(const Product& p1, const Product& p2, const Product& p3) {
    // Goi getPrice() tren const object - OK vi getPrice() la const
    cout << "
  Gia: " << p1.getName() << "=" << p1.getPrice()
         << ", " << p2.getName() << "=" << p2.getPrice()
         << ", " << p3.getName() << "=" << p3.getPrice() << endl;
}

int main() {
    cout << "======================================" << endl;
    cout << "     DEMO THUOC TINH & PHUONG THUC    " << endl;
    cout << "======================================" << endl;

    Product phone("iPhone 15", 25000000.0, 50, "Dien tu");
    Product shirt("Ao thun", 150000.0, 200, "Thoi trang");
    Product laptop("MacBook Pro", 45000000.0, 10, "Dien tu");

    cout << "
[1] Getter va in thong tin:" << endl;
    phone.printInfo();

    cout << "
[2] Setter co validation:" << endl;
    phone.setPrice(23000000);  // Hop le
    phone.setPrice(-1000);     // Khong hop le

    cout << "
[3] Business logic - Ban va nhap kho:" << endl;
    phone.sell(10);
    phone.sell(100);  // Qua so luong
    phone.restock(20);
    phone.sell(5);
    cout << "  Ton kho sau giao dich: " << phone.getQuantity() << endl;

    cout << "
[4] Ap dung giam gia:" << endl;
    shirt.applyDiscount(20);  // Giam 20%
    shirt.printInfo();

    cout << "
[5] Mutable viewCount - theo doi luot xem:" << endl;
    const Product constPhone("Samsung S24", 20000000.0, 30, "Dien tu");
    // constPhone la const object - chi goi duoc const functions
    checkPrices(phone, shirt, laptop);
    cout << "  Luot xem phone: " << phone.getViewCount() << endl;
    cout << "  Luot xem shirt: " << shirt.getViewCount() << endl;

    cout << "
[6] Kiem tra trang thai bang getter:" << endl;
    cout << "  " << phone.getName() << " dat tien? "
         << (phone.isExpensive() ? "Co" : "Khong") << endl;
    cout << "  " << shirt.getName() << " con hang? "
         << (shirt.isAvailable() ? "Co" : "Khong") << endl;

    return 0;
}
````

# PHẦN IV: MỨC TRUY CẬP (ACCESS SPECIFIERS)

## IV.1, IV.2, IV.3. `private`, `public`, `protected`

### 🗝️ Bảng quyền truy cập

| Từ khóa | Trong class | Class con (Inheritance) | Bên ngoài class |
|---------|:-----------:|:-----------------------:|:---------------:|
| `private` | ✅ | ❌ | ❌ |
| `protected` | ✅ | ✅ | ❌ |
| `public` | ✅ | ✅ | ✅ |

### 📌 Mặc định

```cpp
class MyClass {
    int x;   // ← Mặc định là PRIVATE (class)
};

struct MyStruct {
    int x;   // ← Mặc định là PUBLIC (struct)
};
```

> 💡 **Khác biệt duy nhất** giữa `class` và `struct` trong C++ là mức truy cập mặc định.

---

## IV.4. Quy tắc Truy cập Dữ liệu

**Quy tắc vàng (Good Practice):**

```
private   → Tất cả data members
           → Helper functions (chỉ dùng nội bộ)

public    → Constructor / Destructor
           → Getter / Setter (với validation)
           → Các hàm nghiệp vụ (business logic) cần gọi từ ngoài

protected → Data / Function muốn lớp con dùng được
           → Khi thiết kế hệ thống kế thừa
```

**Tại sao không để tất cả là `public`?**
- Không kiểm soát được ai thay đổi dữ liệu
- Không thể validate dữ liệu đầu vào
- Khó thay đổi cách lưu trữ nội bộ sau này
- Vi phạm nguyên lý Encapsulation

---

## IV.5. Thiết kế Đóng gói Dữ liệu

**Nguyên tắc "Minimum Exposure" — chỉ lộ ra những gì thực sự cần:**
1. Bắt đầu với `private` cho tất cả
2. Chỉ nâng lên `public`/`protected` khi thực sự cần
3. Dùng `protected` khi thiết kế hệ thống kế thừa

````python
%%writefile access_demo.cpp
#include <iostream>
#include <string>
using namespace std;

// ============================================================
// DEMO MUC TRUY CAP: private / protected / public
// He thong phan cap nhan vien
// ============================================================

class Person {
// === PRIVATE: TUYET DOI KHONG CHO NGOAI TRUY CAP ===
private:
    string nationalId;   // CCCD - tuyet mat, du class con cung khong biet
    string password;

    // Private helper function - chi Person biet
    bool isValidAge(int a) const { return a >= 0 && a <= 150; }
    bool isValidName(const string& n) const { return n.length() >= 2; }

// === PROTECTED: Lop con ke thua co the dung, ngoai khong duoc ===
protected:
    string name;
    int age;
    string email;

    // Protected method: chi lop con co the goi
    void updateEmail(const string& e) { email = e; }

// === PUBLIC: Mo hoang - ai cung truy cap duoc ===
public:
    Person(string n, int a, string id)
        : name(n), age(a), nationalId(id), password("default") {
        email = n + "@example.com";
        cout << "  [Person] Tao: " << n << endl;
    }

    // Public getters
    string getName() const { return name; }
    int getAge() const { return age; }
    string getEmail() const { return email; }

    // Public setters co validation
    bool setAge(int a) {
        if (!isValidAge(a)) {
            cout << "  [!] Tuoi " << a << " khong hop le!" << endl;
            return false;
        }
        age = a;
        return true;
    }

    virtual void introduce() const {
        cout << "  Xin chao! Toi la " << name
             << ", " << age << " tuoi." << endl;
    }

    virtual ~Person() {}
};

// ============================================================
// Lop con - ke thua PROTECTED members cua Person
// ============================================================
class Employee : public Person {
private:
    double salary;
    string department;

public:
    Employee(string n, int a, string id, string dept, double sal)
        : Person(n, a, id) {   // Goi constructor lop cha
        department = dept;
        salary = sal;

        // CO THE truy cap protected members cua Person
        this->name = n;               // protected → OK
        updateEmail(n + "@company.com"); // protected method → OK

        // KHONG THE truy cap private members
        // this->nationalId = "...";  // LOI - private!
        // this->password = "...";    // LOI - private!

        cout << "  [Employee] Tuyen: " << name
             << " | Phong: " << dept << endl;
    }

    void introduce() const override {
        // Truy cap protected 'name', 'age' tu lop cha
        cout << "  Toi la " << name  // protected → OK
             << ", " << age          // protected → OK
             << " tuoi, nhan vien phong " << department << endl;
    }

    double getSalary() const { return salary; }
    string getDept() const { return department; }

    void getSalaryInfo() const {
        cout << "  Luong: " << salary
             << " | Email: " << email  // protected → OK
             << endl;
    }
};

class Manager : public Employee {
private:
    int teamSize;

public:
    Manager(string n, int a, string id, int team)
        : Employee(n, a, id, "Management", 30000000.0) {
        teamSize = team;
        // name la protected (ke thua qua Employee) → van truy cap duoc
        cout << "  [Manager] Tao: " << name << endl;
    }

    void introduce() const override {
        cout << "  Toi la quan ly " << name
             << ", quan ly nhom " << teamSize << " nguoi." << endl;
    }
};

int main() {
    cout << "======================================" << endl;
    cout << "     DEMO MUC TRUY CAP OOP     " << endl;
    cout << "======================================" << endl;

    cout << "
[1] TAO OBJECTS:" << endl;
    Person p1("Nguyen Van A", 30, "001234567890");
    Employee e1("Tran Thi B", 25, "009876543210", "Engineering", 20000000);
    Manager m1("Le Van C", 40, "001111111111", 5);

    cout << "
[2] PUBLIC truy cap tu moi noi:" << endl;
    cout << "  p1.getName(): " << p1.getName() << endl;    // OK - public
    cout << "  e1.getAge(): "  << e1.getAge()  << endl;    // OK - public (ke thua)
    cout << "  m1.getDept(): " << m1.getDept() << endl;    // OK - public

    cout << "
[3] PRIVATE khong the truy cap tu ngoai:" << endl;
    // p1.nationalId   → LOI BIEN DICH
    // p1.password     → LOI BIEN DICH
    // p1.isValidAge() → LOI BIEN DICH
    cout << "  (p1.nationalId, p1.password → LOI BIEN DICH)" << endl;
    cout << "  Ly do: private khong lo ra ngoai class Person" << endl;

    cout << "
[4] PROTECTED - lop con dung duoc, ngoai khong:" << endl;
    // p1.name = "Hack";  // LOI - protected, ngoai class khong duoc!
    // p1.email = "...";  // LOI - protected!
    cout << "  (p1.name tu ngoai → LOI BIEN DICH)" << endl;
    cout << "  Nhung Employee/Manager co the dung name, email" << endl;

    cout << "
[5] POLYMORPHISM - introduce() theo tung lop:" << endl;
    p1.introduce();   // Person::introduce
    e1.introduce();   // Employee::introduce (override)
    m1.introduce();   // Manager::introduce (override)

    cout << "
[6] SETTER co validation:" << endl;
    p1.setAge(35);    // OK
    p1.setAge(-10);   // Bi tu choi
    p1.setAge(200);   // Bi tu choi

    cout << "
[7] CLASS vs STRUCT (mac dinh truy cap):" << endl;
    struct Point { int x, y; };     // Mac dinh PUBLIC
    class SecretPoint { int x, y; }; // Mac dinh PRIVATE

    Point pt; pt.x = 5; pt.y = 3;  // OK - public
    // SecretPoint sp; sp.x = 5;   // LOI - private!
    cout << "  struct Point{int x,y}: truy cap truc tiep OK" << endl;
    cout << "  class SecretPoint{int x,y}: truy cap truc tiep LOI" << endl;

    return 0;
}
````

# PHẦN V: CONSTRUCTOR VÀ DESTRUCTOR

## V.1. Constructor là gì?

**Constructor** (Hàm tạo) là **member function đặc biệt** được gọi **tự động** ngay khi object được tạo.

**Đặc điểm cần nhớ:**
- **Tên = Tên class** (phân biệt hoa/thường)
- **Không có kiểu trả về** (kể cả `void`)
- Có thể **quá tải** (overload) — nhiều constructor với tham số khác nhau
- Được gọi **ngay lập tức** khi tạo object — không cần gọi thủ công

```cpp
class Car {
public:
    Car() { /* gọi tự động khi: Car c; */ }
    Car(string color) { /* gọi tự động khi: Car c("red"); */ }
    // KHÔNG CÓ: void Car() {} hay Car Car() {} — sai cú pháp!
};
```

## V.2. Default Constructor

Constructor **không tham số** (hoặc tất cả tham số có giá trị mặc định).

```cpp
class Point {
```

- Bạn không khai báo constructor nào → compiler tạo default constructor rỗng

## V.5. Copy Constructor — Sao chép Object

**Cú pháp:** `ClassName(const ClassName& other)`

**Khi nào tự động được gọi?**
1. `B b = a;` (copy initialization)
2. `B b(a);` (direct initialization)  
3. Truyền object vào hàm **theo giá trị** (pass by value)
4. Hàm trả về object **theo giá trị** (return by value)

### Shallow Copy vs Deep Copy

```
Shallow Copy (compiler tạo):   Deep Copy (tự viết):
┌──────────┐                    ┌──────────┐
│ a.data ──┼──→ [1,2,3] Heap   │ a.data ──┼──→ [1,2,3] Heap
└──────────┘                    └──────────┘
     ↓ copy                          ↓ deep copy
┌──────────┐                    ┌──────────┐
│ b.data ──┼──→ CÙNG ô nhớ!!   │ b.data ──┼──→ [1,2,3] Heap MỚI
└──────────┘  ⚠️ NGUY HIỂM      └──────────┘  ✅ ĐỘC LẬP
```

> **Shallow copy**: Khi hủy cả a và b → `delete` cùng một vùng nhớ 2 lần → **Undefined Behavior/Crash!**

---

## V.6. Move Constructor (C++11) — "Trộm" Tài nguyên

**Cú pháp:** `ClassName(ClassName&& other) noexcept`

**Ý tưởng**: Thay vì sao chép dữ liệu (tốn thời gian), "**chiếm quyền sở hữu**" tài nguyên từ object tạm:

```cpp
// Copy constructor: SAO CHÉP → tốn O(n)
MyString a("hello");
MyString b = a;          // Cấp phát bộ nhớ mới, copy từng byte

// Move constructor: LẤY CON TRỎ → O(1)!
MyString c = std::move(a);  // c.data = a.data (lấy pointer)
                             // a.data = nullptr  (a rỗng, nhưng hợp lệ)
```

## V.7. Initializer List — Danh sách Khởi tạo

**Cú pháp:** Đặt sau `:` trước `{}` của constructor

```cpp
class Point {
    int x, y;
    const int ID;  // ← PHẢI dùng initializer list cho const
public:
    Point(int x, int y, int id)
        : x(x), y(y), ID(id)   // ← Initializer list
    {
        // body
    }
};
```

**BẮT BUỘC dùng Initializer List khi:**
1. Data member là `const`
2. Data member là `reference (&)`
3. Gọi constructor của base class
4. Data member không có default constructor

**Initializer List luôn tốt hơn gán trong body:**
```cpp
// ❌ Kém hiệu quả: Khởi tạo mặc định rồi gán lại
Point(int x) { this->x = x; }  // 2 bước: default init + assignment

// ✅ Tốt hơn: Khởi tạo trực tiếp ngay từ đầu
Point(int x) : x(x) {}          // 1 bước: direct initialization
```

> ⚠️ **Thứ tự khởi tạo**: Theo thứ tự **khai báo trong class**, không theo thứ tự trong initializer list!

```cpp
class MyClass {
public:
    ~MyClass() {    // Tên = ~ + tên class, không có tham số, không trả về
        delete[] data;    // Giải phóng heap memory

        // Đóng socket, database connection, v.v.
    }

```

**Khi nào destructor được gọi?**
- **Stack object**: Khi ra khỏi scope `{}`
- **Heap object**: Khi gọi `delete`
- **Thành viên của class**: Khi class chứa nó bị hủy

## V.9. Thứ tự Gọi Constructor & Destructor

```
Kế thừa:          Base() → Derived()     (cha trước, con sau)
Hủy:              ~Derived() → ~Base()   (ngược lại — LIFO)

Cùng scope:       a() → b() → c()
Hủy:              ~c() → ~b() → ~a()    (LIFO - Last In, First Out)
```

````python
%%writefile constructors_full.cpp
#include <iostream>
#include <string>
#include <utility>  // std::move
using namespace std;

// ============================================================
// CLASS MyArray - minh hoa tat ca loai constructor
// Quan ly mang dong tren Heap
// ============================================================

int instanceCounter = 0;

class MyArray {
private:
    int* data;      // Con tro toi vung nho heap
    int size;
    int id;

    void log(const string& msg) const {
        cout << "  [ID=" << id << " | size=" << size
             << " | ptr=" << (data ? "valid" : "null")
             << "] " << msg << endl;
    }

public:
    // ============================================================
    // V.2 - DEFAULT CONSTRUCTOR
    // ============================================================
    MyArray() : data(nullptr), size(0), id(++instanceCounter) {
        log("Default Constructor (array rong)");
    }

    // ============================================================
    // V.3 - CONSTRUCTOR CO THAM SO + INITIALIZER LIST
    // ============================================================
    MyArray(int n, int defaultVal = 0)  // Overload: co the truyen 1 hoac 2 tham so
        : size(n > 0 ? n : 0),           // V.7: Initializer List
          id(++instanceCounter),
          data(n > 0 ? new int[n] : nullptr)  // Cap phat heap
    {
        // Body constructor: dien gia tri
        for (int i = 0; i < size; i++) data[i] = defaultVal;
        log("Param Constructor (cap phat " + to_string(size) + " int)");
    }

    // ============================================================
    // V.4 - CONSTRUCTOR OVERLOADING (constructor thu 3)
    // ============================================================
    MyArray(int n, int start, int step)  // Tao day so tang dan
        : size(n > 0 ? n : 0),
          id(++instanceCounter),
          data(n > 0 ? new int[n] : nullptr)
    {
        for (int i = 0; i < size; i++) data[i] = start + i * step;
        log("Overloaded Constructor (day tang dan)");
    }

    // ============================================================
    // V.5 - COPY CONSTRUCTOR (DEEP COPY - Sao chep sau)
    // ============================================================
    MyArray(const MyArray& other)
        : size(other.size),
          id(++instanceCounter),
          data(other.size > 0 ? new int[other.size] : nullptr)  // HEAP MOI!
    {
        // Copy tung phan tu - KHONG chia se cung dia chi vung nho
        for (int i = 0; i < size; i++) data[i] = other.data[i];
        log("Copy Constructor - DEEP copy (tao heap moi doc lap)");
    }

    // ============================================================
    // V.6 - MOVE CONSTRUCTOR (C++11) - "Trom" tai nguyen
    // ============================================================
    MyArray(MyArray&& other) noexcept
        : size(other.size),
          id(++instanceCounter),
          data(other.data)    // LAY pointer - KHONG cap phat moi!
    {
        // Set 'other' ve trang thai trong rong (nhung hop le)
        other.data = nullptr;
        other.size = 0;
        log("Move Constructor - LAY tai nguyen tu other (O(1)!)");
    }

    // ============================================================
    // V.8 - DESTRUCTOR
    // ============================================================
    ~MyArray() {
        log("Destructor - giai phong heap");
        delete[] data;  // Giai phong vung nho da cap phat
        data = nullptr;
    }

    // Helper methods
    void print() const {
        cout << "  [" << id << "] [";
        for (int i = 0; i < size; i++) {
            cout << data[i];
            if (i < size - 1) cout << ", ";
        }
        cout << "] (size=" << size << ")" << endl;
    }

    int getSize() const { return size; }
    int getId() const { return id; }
    bool isEmpty() const { return size == 0 || data == nullptr; }
};

// ============================================================
// Demo Order of Construction/Destruction
// ============================================================
class Base {
public:
    Base() { cout << "  Base() - Constructor lop cha" << endl; }
    virtual ~Base() { cout << "  ~Base() - Destructor lop cha" << endl; }
};

class Derived : public Base {
public:
    Derived() { cout << "  Derived() - Constructor lop con" << endl; }
    ~Derived() { cout << "  ~Derived() - Destructor lop con" << endl; }
};

// Ham nhan theo gia tri - copy constructor duoc goi!
void processArray(MyArray arr) {
    cout << "    Trong ham processArray: ";
    arr.print();
}  // Destructor cua 'arr' (ban sao) duoc goi khi ra khoi ham

// ============================================================
int main() {
    cout << "============================================" << endl;
    cout << "  DEMO TOAN DIEN CONSTRUCTOR & DESTRUCTOR  " << endl;
    cout << "============================================" << endl;

    // V.2 - Default
    cout << "
[1] DEFAULT CONSTRUCTOR:" << endl;
    MyArray a1;  // Goi default constructor

    // V.3 - Co tham so + Initializer List
    cout << "
[2] PARAM CONSTRUCTOR + INITIALIZER LIST:" << endl;
    MyArray a2(5, 10);    // 5 phan tu, gia tri 10
    MyArray a3(3);        // 3 phan tu, mac dinh = 0
    a2.print();

    // V.4 - Overloaded constructor
    cout << "
[3] OVERLOADED CONSTRUCTOR:" << endl;
    MyArray a4(5, 1, 2);  // [1, 3, 5, 7, 9]
    a4.print();

    // V.5 - Copy Constructor (Deep Copy)
    cout << "
[4] COPY CONSTRUCTOR (Deep Copy):" << endl;
    MyArray a5 = a4;      // Copy constructor
    MyArray a6(a4);       // Cach viet khac, cung la copy constructor
    cout << "  a4 goc: "; a4.print();
    cout << "  a5 sau copy: "; a5.print();
    cout << "  a5 va a4 doc lap nhau (khac heap)!" << endl;

    // V.6 - Move Constructor
    cout << "
[5] MOVE CONSTRUCTOR:" << endl;
    MyArray source(4, 5, 3);  // source = [5, 8, 11, 14]
    cout << "  Truoc khi move: "; source.print();
    MyArray moved = std::move(source);  // Move - lay tai nguyen
    cout << "  Sau khi move:" << endl;
    cout << "    moved: "; moved.print();
    cout << "    source (bi lay het): isEmpty=" << source.isEmpty() << endl;

    // Copy khi truyen vao ham (copy constructor duoc goi)
    cout << "
[6] COPY KHI TRUYEN VAO HAM (by value):" << endl;
    cout << "  Truoc khi goi ham: "; a4.print();
    processArray(a4);  // Copy constructor -> processArray -> destructor ban sao
    cout << "  Sau khi goi ham: "; a4.print();  // a4 van nguyen!

    // V.9 - Thu tu Constructor/Destructor
    cout << "
[7] THU TU CONSTRUCTOR & DESTRUCTOR (ke thua):" << endl;
    cout << "  Tao doi tuong Derived:" << endl;
    {
        Derived d;  // Base() truoc, Derived() sau
    }  // ~Derived() truoc, ~Base() sau (LIFO)

    cout << "
[8] THU TU TRONG CUNG SCOPE:" << endl;
    {
        cout << "  Bat dau scope:" << endl;
        MyArray x(2, 1);  // Constructor x
        MyArray y(2, 2);  // Constructor y
        cout << "  -> Trong scope: x va y ton tai" << endl;
    }  // ~y truoc, ~x sau (LIFO - Last In First Out)

    cout << "
--- Ket thuc main ---" << endl;
    return 0;
}  // a1..a6, moved bi huy theo thu tu nguoc lai
````

### 🎯 Tóm tắt: Khi nào dùng loại Constructor nào?

| Constructor | Mục đích | Ghi chú |
|-------------|----------|---------|
| **Default** `C()` | Tạo object "trống", khởi tạo về 0/null | Compiler tự tạo nếu không khai báo gì |
| **Param** `C(args)` | Tạo object với dữ liệu cụ thể | Thường dùng nhất trong thực tế |
| **Copy** `C(const C&)` | Tạo bản sao độc lập | Tự viết khi class có con trỏ (deep copy!) |
| **Move** `C(C&&)` | Chuyển tài nguyên, không copy | C++11, dùng `std::move()`, O(1) |
| **Initializer List** | Luôn ưu tiên dùng | Bắt buộc với `const`/`reference`/base class |


# PHẦN VI: THÀNH VIÊN TĨNH (STATIC MEMBERS)

## VI.1. Biến thành viên tĩnh (static data member)

### 🔑 Khái niệm cốt lõi

| | Non-static | Static |
|-|-----------|--------|
| **Bộ nhớ** | Mỗi object có bản riêng | Một bản duy nhất cho tất cả |
| **Thuộc về** | Từng object | Cả class |
| **Truy cập** | `obj.member` | `ClassName::member` |
| **Tồn tại khi** | Object được tạo | Chương trình chạy (kể cả khi không có object nào) |

```
Non-static:                        Static:
┌────────┐  x=5                    ┌────────┐  x=5 ──┐
│ obj1   │                         │ obj1   │         │ count=3
├────────┤  x=3  ← Bộ nhớ riêng   ├────────┤  x=3   │ (chia sẻ)
│ obj2   │                         │ obj2   │ ────────┘
└────────┘  x=8                    └────────┘  x=8
│ obj3   │
└────────┘
```

### Cú pháp

```cpp
class MyClass {
    static int count;   // Khai báo bên trong class (không tạo biến thực!)
};
// Định nghĩa bên ngoài class (BẮT BUỘC — tạo biến thực trên memory)
int MyClass::count = 0;

// C++17: Có thể dùng inline static
class MyClass17 {
    inline static int count = 0;  // Khai báo + định nghĩa luôn trong class
};
```

> ⚠️ **Lỗi linker phổ biến**: Khai báo trong class nhưng **quên định nghĩa** bên ngoài → `undefined reference to 'MyClass::count'`

## VI.4. Ứng dụng của `static`

**1. Đếm số instance đang tồn tại:**
```cpp
static int count;
MyClass() { count++; }
~MyClass() { count--; }
```

**2. Singleton Pattern — Đảm bảo chỉ 1 instance:**
```cpp
class DB { static DB* instance; DB(){} public:
    static DB* get() { if(!instance) instance=new DB(); return instance; }
};
```

**3. Hằng số của lớp:**
```cpp
class Circle { static constexpr double PI = 3.14159265; };
```

**4. Cache / Shared resource giữa tất cả objects:**
```cpp
class DBConnection { static Connection* sharedConn; };
```

````python
%%writefile static_full.cpp
#include <iostream>
#include <string>
#include <vector>
using namespace std;

// ============================================================
// DEMO 1: STATIC DATA MEMBER + STATIC MEMBER FUNCTION
// He thong theo doi nhan vien cong ty
// ============================================================

class Employee {
private:
    string name;
    string department;
    double salary;
    int empId;

    // === STATIC DATA MEMBERS ===
    // Chi co MOT ban trong toan bo chuong trinh
    static int totalEmployees;    // Tong so nhan vien dang lam viec
    static double totalSalaryBudget; // Quy luong tong
    static int nextId;            // ID tu dong tang

public:
    // V.7 Initializer List
    Employee(string n, string dept, double sal)
        : name(n), department(dept), salary(sal),
          empId(++nextId)  // nextId la static - chia se
    {
        totalEmployees++;        // Tac dong len bien static chung
        totalSalaryBudget += salary;
        cout << "  [+] Tuyen: " << name
             << " (ID:" << empId << " | " << dept << ")" << endl;
    }

    ~Employee() {
        totalEmployees--;
        totalSalaryBudget -= salary;
        cout << "  [-] Nghi viec: " << name
             << " (ID:" << empId << ")" << endl;
    }

    // Non-static getters - thuoc ve tung object
    string getName()   const { return name; }
    string getDept()   const { return department; }
    double getSalary() const { return salary; }
    int    getId()     const { return empId; }

    bool setSalary(double newSal) {
        if (newSal <= 0) { cout << "  [!] Luong khong hop le" << endl; return false; }
        totalSalaryBudget -= salary;  // Tru luong cu
        salary = newSal;
        totalSalaryBudget += salary;  // Cong luong moi
        return true;
    }

    // === STATIC MEMBER FUNCTIONS ===
    // Khong co 'this', chi truy cap static members
    static int    getTotalEmployees()   { return totalEmployees; }
    static double getTotalBudget()      { return totalSalaryBudget; }
    static double getAverageSalary() {
        if (totalEmployees == 0) return 0.0;
        return totalSalaryBudget / totalEmployees;
    }
    static void printStats() {
        cout << "  === THONG KE CONG TY ===" << endl;
        cout << "  Tong NV    : " << totalEmployees << " nguoi" << endl;
        cout << "  Quy luong  : " << totalSalaryBudget / 1e6 << "M VND" << endl;
        cout << "  Luong TB   : " << getAverageSalary() / 1e6 << "M VND/thang" << endl;
        // Luu y: static function KHONG THE dung 'this->name' o day!
    }

    // Static factory methods (cach tao co ten ro rang)
    static Employee createIntern(string name) {
        return Employee(name, "Intern", 5000000.0);
    }
    static Employee createSeniorDev(string name) {
        return Employee(name, "Engineering", 35000000.0);
    }

    void printInfo() const {
        cout << "  ID:" << empId << " | " << name
             << " | " << department
             << " | " << salary/1e6 << "M" << endl;
    }
};

// === Dinh nghia bien static NGOAI class (BAT BUOC!) ===
int    Employee::totalEmployees    = 0;
double Employee::totalSalaryBudget = 0.0;
int    Employee::nextId            = 1000;  // ID bat dau tu 1001

// ============================================================
// DEMO 2: SINGLETON PATTERN
// Patter khi can dam bao chi co DUY NHAT 1 instance
// ============================================================
class AppLogger {
private:
    // Private constructor - ngan tao truc tiep tu ngoai
    AppLogger() {
        cout << "
  [Logger] Khoi tao Singleton (lan nay thoi!)" << endl;
    }

    vector<string> logs;
    static AppLogger* instance;  // Con tro den instance duy nhat

public:
    // Cam ket khong copy/move Logger
    AppLogger(const AppLogger&)            = delete;
    AppLogger& operator=(const AppLogger&) = delete;

    static AppLogger* getInstance() {
        if (instance == nullptr) {
            instance = new AppLogger();  // Chi tao 1 lan!
        }
        return instance;
    }

    void log(const string& msg) {
        logs.push_back(msg);
        cout << "  [LOG] " << msg << endl;
    }

    void printHistory() const {
        cout << "  [HISTORY - " << logs.size() << " entries]" << endl;
        for (size_t i = 0; i < logs.size(); i++)
            cout << "    " << (i+1) << ". " << logs[i] << endl;
    }

    static void cleanup() { delete instance; instance = nullptr; }
};
AppLogger* AppLogger::instance = nullptr;

// ============================================================
int main() {
    cout << "======================================" << endl;
    cout << "       DEMO STATIC MEMBERS           " << endl;
    cout << "======================================" << endl;

    cout << "
[0] Trang thai ban dau (chua co NV nao):" << endl;
    Employee::printStats();  // Goi qua ten lop - khong can object!

    cout << "
[1] Tuyen dung nhan vien:" << endl;
    Employee e1("Nguyen Van A", "Engineering", 25000000.0);
    Employee e2("Tran Thi B", "Marketing", 18000000.0);
    Employee e3 = Employee::createIntern("Le Van C");    // Static factory
    Employee e4 = Employee::createSeniorDev("Pham Van D");
    Employee::printStats();

    cout << "
[2] Truy cap static function qua ten lop vs qua object:" << endl;
    cout << "  Qua ten lop   : Employee::getTotalEmployees() = "
         << Employee::getTotalEmployees() << endl;
    cout << "  Qua object    : e1.getTotalEmployees()        = "
         << e1.getTotalEmployees() << "  (KO khuyen khich)" << endl;

    cout << "
[3] Tang luong mot nhan vien -> cap nhat static budget:" << endl;
    cout << "  Truoc: quy luong = " << Employee::getTotalBudget()/1e6 << "M" << endl;
    e1.setSalary(30000000.0);
    cout << "  Sau: quy luong  = " << Employee::getTotalBudget()/1e6 << "M" << endl;

    cout << "
[4] Nhan vien nghi viec (destructor giam static count):" << endl;
    {
        Employee temp = Employee::createIntern("Temp Worker");
        cout << "  Trong scope: " << Employee::getTotalEmployees() << " NV" << endl;
    }  // temp bi huy → totalEmployees--
    cout << "  Ngoai scope: " << Employee::getTotalEmployees() << " NV" << endl;

    cout << "
[5] SINGLETON PATTERN:" << endl;
    AppLogger* log1 = AppLogger::getInstance();
    AppLogger* log2 = AppLogger::getInstance();  // Tra ve CUNG instance!
    AppLogger* log3 = AppLogger::getInstance();  // Van la instance do!

    cout << "  Tat ca pointer cung dia chi? "
         << (log1 == log2 && log2 == log3 ? "YES - Singleton!" : "NO") << endl;

    log1->log("App khoi dong");
    log2->log("User dang nhap: admin");  // log2 cung la log1!
    log3->log("Database ket noi thanh cong");

    log1->printHistory();  // Thay tat ca log tu moi noi

    AppLogger::cleanup();

    cout << "
[6] Thong ke cuoi cung:" << endl;
    Employee::printStats();

    cout << "
--- Ket thuc main ---" << endl;
    return 0;
}
````

# 🏆 TỔNG KẾT TOÀN BỘ – CHEAT SHEET OOP C++

## 📋 Quy tắc vàng cần nhớ

| Quy tắc | Mô tả |
|---------|-------|
| **Rule of Zero** | Dùng smart pointer → không cần viết destructor/copy/move |
| **Rule of Three** | Có custom destructor → cần Copy Constructor + Copy Assignment |
| **Rule of Five** (C++11) | + Move Constructor + Move Assignment |
| **const correctness** | Hàm không sửa data → **luôn** đánh dấu `const` |
| **Encapsulation** | Data member → `private`, Interface → `public` |
| **Initializer List** | Dùng `:` thay vì gán trong body constructor |
| **Static definition** | Biến `static` non-const → **phải** định nghĩa ngoài class |
| **Minimum Exposure** | Chỉ public những gì thực sự cần thiết |

---

## 🔥 Skill của lập trình viên kinh nghiệm

1. **Luôn dùng Initializer List** — nhanh hơn, bắt buộc với const/reference/base class
2. **Viết Virtual Destructor** — bắt buộc nếu class là base class (tránh memory leak khi dùng polymorphism)
3. **Đánh dấu `const` cho mọi getter** — cho phép dùng với const object
4. **Dùng `= delete`** để cấm copy/move khi cần — rõ ràng hơn là để lỗi runtime
5. **`override` keyword** — tránh lỗi chính tả khi ghi đè virtual function
6. **Method chaining với `return *this`** — tăng tính đọc của code
7. **`explicit` constructor** — ngăn implicit conversion không mong muốn
8. **Tránh Most Vexing Parse** — dùng `MyClass obj{}` thay vì `MyClass obj()`
9. **Static factory methods** — đặt tên có nghĩa cho các cách tạo object khác nhau
10. **Prefer composition over inheritance** — kế thừa mạnh nhưng coupling cao

---

## ⚠️ Lỗi thường gặp nhất (TOP 10)

```cpp
// 1. Quên dấu ; sau class (lỗi phổ biến nhất của người mới)
class MyClass { }  // ← LOI! Thieu ;

// 2. Most Vexing Parse
MyClass obj();  // Khai bao HAM, khong phai object
MyClass obj{};  // ✅ Dung

// 3. Quên define static member
class A { static int x; };
// int A::x = 0;  ← Quen → linker error: undefined reference

// 4. Shallow copy voi con tro → double free crash
class B { int* p; public: B(int v){p=new int(v);}; ~B(){delete p;} };
B b1(5); B b2 = b1;  // b1.p == b2.p → 2x delete → crash!

// 5. Quen viet Virtual Destructor o base class
class Base { public: ~Base() {} };         // LOI! Khi: Base* p = new Derived(); delete p;
class BaseGood { public: virtual ~BaseGood() {} };  // ✅ Dung

// 6. Thứ tự initializer list ≠ thứ tự khai báo
class Bad { int b, a; public: Bad(int x): a(x), b(a) {} };  // b khởi tạo trước a!

// 7. Goi virtual function trong constructor
class Base2 { public: virtual void f(){cout<<"Base";}; Base2(){f();} };
class Der : public Base2 { public: void f() override {cout<<"Der";} };
Der d;  // In "Base" chứ không phải "Der"! (vtable chưa sẵn sàng trong constructor)

// 8. Memory leak - quên delete
void bad() { MyClass* p = new MyClass; /* quên delete p; */ }

// 9. Truy cap this trong constructor list (thứ tự)
class C { int a, b; public: C() : a(1), b(a+1) {} };  // OK: a khởi tạo trước b
class D { int b, a; public: D() : a(1), b(a+1) {} };  // BAD! b khởi tạo trước a

// 10. Static method gọi non-static member
class E { int x; public: static void f() { x = 5; } };  // LOI: static khong co 'this'
```

## 🗺️ Sơ đồ tổng quan Access Specifiers trong Kế thừa

```
BASE CLASS    INHERITED AS    DERIVED CLASS
─────────────────────────────────────────
private   → (bất kỳ)    → Không kế thừa được
protected → public      → protected (class con của Derived vẫn thấy)
protected → protected   → protected
protected → private     → private (bị đóng lại)
public    → public      → public
public    → protected   → protected
public    → private     → private
```

## 🧪 BÀI TẬP TỔNG HỢP — Áp dụng Kiến thức

Thực hành tốt nhất là **tự thiết kế và code**. Dưới đây là một bài tập tổng hợp nhỏ.

````python
%%writefile exercise_solution.cpp
// ============================================================
// BAI TAP TONG HOP: He thong quan ly Thu vien
// Ap dung: Class, Access Specifiers, Constructor,
//          Static Members, Getter/Setter, Inheritance
// ============================================================
#include <iostream>
#include <string>
#include <vector>
using namespace std;

// ============================================================
// Lop co so: Item (Vat pham trong thu vien)
// ============================================================
class Item {
protected:
    string title;
    string author;
    int year;
    bool isAvailable;
    int itemId;
    static int nextItemId;  // Static: ID tu dong tang

public:
    Item(string t, string a, int y)
        : title(t), author(a), year(y),
          isAvailable(true), itemId(++nextItemId) {}

    // Getters
    string getTitle()  const { return title; }
    string getAuthor() const { return author; }
    int    getYear()   const { return year; }
    int    getId()     const { return itemId; }
    bool   available() const { return isAvailable; }

    // Muon / Tra
    bool borrow() {
        if (!isAvailable) {
            cout << "  [!] '" << title << "' dang duoc muon!" << endl;
            return false;
        }
        isAvailable = false;
        cout << "  [Muon] '" << title << "' (ID:" << itemId << ")" << endl;
        return true;
    }

    bool returnItem() {
        if (isAvailable) {
            cout << "  [!] '" << title << "' chua duoc muon!" << endl;
            return false;
        }
        isAvailable = true;
        cout << "  [Tra] '" << title << "' da duoc tra" << endl;
        return true;
    }

    virtual void printInfo() const {
        cout << "  [" << itemId << "] " << title
             << " - " << author << " (" << year << ")"
             << (isAvailable ? " [San co]" : " [Dang muon]") << endl;
    }

    virtual string getType() const = 0;  // Abstract
    virtual ~Item() {}
};
int Item::nextItemId = 0;

// ============================================================
// Lop con: Book ke thua Item
// ============================================================
class Book : public Item {
private:
    string isbn;
    int pages;

public:
    Book(string t, string a, int y, string i, int p)
        : Item(t, a, y), isbn(i), pages(p) {}

    string getType() const override { return "Sach"; }

    void printInfo() const override {
        Item::printInfo();  // Goi ham cha
        cout << "     ISBN: " << isbn << " | " << pages << " trang" << endl;
    }
};

// ============================================================
// Lop con: Magazine ke thua Item
// ============================================================
class Magazine : public Item {
private:
    int issueNumber;
    string month;

public:
    Magazine(string t, string a, int y, int issue, string m)
        : Item(t, a, y), issueNumber(issue), month(m) {}

    string getType() const override { return "Tap chi"; }

    void printInfo() const override {
        Item::printInfo();
        cout << "     So: " << issueNumber << " | Thang: " << month << endl;
    }
};

// ============================================================
// Lop Library quan ly tat ca Item
// ============================================================
class Library {
private:
    string name;
    vector<Item*> items;  // Polymorphism: luu ca Book va Magazine
    static int totalBorrowCount;  // Dem tong luot muon

public:
    Library(string n) : name(n) {
        cout << "  Thu vien '" << n << "' mo cua!" << endl;
    }

    void addItem(Item* item) {
        items.push_back(item);
        cout << "  [+] Bo sung: [" << item->getType() << "] "
             << item->getTitle() << endl;
    }

    bool borrowItem(int id) {
        for (Item* item : items) {
            if (item->getId() == id) {
                if (item->borrow()) { totalBorrowCount++; return true; }
                return false;
            }
        }
        cout << "  [!] Khong tim thay ID: " << id << endl;
        return false;
    }

    bool returnItem(int id) {
        for (Item* item : items) {
            if (item->getId() == id) return item->returnItem();
        }
        return false;
    }

    void printCatalog() const {
        cout << "
  === DANH MUC: " << name << " ===" << endl;
        for (const Item* item : items) item->printInfo();  // Polymorphism!
    }

    void printAvailable() const {
        cout << "
  === VAT PHAM SAN CO ===" << endl;
        for (const Item* item : items)
            if (item->available()) item->printInfo();
    }

    static int getTotalBorrows() { return totalBorrowCount; }

    ~Library() {
        cout << "
  [Library] Dong cua. Xoa bo nho..." << endl;
        for (Item* item : items) delete item;
    }
};
int Library::totalBorrowCount = 0;

// ============================================================
int main() {
    cout << "==========================================" << endl;
    cout << "  BAI TAP: HE THONG QUAN LY THU VIEN  " << endl;
    cout << "==========================================" << endl;

    Library lib("Thu vien Trung tam");

    // Them sach va tap chi (Polymorphism: Item* chua ca 2 loai)
    lib.addItem(new Book("Lap trinh C++", "Bjarne Stroustrup", 2023, "978-0-321", 1376));
    lib.addItem(new Book("Clean Code", "Robert Martin", 2008, "978-0-132", 431));
    lib.addItem(new Magazine("PCWorld", "PCWorld Team", 2024, 145, "June"));
    lib.addItem(new Book("Design Patterns", "Gang of Four", 1994, "978-0-201", 395));
    lib.addItem(new Magazine("Forbes Vietnam", "Forbes", 2024, 89, "May"));

    // In danh muc
    lib.printCatalog();

    // Muon sach
    cout << "
--- GIAO DICH MUON/TRA ---" << endl;
    lib.borrowItem(1);  // Muon sach thu 1
    lib.borrowItem(3);  // Muon tap chi
    lib.borrowItem(1);  // Thu muon lan 2 - bi tu choi

    // Tra sach
    lib.returnItem(1);

    // In con hang
    lib.printAvailable();

    cout << "
Tong luot muon: " << Library::getTotalBorrows() << endl;

    return 0;
}  // Library destructor tu dong xoa het Book va Magazine
````
